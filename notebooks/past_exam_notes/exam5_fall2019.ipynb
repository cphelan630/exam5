**A3**  
*1.75 points*

*Candidates were expected to demonstrate knowledge of the determination and relationships 
among written, earned, and in-force exposures. Candidates were expected to determine the 
earning pattern based on quarterly written and earned exposures, which was provided.   The 
exposures were not earned evenly, and candidates struggled with determining the earning 
pattern.  Many candidates incorrectly assumed an even earning pattern and proceeded with the 
calculation on that basis.*

#### ChatGPT notes based on examiner report
For full credit on this problem, know this:

1. **The earning pattern is age-based, not calendar-quarter-based.**
   A policy written at the start of a quarter earns by **quarters since written**.

2. **Do not assume uniform earning.**
   The examiner explicitly says that was the main mistake.

3. **You only really need the first-quarter earning % to solve part a and part d.**
   From 2017 Q1:
   $
   5/100=5%
   $

4. **Part a: PY 2017 earned as of 3/31/2018**
   Best method:
   $
   (5+247.5+427.5+52.5+53.75)-125(.05)=780
   $
   Interpretation: total earned through 2018 Q1 minus the piece earned from the 2018 Q1 cohort.

5. **Part b: in-force exposures at 5/31/2018**
   Count policies still active on that date:
   $
   400+100+125+550=1175
   $
   Use **written exposures**, not earning percentages.

6. **Part c: CY 2018 unearned exposures**
   Safest method is remaining unearned % × written cohort:
   $
   550(.05)+475(.50)+30(.95)=293.5
   $
   This uses the non-uniform earning pattern and applies it to **written** exposures.

7. **Part d: CY 2019 Q1 earned exposures**
   Use **in-force exposures at 12/31/2018** as the base:
   $
   (550+475+30)(.05)=52.75
   $
   The examiner specifically warns not to use unearned exposures here.

8. **Use written cohorts for earned/unearned rollforward logic.**
   A major mistake was applying the earning pattern to **earned exposures instead of written exposures**.

9. **Know the key formulas**
   $$
   \text{EOY Unearned}=\text{BOY Unearned}+\text{Written}-\text{Earned}
   $$
   and
   $$
   \text{Unearned}=\text{remaining unearned \%}\times \text{written cohort}
   $$

10. **Show work clearly.**
    This exam report repeatedly emphasizes that graders expect enough support to follow your calculation.

Key takeaway: **Use age-based earning on written cohorts, not uniform earning and not earned-cohort shortcuts.**

* Part d: discuss if it's appropriate for this insurer to estimate earned premium for the current year by averaging the in-force premium at the end of the current year and prior year 
    * Answer 1: No, it doesn't take into account if the book is growing/shrinking. This book is not written uniformly and has little data. The estimate would be volatile.   
    * Answer 2: This is not an appropriate estimation in this example because not all policies have the same duration nor are they written uniformly throughout the year  
    * Answer 3: Given the policies are not written uniformly throughout the year, plus some policies (i.e. pol B, F) has less than a year premium, using the average of in force will not be correct.  
    * Answer 4: No, it is not appropriate for his insurer to estimate EP that way. Not all policies are annual and premium is written during different times of the year. This would not be a good/accurate indication. 

    * *Comments from Examiner Report*: 
        * saying "timing of policies", "different effective dates", "cancellation of policies" without additional detail was not enough to receive full marks. Candidates were expected to connect the idea that the more uniform their policies were written, the more accurate the estimate would be.  
        * mistake: mentioning the "stability of company" without specifying which elements of the company needed to be stable  
        * mistake: saying that the method of using in-force premium to estimate earned exposures was appropriate in the general case, without identifying any restrictions that would make the estimation method inaccurate

    * compare at different times to see if book is shrinking or expanding

### Fall 2019 Q6  

**A9**  
*2.25 points*  
*Fully qualitative*

* Non-proportional reinsurance should be addressed explicitly because amounts aren't proportional to total and will distort triangles and loss ratios  
    * Both stop loss and EOL would cause changes to the loss ratio which needs to be addressed in ratemaking  
* Trending and expenses:  
    * Variable expenses vary with premium.  
        * Since premium is usually trended, VE don't also need to be trended.  
    * Fixed expenses when using the exposure-based projection method:  
        * if exposure is inflation sensitive and expenses trend at different rate, then trending is necessary  
        * if exposure is not inflation sensitive, then trending is needed  
        * Trend can be used if fixed expense trend is different from exposure trend  
        * fixed expenses using exposure base need to be trended if the underlying exposure is not inflation sensitive  
        * *From Examiner's Report*: Candidates did not necessarily have to understand that some exposures are already inflation sensitive. An example of why or why not to trend fixed expenses relative to the type of exposure was acceptable. Stating that fixed expenses can trend at a different rate than payroll was acceptable. 
    * Fixed expenses when using premium-based projection method 
        * When using the premium-based projection method fixed expenses need to be trended independently unless fixed expenses are assumed to trend at the same rate as premium. This is an explicit assumption that needs to be made, unlike variable expenses which are defined to be a ratio to premium. 
    * Common Mistakes: 
        * referring to loss trend instead of fixed expense trend  
        * stating that ther eis no need to trend fixed expense if exposure is/isn't inflation sensitive  
        * stating that fixed expenses don't need to be trended when using premium-based projection method because fixed expense has been ratio-ed to premium
