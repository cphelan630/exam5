## Formula Sheet Quick Reference

### Calendar Year Ratio Method

$$\text{CY ULAE Ratio} = \frac{\text{Paid ULAE}_{CY}}{\text{Paid Losses + Paid ALAE}_{CY}}$$

$$\text{ULAE Provision}_{AY} = \text{Selected Ratio} \times (\text{Reported Losses + Reported ALAE})_{AY}$$

### Kittel Refinement

**Adjusted denominator** (reflects that ULAE costs are incurred partly when claims open and partly when they close):

$$\text{Adjusted Losses + ALAE} = 0.5 \times \text{Paid (Losses + ALAE)} + 0.5 \times \text{Reported (Losses + ALAE)}$$

$$\text{Kittel Ratio} = \frac{\text{Paid ULAE}_{CY}}{0.5 \times \text{Paid (L+ALAE)}_{CY} + 0.5 \times \text{Reported (L+ALAE)}_{CY}}$$

$$\text{ULAE Reserve}_{AY} = \text{Selected Kittel Ratio} \times 0.5 \times \text{IBNR (Losses + ALAE)}_{AY}$$

> **Note:** The factor of 0.5 on IBNR reflects that half of remaining ULAE will be incurred when future claims are opened/reported, and half when they are finally closed.

## ULAE Overview

### What is ULAE?

**Unallocated Loss Adjustment Expenses (ULAE)** are general claims department costs that *cannot* be traced to an individual claim. Unlike ALAE, ULAE is not recorded at the claim level — it represents the overhead of running the claims function.

**Common ULAE items:**
- Claims adjuster salaries and benefits
- Claims department management and overhead
- Claims systems, technology, and software costs
- Quality review and audit staff
- Rent and occupancy for claims offices

### ULAE vs. ALAE

| Feature | ALAE | ULAE |
|---|---|---|
| Assignable to a specific claim | Yes — tracked in the claim file | No — general overhead |
| Recorded at claim level | Yes | No |
| Examples | Defense counsel fees, court costs, expert witnesses | Adjuster salaries, claims dept overhead |
| NAIC equivalent | DCC (Defense & Cost Containment) + part of A&O | Part of A&O |
| Estimation approach | Develop with loss triangles or ratio methods | Calendar year ratio or count-based methods |
| Trend drivers | Litigation rates, attorney involvement | Staffing levels, overhead, inflation |

> **See also:** `exam5_alae_estimation.ipynb` for ALAE estimation methods.

---

### NAIC Statutory Classification

The NAIC annual statement uses a different breakdown that actuaries must also know:

| NAIC Term | Relationship to ALAE/ULAE | Description |
|---|---|---|
| **DCC** (Defense and Cost Containment) | Subset of ALAE | Costs of defending claims — outside legal counsel, cost containment services (nurse case managers, utilization review) |
| **A&O** (Adjusting and Other) | Remainder of ALAE + ULAE | Claims adjusting expenses, investigation costs, and all other LAE not in DCC |

The mapping: **ALAE** splits into DCC (defense portion) + part of A&O (adjusting portion). **ULAE** flows entirely into A&O.

> **Exam tip** — Know both classification systems. ALAE/ULAE is the actuarial framework (used for development and trending). DCC/A&O is the statutory financial reporting framework. A common exam trap: confusing which classification belongs to which framework.

## Method 1: Calendar Year Paid Development Method

The simplest and most common ULAE estimation approach. Because ULAE cannot be assigned to individual claims, a ratio of ULAE to losses + ALAE is computed on a calendar year basis and then applied to each accident year's reserve base.

### Step-by-Step Procedure

1. Gather **calendar year paid ULAE** for several recent years (from statutory Schedule P or internal records)
2. Gather **calendar year paid losses + paid ALAE** for the same calendar years
3. Compute the annual ratio: ULAE / (Paid Losses + Paid ALAE)
4. Select a ratio based on observed trend and actuarial judgment
5. Apply the selected ratio to each accident year's **reported losses + reported ALAE** to produce the ULAE provision

### Worked Example

**Calendar year experience:**

| CY | Paid ULAE | Paid Losses + ALAE | CY Ratio |
|---|---|---|---|
| 2022 | $580,000 | $5,200,000 | 11.2% |
| 2023 | $610,000 | $5,500,000 | 11.1% |
| 2024 | $650,000 | $5,800,000 | 11.2% |
| **Selected** | | | **11.2%** |

The ratios are stable, so 11.2% is selected.

**Apply to accident year reserves:**

| AY | Reported Losses + ALAE | Selected Ratio | ULAE Provision |
|---|---|---|---|
| 2022 | $5,600,000 | 11.2% | $627,200 |
| 2023 | $5,100,000 | 11.2% | $571,200 |
| 2024 | $4,800,000 | 11.2% | $537,600 |

### Assumptions and Limitations

**Assumes:** ULAE costs are incurred proportionally throughout the life of a claim — equally at every dollar of loss paid or reserved.

**Limitation:** This is generally *not* true. In practice, significant ULAE is incurred when a claim is **opened** (initial investigation, file setup) and again when it is **closed** (final settlement, file closure). Relatively little overhead is incurred during the dormant middle period.

This means the basic ratio method can overstate or understate ULAE provisions depending on the mix of open vs. closed claims in the accident year cohort. The Kittel Refinement addresses this.

```python
import pandas as pd
import numpy as np

# Calendar year experience data
cy_data = pd.DataFrame({
    'cy': [2022, 2023, 2024],
    'paid_ulae': [580_000, 610_000, 650_000],
    'paid_losses_alae': [5_200_000, 5_500_000, 5_800_000],
})

cy_data['cy_ratio'] = cy_data['paid_ulae'] / cy_data['paid_losses_alae']

print("Calendar Year ULAE Ratios:")
display(cy_data.assign(
    paid_ulae=cy_data['paid_ulae'].map('${:,.0f}'.format),
    paid_losses_alae=cy_data['paid_losses_alae'].map('${:,.0f}'.format),
    cy_ratio=cy_data['cy_ratio'].map('{:.1%}'.format),
).rename(columns={'cy': 'CY', 'paid_ulae': 'Paid ULAE',
                   'paid_losses_alae': 'Paid Losses + ALAE', 'cy_ratio': 'CY Ratio'}))

# Select ratio (stable trend => use most recent / simple average)
selected_ratio = cy_data['cy_ratio'].iloc[-1]
print(f"\nSelected ULAE Ratio: {selected_ratio:.1%}")

# Apply to accident year reserve base
ay_data = pd.DataFrame({
    'ay': [2022, 2023, 2024],
    'reported_losses_alae': [5_600_000, 5_100_000, 4_800_000],
})

ay_data['ulae_provision'] = ay_data['reported_losses_alae'] * selected_ratio

print("\nAccident Year ULAE Provisions (Calendar Year Ratio Method):")
display(ay_data.assign(
    reported_losses_alae=ay_data['reported_losses_alae'].map('${:,.0f}'.format),
    ulae_provision=ay_data['ulae_provision'].map('${:,.0f}'.format),
).rename(columns={'ay': 'AY', 'reported_losses_alae': 'Reported Losses + ALAE',
                   'ulae_provision': 'ULAE Provision'}))

print(f"\nTotal ULAE Provision: ${ay_data['ulae_provision'].sum():,.0f}")
```

## Method 2: Kittel Refinement

The Kittel Refinement (named for G. F. Kittel) improves on the basic CY ratio method by recognizing that ULAE costs are *not* incurred uniformly — they are concentrated at two points in the claim lifecycle:

- **~50% when a claim is opened** (initial investigation, file setup, first adjuster contact)
- **~50% when a claim is closed** (final negotiation, settlement processing, file closure)

### Why This Matters

The basic ratio method uses **paid losses + ALAE** as the denominator, which acts as a proxy for claims that are fully or partially closed. But it ignores the ULAE cost already absorbed when those claims were first opened.

Kittel uses a **weighted average** of paid and reported losses + ALAE (each weighted 50%) as the denominator. This reflects that the CY ULAE spend is driven equally by:
- Claims closing (captured by paid losses + ALAE)
- Claims that were open during the year (captured by reported losses + ALAE, which includes case reserves)

### Formula

**Step 1 — Compute the Kittel ratio:**

$$\text{Kittel Denominator}_{CY} = 0.5 \times \text{Paid (L+ALAE)}_{CY} + 0.5 \times \text{Reported (L+ALAE)}_{CY}$$

$$\text{Kittel Ratio}_{CY} = \frac{\text{Paid ULAE}_{CY}}{\text{Kittel Denominator}_{CY}}$$

**Step 2 — Apply to the ULAE reserve:**

The ULAE reserve for an accident year is the ULAE still to be incurred on **unreported and unpaid claims** (IBNR + case reserve deficiency). Since those future costs are 50% at open and 50% at close, only half of the IBNR base generates future ULAE costs:

$$\text{ULAE Reserve}_{AY} = \text{Kittel Ratio} \times 0.5 \times \text{IBNR (L+ALAE)}_{AY}$$

The ULAE **provision** (which includes amounts already paid + the reserve) uses reported losses + ALAE as the base, same as the basic method, but with the Kittel ratio.

### Worked Example

**Calendar year data (adding reported losses + ALAE):**

| CY | Paid ULAE | Paid L+ALAE | Reported L+ALAE | Kittel Denom | Kittel Ratio |
|---|---|---|---|---|---|
| 2022 | $580,000 | $5,200,000 | $5,900,000 | $5,550,000 | 10.5% |
| 2023 | $610,000 | $5,500,000 | $6,200,000 | $5,850,000 | 10.4% |
| 2024 | $650,000 | $5,800,000 | $6,500,000 | $6,150,000 | 10.6% |
| **Selected** | | | | | **10.5%** |

**Apply to accident year ULAE reserve (using IBNR base):**

| AY | Reported L+ALAE | Ultimate L+ALAE | IBNR L+ALAE | ULAE Reserve |
|---|---|---|---|---|
| 2022 | $5,600,000 | $5,700,000 | $100,000 | $5,250 |
| 2023 | $5,100,000 | $5,400,000 | $300,000 | $15,750 |
| 2024 | $4,800,000 | $5,350,000 | $550,000 | $28,875 |

ULAE Reserve = 10.5% × 0.5 × IBNR

### Comparison to Basic Method

The Kittel ratio (10.5%) is slightly lower than the basic ratio (11.2%) because the denominator is larger (including both paid and reported), but the reserve per dollar of IBNR is lower too — reflecting that some ULAE was already incurred when those claims were opened. The net effect varies by the maturity of the accident year cohort.

```python
import pandas as pd
import numpy as np

# Calendar year data for Kittel
cy_kittel = pd.DataFrame({
    'cy': [2022, 2023, 2024],
    'paid_ulae': [580_000, 610_000, 650_000],
    'paid_losses_alae': [5_200_000, 5_500_000, 5_800_000],
    'reported_losses_alae': [5_900_000, 6_200_000, 6_500_000],
})

# Kittel denominator: 50% paid + 50% reported
cy_kittel['kittel_denom'] = (0.5 * cy_kittel['paid_losses_alae']
                             + 0.5 * cy_kittel['reported_losses_alae'])
cy_kittel['kittel_ratio'] = cy_kittel['paid_ulae'] / cy_kittel['kittel_denom']

print("Kittel Refinement — Calendar Year Ratios:")
display(cy_kittel[['cy', 'paid_ulae', 'paid_losses_alae', 'reported_losses_alae',
                    'kittel_denom', 'kittel_ratio']].assign(
    paid_ulae=cy_kittel['paid_ulae'].map('${:,.0f}'.format),
    paid_losses_alae=cy_kittel['paid_losses_alae'].map('${:,.0f}'.format),
    reported_losses_alae=cy_kittel['reported_losses_alae'].map('${:,.0f}'.format),
    kittel_denom=cy_kittel['kittel_denom'].map('${:,.0f}'.format),
    kittel_ratio=cy_kittel['kittel_ratio'].map('{:.1%}'.format),
).rename(columns={
    'cy': 'CY', 'paid_ulae': 'Paid ULAE',
    'paid_losses_alae': 'Paid L+ALAE', 'reported_losses_alae': 'Reported L+ALAE',
    'kittel_denom': 'Kittel Denom (50/50)', 'kittel_ratio': 'Kittel Ratio'
}))

selected_kittel = cy_kittel['kittel_ratio'].mean()
print(f"\nSelected Kittel Ratio: {selected_kittel:.1%}")

# Apply to accident year ULAE reserve
ay_kittel = pd.DataFrame({
    'ay': [2022, 2023, 2024],
    'reported_losses_alae': [5_600_000, 5_100_000, 4_800_000],
    'ultimate_losses_alae': [5_700_000, 5_400_000, 5_350_000],
})

ay_kittel['ibnr_losses_alae'] = ay_kittel['ultimate_losses_alae'] - ay_kittel['reported_losses_alae']

# ULAE reserve = Kittel ratio × 0.5 × IBNR
# (half of the unreported ULAE cost is incurred at claim open, half at close)
ay_kittel['ulae_reserve'] = selected_kittel * 0.5 * ay_kittel['ibnr_losses_alae']

print("\nAccident Year ULAE Reserves (Kittel Refinement):")
display(ay_kittel.assign(
    reported_losses_alae=ay_kittel['reported_losses_alae'].map('${:,.0f}'.format),
    ultimate_losses_alae=ay_kittel['ultimate_losses_alae'].map('${:,.0f}'.format),
    ibnr_losses_alae=ay_kittel['ibnr_losses_alae'].map('${:,.0f}'.format),
    ulae_reserve=ay_kittel['ulae_reserve'].map('${:,.0f}'.format),
).rename(columns={
    'ay': 'AY',
    'reported_losses_alae': 'Reported L+ALAE',
    'ultimate_losses_alae': 'Ultimate L+ALAE',
    'ibnr_losses_alae': 'IBNR L+ALAE',
    'ulae_reserve': 'ULAE Reserve',
}))

print(f"\nTotal ULAE Reserve: ${ay_kittel['ulae_reserve'].sum():,.0f}")
```

## Method Comparison and Selection

### Summary

| Feature | CY Ratio Method | Kittel Refinement |
|---|---|---|
| **Data required** | Paid ULAE, Paid (L+ALAE) | Paid ULAE, Paid (L+ALAE), Reported (L+ALAE) |
| **Underlying assumption** | ULAE incurred proportionally to payments | ULAE incurred 50% at open, 50% at close |
| **Complexity** | Simple | Moderate |
| **Accuracy** | Good when claim lifecycles are short | Better when open/close ULAE mix varies by AY maturity |
| **Common use** | Quick analyses, lines with fast closure | More rigorous reserving analyses |

### When to Use Each Method

**Use the CY Ratio Method when:**
- Data limitations prevent access to both paid and reported loss + ALAE on a CY basis
- The line of business has very short-tailed claims (most ULAE incurred at close, little open-period cost)
- ULAE is a small component of total reserves and precision is less critical
- A rough check or benchmark is needed

**Use the Kittel Refinement when:**
- More accurate results are required
- The accident year cohorts have meaningfully different maturity (early AYs mostly closed, recent AYs mostly open)
- Long-tailed lines where claims remain open for years
- Claims handling costs are front-loaded (heavy investigation at open) or back-loaded (complex settlement)

### Judgment Factors in Ratio Selection

Regardless of which method is used, the actuary must select a ratio from the historical CY data. Key considerations:

- **Trend:** Is the ULAE ratio stable, increasing, or decreasing? Rising adjuster salaries or headcount changes may drive a trend.
- **Operational changes:** Outsourcing, offshoring, or automation of claims handling can cause step-changes in ULAE levels.
- **Volume effects:** Fixed costs create leverage — a growing book may show a declining ratio as overhead is spread over more claims.
- **Mix shift:** Changes in lines of business mix affect the ULAE ratio if different lines have different overhead allocations.
