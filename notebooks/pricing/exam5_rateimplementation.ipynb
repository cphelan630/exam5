# Rate Implementation & Marketing Considerations

The ratemaking process produces an **indicated rate change** — the actuarially justified change required to achieve rate adequacy. But the indicated change is rarely implemented as-is. Translating an indication into an actual filed rate requires additional steps: constructing the rate algorithm (base rates, relativities, additive fees), navigating regulatory and operational constraints, and weighing the competitive and marketing consequences of the change.

This notebook covers the final stage of the ratemaking process tested on CAS Exam 5, drawing from Werner & Modlin Chapters 15–16. By the end you should be able to:

1. **Distinguish variable, additive, and expense-fee premium components** and apply the correct formula for each.
2. **Derive a proposed base rate** in two scenarios: when relativities are unchanged, and when relativities are revised simultaneously.
3. **Identify types of regulatory constraints** on rates and describe the options available to a company facing each constraint.
4. **Apply competitive comparison metrics** — percent competitive position, dollar competitive position, percent win, and rank — to evaluate market position.
5. **Describe distributional analysis, dislocation analysis, lifetime value analysis, and optimized pricing** and explain when each is appropriate.
6. **Define soft and hard market conditions** and explain how pricing strategy differs across the underwriting cycle.

## Formula Sheet Quick Reference

### Rate Algorithm Components

| Formula | Expression |
|---|---|
| **Expense fee per exposure** | $\text{Expense Fee} = \dfrac{\text{Fixed Expense per Exposure}}{\text{PLR}}$ |
| **Additive premium per exposure** | $\text{Additive Prem} = \dfrac{\text{Additive Amount per Exposure}}{1 - VE\% - TP\%}$ |
| **Variable permissible loss ratio** | $VPLR = 1 - \text{Expense Ratio} - \text{Target Return}$ |
| **Total permissible loss ratio** | $PLR = 1 - \text{LAE Ratio} - \text{Expense Ratio} - \text{Target Return}$ |

### Deriving the Proposed Base Rate

| Scenario | Formula |
|---|---|
| **No relativity change** (all variable premium) | $\text{Proposed Base Rate} = \dfrac{\text{Proposed Variable Premium}}{\text{Current Variable Premium}} \times \text{Current Base Rate}$ |
| **No relativity change** (shortcut) | $\text{Proposed Base Rate} = (1 + \%\text{ Change}) \times \text{Current Base Rate}$ |
| **With relativity changes** | $\text{Proposed Base Rate} = \dfrac{\text{Proposed Overall Avg Premium}}{\text{Current Overall Avg Premium}} \times \text{Current Base Rate} \times \dfrac{\text{Avg Current Relativity}}{\text{Avg Proposed Relativity}}$ |

### Marketing Metrics

| Formula | Expression |
|---|---|
| **% Competitive position** | $\dfrac{\text{Competitor Premium}}{\text{Company Premium}} - 1$ |
| **\$ Competitive position** | $\text{Competitor Premium} - \text{Company Premium}$ |
| **% Win** | $\dfrac{\text{\# Risks Meeting Criteria (e.g., company cheaper)}}{\text{Total \# of Risks}}$ |
| **Close ratio** | $\dfrac{\text{\# Accepted Quotes}}{\text{Total \# of Quotes}}$ |
| **Retention ratio** | $\dfrac{\text{\# Policies Renewed}}{\text{\# Potential Renewals}}$ |
| **% Policy growth** | $\dfrac{\text{New Policies Written} - \text{Lost Policies}}{\text{Total Policies}}$ |

## Expense Fees and Additive Premium

### Variable vs. Additive Premium

Not all premium components scale proportionally with the base rate and rating factor relativities. W&M distinguishes three types:

| Component | Definition | Scales with risk characteristics? |
|---|---|---|
| **Variable premium** | Product of base rate and all relativities | Yes — reflects individual risk characteristics |
| **Additive premium** | Flat dollar amount added to every policy | No — same for all policyholders |
| **Expense fee** | Fixed expense per exposure backed out of PLR | No — reflects fixed costs per policy/exposure |

**When additive or expense fees are appropriate:** Some costs are roughly constant per policy regardless of size (e.g., policy issuance, mailing, state filing fees). Treating these as a constant percentage of variable premium overcharges large risks (constant cost is a small % of large premium) and undercharges small risks (constant cost is a large % of small premium). The additive/expense-fee approach corrects this.

### Expense Fee Formula

$$\text{Expense Fee per Exposure} = \frac{\text{Fixed Expense per Exposure}}{PLR}$$

The denominator $PLR$ (permissible loss ratio) converts the net fixed expense cost to a gross premium equivalent — it "grosses up" the expense amount to account for the variable expenses and profit margin that must also be recovered from each premium dollar.

### Additive Premium Formula

$$\text{Additive Premium per Exposure} = \frac{\text{Additive Amount per Exposure}}{1 - VE\% - TP\%}$$

where $VE\%$ = variable expense percentage and $TP\%$ = target profit percentage. The denominator again grosses up the additive cost, since each dollar of additive premium must also cover variable expenses and profit.

### Worked Example

**Given:**

| Item | Value |
|---|---|
| Fixed expense per exposure (policy issuance + fees) | $45 |
| Variable expense ratio | 20% |
| Target profit ratio | 5% |
| Expected LAE ratio | 10% |

**Step 1: Permissible loss ratio**

$$PLR = 1 - 0.10 - 0.20 - 0.05 = 0.65$$

**Step 2: Expense fee**

$$\text{Expense Fee} = \frac{\$45}{0.65} = \$69.23 \text{ per exposure}$$

**Interpretation:** Every policy is charged a flat $\$69.23$ additive fee regardless of the variable premium amount. A risk with a $\$200$ variable premium pays $\$269.23$ total; a risk with a $\$5{,}000$ variable premium also pays the same flat $\$69.23$ on top.

> **Exam tip:** The expense fee uses the total PLR in the denominator (including LAE). The additive premium formula uses $1 - VE\% - TP\%$ (excluding LAE). Know which denominator applies to which formula — the difference is whether LAE is included as a variable cost.

## Deriving the Proposed Base Rate

Once the overall indicated average premium is determined, the base rate must be set so that the rating algorithm produces that average. The formula depends on whether relativities are being revised simultaneously.

### Case 1: No Change to Relativities (All Variable Premium)

When all premium is variable (no additive fee) and relativities are not changing:

$$\text{Proposed Base Rate} = \frac{\text{Proposed Variable Premium}}{\text{Current Variable Premium}} \times \text{Current Base Rate}$$

Equivalently, if you know the overall percentage change:

$$\text{Proposed Base Rate} = (1 + \%\text{ Rate Change}) \times \text{Current Base Rate}$$

**Worked Example — No Relativity Change:**

| Item | Value |
|---|---|
| Current base rate | $800 |
| Current average variable premium | $1,200 |
| Proposed average variable premium | $1,296 (from indication: +8%) |

$$\text{Proposed Base Rate} = \frac{1{,}296}{1{,}200} \times 800 = 1.08 \times 800 = \$\mathbf{864}$$

All relativities stay the same; only the base rate shifts.

### Case 2: With Changes to Relativities

When relativities are revised at the same time as the overall level change, changing the base rate proportionally would produce the wrong average premium — the revised relativities shift the weighted average relativity, so the base rate must be adjusted further:

$$\text{Proposed Base Rate} = \frac{\text{Proposed Overall Avg Premium}}{\text{Current Overall Avg Premium}} \times \text{Current Base Rate} \times \frac{\text{Avg Current Relativity}}{\text{Avg Proposed Relativity}}$$

The last factor corrects for the shift in the exposure-weighted average relativity caused by the revised factors.

**Worked Example — With Relativity Changes:**

| Item | Value |
|---|---|
| Current base rate | $800 |
| Current avg premium (overall) | $1,200 |
| Proposed avg premium (overall) | $1,296 (+8%) |
| Exposure-weighted avg current relativity | 1.500 |
| Exposure-weighted avg proposed relativity | 1.440 (relativities revised downward) |

$$\text{Proposed Base Rate} = \frac{1{,}296}{1{,}200} \times 800 \times \frac{1.500}{1.440} = 1.08 \times 800 \times 1.0417 = \$\mathbf{900}$$

Because the average relativity decreased (relativities were reduced), the base rate must be set higher than the simple 8% increase would suggest, to maintain the intended average premium.

> **Exam tip:** The ratio $\text{Avg Current Relativity} / \text{Avg Proposed Relativity}$ is often called the **off-balance correction**. If proposed relativities are lower on average than current (e.g., a discount tier was narrowed), this ratio exceeds 1.0 and the base rate must increase by more than the overall indicated change. The logic: the base rate and the relativities must combine to produce the correct average; if relativities decrease, the base rate must compensate.

## Regulatory Constraints

In the United States, insurance rates are regulated at the state level. Regulators can impose constraints that prevent a company from implementing the actuarially indicated rate, creating a gap between what the actuary indicates and what can actually be filed.

### Types of Regulatory Constraints

| Constraint Type | Description | Example |
|---|---|---|
| **Rate suppression** | Regulators restrict the overall average rate change below the indicated level | Indication: +12%; approved: +6% |
| **Individual rate caps** | Limits on how much any single policy's rate can increase in one period | No single risk may see more than +25% year over year |
| **Variable prohibitions** | Certain rating variables cannot be used | Credit scores banned in some states for personal auto |
| **Advance notice requirements** | Rate changes must be filed X days before effective date | 60-day advance notice required |
| **Mandatory techniques** | Regulator prescribes specific ratemaking methods | Sequential analysis required for classification ratemaking |
| **Prior approval** | New rates must be approved before taking effect | Rate changes require commissioner sign-off |

### Company Options When Facing Constraints

When a regulatory constraint prevents a company from implementing the actuarially sound rate, several responses are available:

1. **Take legal action** — Challenge the regulation in court if it constitutes an unconstitutional taking or violates state law.
2. **Revise underwriting guidelines** — Restrict new business or non-renew policies where rates are believed to be inadequate, effectively limiting adverse exposure without changing the filed rate.
3. **Adjust marketing** — Reduce marketing spend or redirect agents away from market segments where rates are constrained, minimizing volume at inadequate rates.
4. **Use a substitute variable** — If a specific rating variable is banned, identify a related but permitted variable that captures similar risk information (e.g., if credit score is banned, use residential stability or homeownership status).
5. **Accept the constraint** — If the regulated rate produces an acceptable combined ratio, implement as allowed and revisit in the next filing cycle.

> **Exam tip:** The exam frequently tests the distinction between a company's options when (a) the overall rate level is suppressed vs. (b) a specific variable is prohibited. Options 2–3 are most relevant for overall suppression; option 4 is specific to variable prohibitions. Option 1 is always available but slow and expensive.

## Operational Constraints

Even when rates are actuarially sound and regulatorily permissible, practical implementation may be constrained by internal organizational limitations.

### Types of Operational Constraints

| Constraint | Examples |
|---|---|
| **Systems limitations** | Policy administration systems cannot process a new rating variable; billing systems cannot handle additive premium structures |
| **Resource constraints** | Insufficient actuarial, IT, or underwriting staff to implement a complex new algorithm on the desired timeline |
| **Limited data access** | Third-party data (e.g., credit bureau scores, telematics) requires vendor contracts, IT integration, and privacy compliance before use |
| **Policyholder disruption** | Implementing large rate changes on renewal may spike cancellations; phased implementation may be operationally safer |

### Cost-Benefit Analysis

When facing operational constraints, actuaries use a **cost-benefit analysis** to evaluate options:

$$\text{Net Benefit} = \text{Expected Improvement in Profitability / Risk Selection} - \text{Implementation Costs}$$

- **Implementation costs** include one-time IT development, ongoing data vendor fees, training, compliance, and testing.
- **Benefits** include improved loss ratios (better risk selection), reduced adverse selection, and competitive advantages from more accurate pricing.
- The analysis is conducted over a **specified time horizon** to allow benefits to accumulate (often 3–5 years).

**Common cost-benefit trade-off:** A new telematics-based rating variable may cost $\$2M$ to implement. If it improves the loss ratio by 1.5 points on a $\$150M$ book, the benefit is $\$2.25M$/year — a payback period of roughly one year. The cost-benefit analysis would support implementation.

> **Exam tip:** Operational constraints differ from regulatory constraints in that the company has full control — operational constraints are internal limitations, not externally imposed. A cost-benefit analysis frames the decision as a business choice rather than a compliance problem.

## Marketing Considerations — Overview

The actuarial indication tells an insurer what rate is needed to cover expected costs. But **demand is elastic** — changing rates affects how many policies are written and renewed. Marketing considerations bridge the gap between actuarial soundness and business strategy.

### New Business vs. Renewal Business

Insurers generally treat new and renewal business separately because:

- **New business** involves prospects shopping the market — price competitiveness is the primary driver of writing new policies.
- **Renewal business** involves existing policyholders — inertia, satisfaction, and switching costs reduce price sensitivity, but large rate increases can trigger cancellations.

Analyzing these segments separately allows the company to understand whether a rate change will primarily affect new business growth or existing policyholder retention.

### Factors Affecting Renewal Propensity

Whether an existing policyholder renews depends on:

- **Competitor pricing** — if the competitor's premium is significantly lower, the policyholder may shop
- **Total cost of the product** — not just base premium but fees, deductibles, and coverage differences
- **Rate change** — even if the premium is still competitive, a large year-over-year increase prompts shopping behavior
- **Customer characteristics** — price sensitivity varies; some segments (e.g., older homeowners) have higher inertia
- **Customer satisfaction and brand loyalty** — claims handling quality and relationship management affect retention independent of price

## Competitive Comparison Metrics

To assess market position, insurers compare their rates to competitors' rates on the same risks. Four standard metrics are used:

### The Four Metrics

**1. Percent Competitive Position**
$$\% \text{ Competitive Position} = \frac{\text{Competitor Premium}}{\text{Company Premium}} - 1$$

Positive value means the competitor charges *more* — company is cheaper (favorable position). Negative value means the company charges more — company is more expensive (unfavorable).

**2. Dollar Competitive Position**
$$\$\text{ Competitive Position} = \text{Competitor Premium} - \text{Company Premium}$$

Same sign convention. Dollar position matters more for high-premium commercial lines where percentage differences are less intuitive.

**3. Percent Win**
$$\% \text{ Win} = \frac{\text{\# Risks Where Company Premium} < \text{Competitor Premium}}{\text{Total \# of Risks}}$$

The proportion of the market where the company has a price advantage. Can be segmented by territory, class, or any other dimension to identify where the company is most/least competitive.

**4. Rank**
$$\text{Rank} = \text{Company's ranking among all competitors by price}$$

e.g., Rank = 2 means the company is the second cheapest among five competitors on a given risk. Rank analysis is useful when multiple competitors are tracked simultaneously.

### Worked Example

For a sample of 1,000 risks, the company's premiums are compared to Competitor A:

| Territory | Avg Company Premium | Avg Competitor A Premium | % Competitive Position | % Win |
|---|---|---|---|---|
| Urban | $1,800 | $1,620 | $1{,}620/1{,}800 - 1 = -10\%$ | 18% |
| Suburban | $1,200 | $1,260 | $1{,}260/1{,}200 - 1 = +5\%$ | 63% |
| Rural | $900 | $945 | $945/900 - 1 = +5\%$ | 71% |

**Interpretation:** The company is uncompetitive in urban areas (10% more expensive than Competitor A; wins only 18% of risks there), but has a slight price advantage in suburban and rural markets. A targeted rate reduction in urban territory — if actuarially supportable — could improve competitive position and new business volume there.

> **Exam tip:** Competitive position formulas always put the competitor in the numerator. A positive % competitive position is favorable for the company (you're cheaper). Know which direction is favorable before interpreting exam scenarios.

## Distributional Analysis and Dislocation Analysis

### Distributional Analysis

**Distributional analysis** studies how the composition of new and renewal business changes over time across customer segments. Rather than looking only at average premium or volume, it tracks *who* is buying and renewing.

**Example use case:** After implementing a new age-based territorial rating variable, distributional analysis might reveal that the high-risk urban segment is growing faster than the low-risk suburban segment — a sign that the new variable is creating adverse selection rather than correcting it.

**Key questions distributional analysis answers:**
- Is the mix of new vs. renewal business shifting?
- Are certain risk segments growing or shrinking disproportionately?
- Is the company attracting the segments where its rates are below cost (adverse selection), or the segments where it is profitable?

### Dislocation Analysis

**Dislocation analysis** studies how rate changes affect individual policyholders' propensity to renew. A rate change is *dislocating* if it causes significant premium increases for specific groups of policyholders, increasing their probability of non-renewal.

**The dislocation problem:** Even when the *average* rate change is modest (+5%), individual policies may experience very different changes. A policyholder who was in a misclassified territory might see +40% upon reclassification. Dislocation analysis identifies how many policyholders fall into large-increase buckets and estimates the expected non-renewal from each bucket.

**Use in implementation decisions:** If dislocation analysis shows that 15% of the book faces increases above +25%, the company might:
- Phase in the change over 2–3 years to moderate the annual impact
- Apply individual increase caps and recoup through future filings
- Accept higher non-renewals if those risks are unprofitable anyway

| Analysis Type | Question Answered | Output |
|---|---|---|
| Distributional analysis | Who is in my book, and is the mix changing? | Segment-level volume and loss ratio trends |
| Dislocation analysis | How does my rate change hit individual policyholders? | Distribution of premium changes; estimated non-renewal by change bucket |

## Lifetime Value Analysis and Optimized Pricing

### Lifetime Value (LTV) Analysis

Traditional ratemaking is **single-period**: it seeks a rate that is adequate for the upcoming policy year in isolation. **Lifetime value analysis** takes a multi-period view, recognizing that a policyholder who renews multiple times generates more total profit than one who churns after one year — even if the single-year loss ratio is identical.

**LTV accounts for:**
- The probability that a policyholder renews for $t$ more years
- The expected underwriting profit per year if retained
- The acquisition cost paid only once (at new business)
- Any changes in risk quality or pricing over time

**Implication for pricing:** LTV analysis may justify offering a slightly below-cost rate to acquire a new policyholder if the expected lifetime profitability — net of the initial underpricing — is positive. Conversely, it may support aggressive retention pricing to keep highly profitable long-term customers.

**Key limitation:** LTV requires reliable estimates of renewal probability, which depend on competitive position, customer satisfaction, and market dynamics — all of which are difficult to forecast. Additionally, using lifetime value to justify an inadequate rate in year one may create regulatory problems.

### Optimized Pricing

**Optimized pricing** uses multivariate statistical models to set prices that maximize a specific objective (usually expected profit or lifetime value) subject to constraints. It explicitly models **demand** in addition to cost:

$$\text{Expected Profit} = \text{Premium} \times P(\text{Buy/Renew at that Premium}) - \text{Expected Costs}$$

By estimating the demand curve (how purchase/renewal probability varies with price), optimized pricing identifies the premium that maximizes profit — which may be higher or lower than the pure cost-based indication depending on competitive dynamics.

**Components of an optimized pricing model:**
1. **Cost model** — expected losses and expenses per risk (standard ratemaking output)
2. **Demand model** — elasticity of purchase and renewal probability with respect to price
3. **Optimization** — find the price that maximizes profit given the demand response, subject to regulatory and operational constraints

> **Exam tip:** Optimized pricing is an *enhancement* to ratemaking, not a replacement. The cost model (indication) remains the foundation. The demand model determines whether to price above, at, or below the indication based on competitive sensitivity. Be able to explain why optimized pricing might charge *less* than the cost-based indication (high-elasticity market; price above cost loses volume and overall profit) or *more* (low-elasticity market; inelastic buyers; pricing at cost leaves money on the table).

## Underwriting Cycles

The insurance market moves through alternating phases of **soft** and **hard** market conditions, driven by the interplay of pricing, competition, capital, and loss experience.

### Soft vs. Hard Market

| Characteristic | Soft Market | Hard Market |
|---|---|---|
| **Price level** | Low (below actuarially adequate) | High (at or above adequate) |
| **Premium growth** | High (easy to write new business) | Low (demand suppressed by price) |
| **Underwriting standards** | Loose (competitors accept marginal risks) | Tight (insurers more selective) |
| **Coverage availability** | Broad (many insurers compete) | Narrow (some risks uninsurable) |
| **Combined ratio** | Deteriorating (prices below cost) | Improving (prices above cost) |
| **Capital position** | Surplus high (capital seeking return) | Surplus depleted (by prior losses) |

### Why Cycles Exist

Several forces combine to create and perpetuate the cycle:

1. **Capital flows:** After profitable years, new capital enters the market seeking returns, increasing competition and driving prices down (soft market onset). After major loss events, capital is depleted and rates harden.
2. **Reserve development:** Insurers may underestimate reserves in competitive soft markets, making results appear better than they are. When adverse reserve development is recognized, it triggers a hard market correction.
3. **Competition lag:** It takes time for new entrants to be licensed, write policies, and collect premiums. This lag means markets overshoot in both directions.
4. **Correlated investment income:** When investment returns are high, insurers can tolerate weaker underwriting results, softening the market. When investment returns fall (as in a recession), underwriting must generate more profit, hardening the market.

### Pricing Strategy Across the Cycle

A disciplined insurer adjusts strategy based on market conditions:

| Strategy | Soft Market | Hard Market |
|---|---|---|
| **Rate level** | Hold firm at cost-based rates; accept volume loss rather than write inadequate business | Capture the hard market opportunity; file indicated increases promptly |
| **Underwriting** | Tighten standards; non-renew marginal risks | Relax selectively; grow volume while prices are adequate |
| **Product design** | Maintain coverage terms; resist broadening for competitive edge | Consider broadening coverage as a value-add without sacrificing adequacy |
| **Reinsurance** | May be able to reduce reinsurance cost | Lock in reinsurance before prices spike further |

> **Exam tip:** The underwriting cycle is caused by external market dynamics, not individual insurer behavior — a single company cannot control the cycle. However, a company can *react* to the cycle better or worse. The exam may ask you to identify soft vs. hard market conditions from given data (high growth + deteriorating combined ratio = soft) or to recommend appropriate strategy.

## Exam Tips & Common Pitfalls

### Common Mistakes

| Topic | Common Mistake | Correction |
|---|---|---|
| **Expense fee** | Using variable expense % in the denominator | Expense fee uses *total* PLR = $1 - LAE\% - VE\% - TP\%$ |
| **Additive premium** | Using PLR in the denominator | Additive premium uses $1 - VE\% - TP\%$ (PLR without the LAE ratio) |
| **Base rate (no relativity change)** | Treating the on-level factor as the base rate multiplier | If relativities are unchanged, proposed base rate = $(1 + \%\text{ change}) \times$ current base rate — the off-balance correction is 1.0 |
| **Base rate (with relativity change)** | Ignoring the off-balance correction | Must multiply by Avg Current Rel / Avg Proposed Rel to maintain correct average premium |
| **% Competitive position** | Putting company premium in the numerator | Formula is Competitor / Company $- 1$; positive = favorable (you're cheaper) |
| **% Win** | Counting ties as wins | % Win counts only risks where company is *strictly* cheaper than the competitor |
| **LTV** | Confusing LTV justification of below-cost pricing with actuarial adequacy | LTV may justify a below-cost year-one price strategically, but this creates regulatory exposure and must be disclosed |
| **Underwriting cycles** | Saying a soft market is "good" | A soft market is characterized by *inadequate rates* — it is financially dangerous even if volume is high |

### Key Conceptual Points for Essay Questions

- **Why is the indicated rate not always the implemented rate?** Three categories of constraints: actuarial (credibility, data limitations), regulatory (rate suppression, variable bans), and operational (systems, resources). The actuary's role is to document the indication and explain departures from it.

- **Why would a company price below the actuarial indication?** Legitimate reasons: LTV analysis shows long-term profitability despite initial underpricing; competitive analysis shows the market will not support the indicated rate. Illegitimate reasons: market share pressure, management pressure to grow. The actuary must evaluate whether below-indication pricing is defensible.

- **What is the difference between distributional analysis and dislocation analysis?** Distributional analysis looks at the *book-level* mix of business (segment composition) over time. Dislocation analysis looks at *individual-policy-level* premium changes from a rate revision (who gets large increases). Both inform implementation decisions but answer different questions.

- **Why does the off-balance correction factor appear when relativities change?** The base rate times the average relativity must equal the target average premium. If relativities shift on average (e.g., all discounts expanded), the base rate must compensate upward to preserve the target, and vice versa. The off-balance factor quantifies this mechanical correction.

- **What causes a hard market?** Capital depletion after large losses (catastrophes, adverse reserve development), combined with departures of marginal competitors, reduces supply and allows prices to rise. High prices attract new capital, which eventually restores competition and softens the market again.
