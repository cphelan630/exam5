## Formula Sheet Quick Reference

| Formula | Expression |
|---|---|
| **% Reported by age $d$** | $\text{RDF}(d) = 1 / \text{CDF}(d \to \text{ult})$ |
| **Step $n$ rate** | $R_n^{CM} = \text{RDF}(12n) \times R^{occ}$ |
| **Step factor** $n \geq 2$ | $SF_n = R_n^{CM} / R_{n-1}^{CM} = \text{LDF}(12(n-1) \to 12n)$ |
| **First step factor** | $SF_1 = R_1^{CM} / R^{occ} = \text{RDF}(12)$ |
| **Mature CM factor** | $\text{MCF} = \text{RDF}(\text{mature age}) \approx 1 - \text{PCF}$ |
| **Prospective conversion factor** | $\text{PCF} = 1 - \text{RDF}(\text{policy expiration age})$ |
| **Tail coverage premium** | $R^{tail} = \text{PCF} \times R^{occ}$ |
| **Extended reporting period** | $\text{PCF}(d_{ERP}) = \text{RDF}(d_{exp} + d_{ERP}) - \text{RDF}(d_{exp})$... relative premium = $\text{PCF}(d_{ERP}) \times R^{occ}$ |

> **Key insight:** CM Step $n$ rates are driven by **report-year** development factors, not accident-year development. The RDF uses the cumulative % of ultimate losses reported by a given age.

## Section 1: Background — Why Claims-Made Policies Exist

### The Report Lag Problem

For **long-tailed lines** (medical malpractice, D&O, E&O, professional liability), there can be a multi-year gap between when a loss *occurs* and when a claim is *reported*. Under occurrence coverage:
- The insurer at risk when the event occurred is responsible — even if that insurer has exited the market
- The occurrence policy must price for all future reports, including claims reported 5–10+ years hence
- Predicting future claim costs over a decade creates enormous pricing uncertainty

**Claims-made (CM) policies** solve this by limiting coverage to claims *reported* during the policy period:
- The *current* insurer (who knows today's cost environment) is on the hook
- Pricing uncertainty is reduced significantly — you're estimating costs for claims reported this year, not claims arising from events this year
- In return, the insured must maintain continuous coverage or purchase tail coverage

### Occurrence vs. Claims-Made: Coverage Trigger

| Feature | Occurrence | Claims-Made |
|---|---|---|
| **Coverage trigger** | Loss *occurs* during policy period | Claim *reported* during policy period |
| **IBNR tail** | Long — claims can emerge years later | Short/none — only IBNER remains |
| **Common lines** | Most P&C lines | Med mal, D&O, E&O, cyber |
| **Retroactive date** | N/A | First inception date; excludes prior known claims |
| **Tail coverage needed** | No | Yes — if switching insurers or canceling |
| **Pricing uncertainty** | Higher (long-tail) | Lower (shorter development) |

### The Retroactive Date

A CM policy has a **retroactive date** — the earliest date from which prior acts are covered. Claims arising from events *before* the retroactive date are excluded.

- **First-year CM** (Step 1): Retroactive date = policy inception. Only covers claims both incurred AND reported in year 1.
- **Second-year CM** (Step 2): Retroactive date still = original inception. Now covers claims from year 1 *reported* in year 2, plus claims from year 2 reported in year 2.
- **Mature CM**: Retroactive date = many years ago. Covers claims from a long history of prior acts, all reported during the current policy year.

## Section 2: The Report Development Pattern

Claims-made pricing uses **report-year development factors** — the cumulative % of ultimate losses that are *reported* (not just paid) by a given age.

Let:
- $\text{CDF}(d)$ = cumulative development factor from age $d$ months to ultimate
- $\text{RDF}(d)$ = reported development factor = $1 / \text{CDF}(d)$ = % of ultimate reported by age $d$

Since $\text{CDF}(d)$ decreases toward 1.0 as $d \to \infty$, $\text{RDF}(d)$ increases toward 1.0.

### Example Development Pattern (Medical Malpractice)

| Age (months) | CDF to Ultimate | RDF = 1/CDF |
|---|---|---|
| 12 | 3.052 | 32.8% |
| 24 | 1.943 | 51.5% |
| 36 | 1.427 | 70.1% |
| 48 | 1.183 | 84.5% |
| 60 | 1.073 | 93.2% |
| 72 | 1.026 | 97.5% |
| 84 | 1.008 | 99.2% |
| Ult | 1.000 | 100.0% |

**Interpretation:** Only 32.8% of medical malpractice claims (by dollar) are reported within the first 12 months of the accident year. By 84 months (7 years), 99.2% have been reported.

This slow report pattern is what drives the need for claims-made coverage on this line — an occurrence policy must reserve for the 67.2% of unreported claims at 12 months, across a decade or more of loss development.

## Section 3: The Step Method

The step method calculates the CM rate for each year of a CM program. The key insight is that each additional year of CM coverage captures the losses that were incurred in prior years but not yet reported.

### Step Rates and Step Factors

**Step $n$ rate** = the CM rate in year $n$ of a continuous CM program:

$$R_n^{CM} = \text{RDF}(12n) \times R^{occ}$$

where $R^{occ}$ is the corresponding occurrence rate.

**Step factor** = the ratio of consecutive step rates:

$$SF_n = \frac{R_n^{CM}}{R_{n-1}^{CM}} = \frac{\text{RDF}(12n)}{\text{RDF}(12(n-1))} \quad (n \geq 2)$$

This simplifies to: $SF_n = \text{LDF}(12(n-1) \to 12n)$ — the age-to-age link ratio for the $n$-th transition.

The **first step factor** (transition from occurrence to year-1 CM) is:
$$SF_1 = \frac{R_1^{CM}}{R^{occ}} = \text{RDF}(12) = \frac{1}{\text{CDF}(12 \to \text{ult})}$$

### Worked Example — Medical Malpractice

Using the development pattern above and an occurrence rate of **\$1,000**:

| Step | Age Covered | RDF | CM Rate | Step Factor |
|---|---|---|---|---|
| Occurrence | — | 100.0% | \$1,000 | — |
| Step 1 (Year 1 CM) | 0–12 mo | 32.8% | \$328 | 0.328 |
| Step 2 (Year 2 CM) | 0–24 mo | 51.5% | \$515 | 1.570 |
| Step 3 (Year 3 CM) | 0–36 mo | 70.1% | \$701 | 1.361 |
| Step 4 (Year 4 CM) | 0–48 mo | 84.5% | \$845 | 1.206 |
| Step 5 (Year 5 CM) | 0–60 mo | 93.2% | \$932 | 1.103 |
| Step 6 (Year 6 CM) | 0–72 mo | 97.5% | \$975 | 1.046 |
| Step 7 (Year 7 CM) | 0–84 mo | 99.2% | \$992 | 1.018 |
| **Mature** | All | ~100% | ~\$1,000 | \to 1.0 |

**Observations:**
1. Year 1 CM is dramatically cheaper than occurrence (only 32.8¢ per dollar of occurrence premium) because only a small fraction of claims are reported in the first year
2. Each subsequent step adds the prior-year development — Step 2 adds claims reported in months 13–24
3. Step factors > 1 (except the first) — the CM rate increases each year
4. Step factors converge to 1.0 at maturity
5. The **mature CM rate ≈ occurrence rate** for a stable book in steady state

### Why Mature CM ≈ Occurrence Rate

In a stable (no trend, constant volume) book, the mature CM policy covers claims reported this year from *all prior accident years combined*. The expected volume of reports equals the total expected claims for one accident year — the same as an occurrence policy. Hence the rates converge.

In practice, mature CM rates are slightly below occurrence rates because a small fraction of claims (the "true tail") are never reported before policy expiration even with many years of coverage — the prospective conversion factor (Section 5).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Development pattern: age-to-age factors
ldfs = pd.Series(
    [3.052/1.943, 1.943/1.427, 1.427/1.183, 1.183/1.073, 1.073/1.026, 1.026/1.008, 1.008],
    index=[(12,24),(24,36),(36,48),(48,60),(60,72),(72,84),(84,'ult')]
)

# CDFs to ultimate and RDFs
ages = [12, 24, 36, 48, 60, 72, 84]
cdfs = [3.052, 1.943, 1.427, 1.183, 1.073, 1.026, 1.008]
rdfs = [1/c for c in cdfs]

occurrence_rate = 1000.0
cm_rates = [r * occurrence_rate for r in rdfs]

# Step factors
sf = [rdfs[0]]  # SF_1 = RDF(12) relative to occurrence
for i in range(1, len(rdfs)):
    sf.append(rdfs[i] / rdfs[i-1])

df = pd.DataFrame({
    'Step': [f'Step {i+1}' for i in range(len(ages))],
    'Age Covered (mo)': [f'0–{a}' for a in ages],
    'CDF to Ult': cdfs,
    'RDF (% Reported)': [f'{r:.1%}' for r in rdfs],
    'CM Rate ($)': [f'${r:,.0f}' for r in cm_rates],
    'Step Factor': [f'{s:.3f}' for s in sf],
})

# Add occurrence row at top
occ_row = pd.DataFrame([{
    'Step': 'Occurrence', 'Age Covered (mo)': '—', 'CDF to Ult': 1.000,
    'RDF (% Reported)': '100.0%', 'CM Rate ($)': f'${occurrence_rate:,.0f}', 'Step Factor': '—'
}])
df = pd.concat([occ_row, df], ignore_index=True)

print("Claims-Made Step Rates (Occurrence Rate = $1,000):")
display(df.set_index('Step'))

# Plot CM rates converging to occurrence
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot([0] + ages, [occurrence_rate] + cm_rates, 'o-', color='steelblue', label='CM Rate by Step')
ax.axhline(occurrence_rate, color='tomato', linestyle='--', label='Occurrence Rate')
ax.set_xlabel('Age of CM Program (months)')
ax.set_ylabel('Rate ($)')
ax.set_title('Claims-Made Step Rates Converging to Occurrence Rate')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Section 4: Mature Claims-Made Rate

Once a CM program has been in place for enough years that the step factors are essentially 1.0, the policy is **mature**. Practically, maturity is often defined as when the remaining development beyond the latest step is less than a few percent.

**Mature claims-made factor (MCF):**

$$\text{MCF} = \text{RDF}(\text{mature age}) = 1 - \text{PCF}$$

In steady state with no trend: MCF ≈ 1.0, so the mature CM rate ≈ occurrence rate.

### Impact of Trend

In practice, the mature CM rate differs from the occurrence rate because of **trend**. Under CM, older accident years' claims are reported in the current year — but those losses occurred at *prior* cost levels. The insurer must trend those prior-year losses to current levels when pricing.

A **loss trend** causes the mature CM rate to be *lower* than the occurrence rate because:
- CM covers reports from prior years (lower cost environment)
- Occurrence covers current-year claims at today's higher cost environment

A trend adjustment factor is applied to each prior accident year's contribution:

$$R^{CM}_{mature, trend} = R^{occ} \times \sum_{j=0}^{\infty} [\text{RDF}(12(j+1)) - \text{RDF}(12j)] \times (1+t)^{-j}$$

where $t$ is the annual loss trend and $j$ indexes prior accident years (0 = current year, 1 = one year prior, etc.).

> **Exam tip:** A positive trend causes the mature CM rate to be *less than* the occurrence rate because you're pricing prior-year losses at inflation-adjusted levels lower than current. This is a common exam conceptual question.

## Section 5: Prospective Conversion Factor and Tail Coverage

### The PCF

When a CM policy expires (or the insured cancels or switches insurers), a fraction of claims from prior accident years remain unreported. These claims are NOT covered by the expired CM policy.

The **Prospective Conversion Factor (PCF)** is the % of ultimate losses that remain unreported at policy expiration:

$$\text{PCF} = 1 - \text{RDF}(d_{exp}) = 1 - \frac{1}{\text{CDF}(d_{exp} \to \text{ult})}$$

For a mature CM policy expiring at the end of year 7 (age 84 months):
$$\text{PCF} = 1 - 0.992 = 0.008 = 0.8\%$$

The tail policy premium = PCF × occurrence rate = 0.008 × \$1,000 = **\$8**.

### Extended Reporting Period ("Tail Coverage")

Instead of unlimited tail coverage, the insured may purchase an **extended reporting period (ERP)** endorsement covering claims reported within a limited additional window (e.g., 12 or 24 months after expiration).

| Tail Type | Coverage | Premium |
|---|---|---|
| **1-year ERP** | Claims reported in months 85–96 | [RDF(96) − RDF(84)] × occurrence rate |
| **2-year ERP** | Claims reported in months 85–108 | [RDF(108) − RDF(84)] × occurrence rate |
| **Unlimited ERP** | All remaining claims | PCF × occurrence rate |

Most professional liability lines require ERP pricing when the CM policy is not renewed.

### Retro-Conversion Factor

The **retro-conversion factor (RCF)** converts a mature CM rate *back* to the equivalent occurrence rate. This is the inverse relationship: 

$$R^{occ} = \frac{R^{CM}_{mature}}{\text{MCF}}$$

Used when: the insurer has CM historical data and wants to price an occurrence policy (or compare CM rates to occurrence benchmarks).

```python
import pandas as pd
import numpy as np

# Extended development past age 84 (years 8-10)
ages_extended = ages + [96, 108, 120]
cdfs_extended = cdfs + [1.003, 1.001, 1.000]
rdfs_extended = [1/c for c in cdfs_extended]

# PCF at each maturity
print("Prospective Conversion Factors and Tail Coverage Pricing:")
print(f"{'Age (mo)':>10} {'RDF':>10} {'PCF':>10} {'Tail Premium ($)':>18}")
print("-" * 52)
for age, rdf in zip(ages_extended, rdfs_extended):
    pcf = 1 - rdf
    tail = pcf * occurrence_rate
    print(f"{age:>10} {rdf:>10.3f} {pcf:>10.3f} {tail:>18.2f}")

print()
print("Extended Reporting Period (ERP) Pricing from a Year-7 Mature Policy (age 84):")
rdf_84 = rdfs_extended[ages_extended.index(84)]
erp_data = []
for age in [96, 108, 120]:
    rdf_age = rdfs_extended[ages_extended.index(age)]
    erp_rdf = rdf_age - rdf_84
    erp_pct = (age - 84) // 12
    erp_premium = erp_rdf * occurrence_rate
    erp_data.append({'ERP Type': f'{erp_pct}-year ERP', 'RDF Increment': f'{erp_rdf:.4f}', 
                     'ERP Premium': f'${erp_premium:.2f}'})
    
# Unlimited tail
pcf_84 = 1 - rdf_84
erp_data.append({'ERP Type': 'Unlimited tail', 'RDF Increment': f'{pcf_84:.4f}',
                 'ERP Premium': f'${pcf_84 * occurrence_rate:.2f}'})

display(pd.DataFrame(erp_data))
```

## Section 6: Transition Scenarios and Exam Traps

### Common Transition Scenarios

**Scenario 1: First-time CM buyer (occurrence → CM)**
- Prior carrier issued occurrence policy; no CM history
- New carrier writes Step 1 CM — charges only RDF(12) × occ_rate
- Rate jumps significantly in Year 2 and beyond as step factors > 1
- Insured must understand that CM rates increase until maturity

**Scenario 2: Switching CM carriers (mature CM → mature CM)**
- Prior carrier had mature CM; new carrier issues mature CM with same retroactive date
- Rate ≈ mature CM rate (no big change)
- Prior carrier has NO remaining obligation (all prior reports were covered on prior policy)
- New carrier takes on ALL future reports from ALL retroactive-date-eligible prior acts

**Scenario 3: Canceling CM coverage**
- All unreported claims from prior years are now uninsured unless tail coverage is purchased
- The tail policy covers the PCF × occurrence rate worth of future reports
- This can be very expensive for long-tail lines where 20–30% of claims may be unreported after 5 years

**Scenario 4: CM → occurrence conversion**
- The insured buys an occurrence policy going forward
- The occurrence policy covers new accident year claims AND future reports of already-occurred claims
- The tail on the CM policy is NOT needed (occurrence policy covers future reports of prior acts)

### Common Exam Traps

| Trap | Correct Approach |
|---|---|
| **Step 1 rate ≈ occurrence rate** | No — Step 1 rate is *much less* than occurrence because only a small % of claims are reported in year 1 |
| **Mature CM rate = occurrence rate always** | Only in a stable book with no trend; positive trend makes mature CM < occurrence |
| **Step factors are all < 1** | Only SF₁ < 1; SF₂ and beyond are > 1 (the CM rate grows each year from step 1 to maturity) |
| **Tail premium is large** | For long-tail lines, the tail can be very significant (>10% of the mature CM rate for med mal) |
| **Occurrence to CM conversion only needs Step 1** | No — the step factors must be tracked through the full maturation period until convergence |
