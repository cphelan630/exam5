## Fundamental Insurance Equation

### The Core Equation

The starting point for all ratemaking is the **Fundamental Insurance Equation**:

$$\text{Premium} = \text{Losses} + \text{LAE} + \text{UW Expenses} + \text{UW Profit}$$

Each component requires estimation:

| Component | Description |
|---|---|
| **Losses** | Amounts paid or owed to claimants under the policy |
| **LAE** (Loss Adjustment Expenses) | Expenses incurred settling claims: ALAE (claim-specific) + ULAE (general claims dept.) |
| **UW Expenses** | Policy acquisition and servicing costs: commissions, other acquisition, general expenses, taxes/licenses/fees |
| **UW Profit** | Target return for the insurer; typically set after accounting for investment income |

  
The goal of ratemaking is to set rates so that premium covers all expected costs (for future period) and achieves the target profit.  


---

### The Four CAS Ratemaking Principles

The CAS *Statement of Principles Regarding P&C Insurance Ratemaking* establishes four principles:

1. A rate is an estimate of the expected value of future costs. (**prospective**)
2. A rate provides for **all** costs associated with the transfer of risk.  
    * i.e. charges enough in total    
3. A rate provides for the costs associated with an **individual** risk transfer.   
    * i.e. charges the right amount to each insured  
4. A rate is reasonable and not excessive, inadequate, or unfairly discriminatory if it is an actuarially sound estimate of the expected value of all future costs associated with an individual risk transfer.

---

### Two Keys to Achieving Balance

**1. Ratemaking is prospective.**

Rates must cover *future* costs, not recoup past losses. Historical data is adjusted for any factor that causes past costs to differ from future expectations — rate changes, law changes, inflation, shifts in mix of business, and so on.

**2. Balance must be attained at both the aggregate and individual levels.**

- **Aggregate balance**: total premium across all policies is sufficient to cover total expected losses, expenses, and target profit.
- **Individual balance**: premium for each risk reflects the expected cost of *that specific risk*. A risk with a higher expected loss should pay a higher premium than a lower-risk insured.

Failure to achieve individual balance leads to **adverse selection**: competitors who better segment risks will attract the better risks away, leaving behind an increasingly poor-quality book. Eventually the insurer must either raise rates further or exit the market.

> **Formula sheet note** — The exam tests both the equation itself and the four principles. Principle 1 (prospective) directly supports the need for all the adjustments covered in this notebook.

## Addressing Large Events and Anomalies

Historical losses often contain extraordinary items — infrequent but very large individual losses, or entire accident years distorted by a catastrophic event. If included at face value, these items make indicated rates swing wildly from year to year. The standard approach is to **remove these extraordinary amounts from historical losses and replace them with a long-run average provision**.

---

### Shock Losses

**Shock losses** (also called large individual losses or excess losses) are individual claims that occur infrequently but are excessively large relative to the typical claim in the portfolio.

**Why they matter:** In a small portfolio, a single outsized claim can spike the indicated rate one year and drop it sharply the next. Including them without adjustment violates the goal of rate stability and prospective relevance.

#### Three ways to handle shock losses

| Approach | Description |
|---|---|
| **Exclude entirely** | Remove shock losses from historical data; add back a separately derived long-run provision |
| **Cap at basic limit** | Censor each loss at the minimum coverage limit (basic limit); the indication reflects only basic-limit exposure |
| **Cap at a threshold** | Cap at a percentile of the loss distribution or a percentage of the insured value |

> **Formula sheet note** — The exam most often tests the **excess loss factor** approach. Know the 4-step procedure cold.

#### 4-Step Excess Loss Provision

1. Sum the **excess losses** (the portion of each claim above the threshold) over the full experience period.
2. Sum the **non-excess losses** (each ground-up loss capped at the threshold) over the full experience period.
3. Calculate the **excess loss ratio** = Total Excess Losses / Total Non-Excess Losses.
4. The **excess loss factor** = 1 + excess loss ratio.

Apply this factor to each year's non-excess losses to produce a loss amount that incorporates the long-run expected shock loss.

#### Worked Example

Threshold = $500,000. Five-year experience period:

| Year | Non-Excess Losses | Excess Losses |
|---|---|---|
| 2019 | $9,500,000 | $700,000 |
| 2020 | $9,800,000 | $0 |
| 2021 | $10,300,000 | $1,200,000 |
| 2022 | $10,100,000 | $0 |
| 2023 | $9,800,000 | $600,000 |
| **Total** | **$49,500,000** | **$2,500,000** |

Excess loss ratio = $2,500,000 / $49,500,000 = **5.05%**

Excess loss factor = 1 + 0.0505 = **1.0505**

For ratemaking, 2023 losses = $9,800,000 × 1.0505 = **$10,295,000** — replacing the actual $600K shock loss with the long-run average expected excess ($9,800,000 × 5.05% ≈ $495,000).

> **Exam tip** — The excess loss factor is applied to **non-excess** losses, not total losses. Make sure you cap the losses before computing the excess.

---

### Catastrophe Losses

**Catastrophe losses** result from natural or man-made disasters that cause a *large number of claims* simultaneously — hurricanes, earthquakes, wildfires, winter storms, etc. They differ from shock losses, which are individual high-severity claims.

Like shock losses, actual catastrophe losses are removed from historical ratemaking data and replaced with a **long-run expected catastrophe provision**.

#### Non-Modeled Catastrophe Analysis

Used for perils that occur with some regularity over many years (e.g., hail storms for auto comprehensive):

1. Follow the same 4-step excess loss procedure, treating catastrophe losses as the "excess" component, **or**
2. Develop a **pure premium or loss ratio** for the catastrophe exposure using a long experience period (10–30 years).

The number of years used should balance **stability** (more years = more stable) with **relevance** (very old data may not reflect today's climate, values, or demographics).

#### Modeled Catastrophe Analysis

Used for infrequent, very high-severity events (major earthquakes, Category 5 hurricanes):

- A **stochastic catastrophe model** simulates thousands of scenarios to estimate expected annual losses from the peril.
- The modeled expected cat losses are added to the non-modeled cat provision to get a **total cat provision**.
- The modeled provision is generally added to non-cat losses that have already been trended and developed to ultimate — do not trend or develop the cat provision again.

---

### Reinsurance

**Reinsurance** is insurance purchased by the primary insurer to transfer some of its financial risk. There are two main types:

| Type | Description |
|---|---|
| **Proportional** | The same proportion of premium and losses is ceded to the reinsurer. Premium and loss effects are symmetric — no explicit adjustment needed in the ratemaking analysis. |
| **Non-proportional** | The primary insurer cedes a portion of premium; the reinsurer pays losses above a retention or within a layer. Premium and loss effects are asymmetric. |

**Non-proportional reinsurance** must be explicitly addressed in a ratemaking analysis. Two approaches:

1. **Reduce projected losses** for expected reinsurance recoveries, and reduce total premium by the cost (ceded premium) of reinsurance.
2. **Include the net cost** of reinsurance (reinsurance premium minus expected recoveries) as a separate expense item.

## Adjusting for One-Time Changes

Historical premiums and losses may reflect rate levels, benefit levels, or coverage definitions that differ from the current environment. These one-time changes must be corrected before using historical data to project future costs.

---

### Rate and Benefit Changes

**Rate changes** affect premium levels. **Benefit or coverage changes** (including those mandated by law) affect loss levels. Both can have two types of effects:

| Effect Type | Description |
|---|---|
| **Direct effects** | Obvious, quantifiable impacts on premium or losses resulting directly from the change |
| **Indirect effects** | Changes in claimant behavior that impact losses (harder to quantify, often estimated judgmentally) |

**Example:** A workers compensation statute raises the wage replacement rate from 60% to 65%. The direct effect on losses is +8.3% (= 65/60 − 1). The indirect effect — workers may have less incentive to return to work promptly under the higher benefit — requires actuarial judgment to estimate.

**Three approaches to calculate the direct effect of benefit/coverage changes on losses:**
1. Restate individual claims under the new benefit structure
2. Simulate losses under new conditions
3. Use aggregate loss distributions

---

### Adjusting Historical Premium to Current Rate Level (On-Level Premium)

The goal is to restate historical earned premium as if it had been written at the **current rate level** — making premium from different years directly comparable for ratemaking.

#### Extension of Exposures Method

1. Re-rate every historical policy using the current rates and current rating characteristics.
2. Recalculate earned premium for each historical period using the updated rates.

This is the **most accurate** approach, but requires detailed policy-level data and significant computing resources. Assumptions must also be made for any new rating variables not present in older data.

#### Parallelogram Method (Geometric Method)

A practical approximation that assumes policies are written **uniformly throughout each period**.

**Steps:**
1. Determine the timing and magnitude of each rate change in the historical experience period.
2. Calculate the **cumulative rate level index** for each rate level group. Start at 1.000; multiply by each successive rate change factor as changes take effect.
3. For each historical period, calculate the **proportion of earned premium** in each rate level group using geometric shapes in a policy-year / calendar-year diagram.
4. Calculate the **weighted average cumulative rate level index** for the historical period.
5. Calculate the **on-level factor**:

$$\text{On-Level Factor} = \frac{\text{Current Cumulative Rate Level Index}}{\text{Weighted Avg Rate Level Index for Historical Period}}$$

6. Multiply each historical period's earned premium by its on-level factor.

**Limitations:**
- Assumes uniform policy writings throughout the period (may not hold for seasonal lines).
- Rate changes are applied at the overall average — class-level variation is ignored.

*If writings are not uniform, shorten the sub-periods to months or quarters, or calculate the actual distribution of policy writings.*

#### Worked Example — Parallelogram Method

Suppose rates changed as follows:

| Effective Date | Rate Change | Cumulative Index |
|---|---|---|
| (base) | — | 1.000 |
| Jan 1, 2022 | +5% | 1.050 |
| Jul 1, 2023 | +8% | 1.134 |

Current rate level index = **1.134**.

For Calendar Year 2022 (annual policies, uniform writings):
- Portion written before Jan 1, 2022: none (the year starts at the change date in this simplified example)
- Using the parallelogram geometry, the weighted avg index for CY 2022 earned premium = weighted average of the 1.000 and 1.050 levels.
- On-level factor = 1.134 / (weighted avg) — this brings 2022 premium to today's rate level.

> **Formula sheet note** — Exam questions often give you a rate change history and ask you to calculate the on-level factor for a specific accident or calendar year. The key is correctly computing the weighted average index using the parallelogram proportions (areas = fractions of earned premium at each rate level).

---

### Adjusting Historical Losses for Benefit Changes

The same parallelogram geometry applies to losses when benefit or coverage changes have occurred. Instead of an on-level factor, calculate a **benefit change loss adjustment factor**:

$$\text{Benefit Change Loss Adj. Factor} = \frac{\text{Current Loss Level}}{\text{Avg Loss Level of Historical Period}}$$

The dividing line between pre-change and post-change losses depends on exactly how the law is written:
- If the change affects all **claims with accident dates** after a certain date → the dividing line is a horizontal cut through the AY diagram.
- If the change affects all **policies written** after a certain date → the dividing line is a diagonal cut (same geometry as the premium parallelogram).
- If the change affects all **open claims** regardless of date → appears on a diagonal, potentially distorting development patterns.

> **Exam tip — double-counting:** If you restate historical losses to the new benefit level *and* your trend is estimated from data already on the new benefit level, no further benefit adjustment is needed. Applying both an explicit benefit factor *and* a trend derived from post-change data will double-count the change.

## Loss Development in Ratemaking

> **Note:** The mechanics of loss development triangles, age-to-age factors, and the chain ladder method are covered in depth in the reserving review notebooks. This section focuses on *why* and *how* development is applied within the ratemaking process.

---

### Why Development Is Needed

At the time of a ratemaking analysis, the most recent accident years are **immature** — claims are still open, new claims may still be reported (IBNR), and case reserves on known claims will change as more information emerges.

Ratemaking requires **estimated ultimate losses** for each accident year — the final amount the insurer expects to pay once all claims are settled. Using reported losses without developing them to ultimate would systematically understate the cost of recent years and distort the indicated rates.

The **chain ladder method** (development technique) is the most common approach: link ratios from older, more mature accident years are selected and applied to the current diagonal to project each accident year to ultimate. See the reserving review notebooks for full details.

---

### Key Ratemaking Considerations

#### Extraordinary losses distort development patterns

Shock losses and catastrophe losses must be **removed from the data before measuring development patterns**. Their effect on link ratios depends on when they enter the data:

- **Reported immediately (in the accident year):** Inflates aggregate losses at the first valuation. This may dampen the 12→24 month link ratio for that year, because the reported amount at 12 months already looks unusually high relative to what is seen at 24 months.
- **Reported with a lag (after the accident year closes):** Creates a large jump from one diagonal to the next, inflating the link ratio for that development period in that accident year.

Either way, if not removed, the resulting selected development factors and ultimate projections will be unreliable.

#### Benefit changes can shift absolute loss levels

- A benefit change affecting all **new accident dates** after a given date will change the absolute level of losses in new accident year rows. The development *pattern* (link ratios) may be unaffected if all losses in a row are uniformly impacted.
- A benefit change affecting all **open claims** (regardless of accident year) as of a given date will appear on a diagonal, distorting the link ratios for that development period across multiple accident years.

#### Short-tailed vs. long-tailed lines

| Line Type | Examples | Development Significance |
|---|---|---|
| **Short-tailed** | Auto physical damage, homeowners property | Claims settle quickly; the most recent AY is near ultimate within 12–18 months. Development adjustment is minor. |
| **Long-tailed** | Commercial general liability, workers comp, medical malpractice | Claims settle over many years. Development adjustments are large and carry significant actuarial uncertainty. |

---

### Output: Estimated Ultimate Losses

The result of applying selected development factors to the latest diagonal is **estimated ultimate losses by accident year** — the $\hat{L}$ (projected loss) used in the pure premium and loss ratio indication formulas.

> **Overlap fallacy:** Applying development factors *and* trend factors does not double-count inflation. Development brings immature losses to their ultimate level on claims that *have already occurred*. Trend then adjusts those ultimate losses to the cost environment of the *future policy period*. They address completely different time intervals.  TLDR: You should apply **both** trends and development to reflect future cost environment.

## Trends

**Loss trend** is the systematic change in frequency, severity, or pure premium over time due to economic, social, legal, or behavioral forces. Because rates will be in effect in the *future*, historical losses must be projected forward to the expected cost level during the future policy period.

---

### Drivers of Loss Trend

| Driver | Examples |
|---|---|
| **Monetary / medical inflation** | Repair costs, medical procedure costs rising with CPI or medical-specific indices |
| **Social inflation** | Larger jury awards, increased attorney involvement, higher litigation rates |
| **Utilization changes** | Higher utilization of medical treatment, increased claims consciousness |
| **Distributional shifts** | Portfolio shifting toward higher- or lower-risk insureds |
| **Safety / technology** | ADAS features reducing auto collision frequency; improved building codes |

Frequency and severity are typically analyzed **separately** because they can move in different directions and for different reasons. For example, higher deductibles will reduce frequency (small claims go unreported) while increasing severity (only large claims are reported).

---

### Trend Analysis Methods

#### Exponential Trend (most common for insurance losses)

An exponential curve fit produces a **constant annual percentage change**. This is preferred because:
- Severity and frequency cannot be negative (a linear model can eventually project below zero for declining trends)
- Insurance cost inflation tends to compound

The exponential model is fit to historical data points via regression. The statistical fit is assessed using **R²** — the proportion of total variance explained by the model.

**Combining frequency and severity trends into a pure premium trend:**

$$\text{Pure Premium Trend} = (1 + \text{Frequency Trend}) \times (1 + \text{Severity Trend}) - 1$$

*Example: −1% annual frequency trend and +2% annual severity trend → pure premium trend = (0.99)(1.02) − 1 = **+0.98%***

#### Linear Trend

A linear fit produces a **constant annual dollar change**. For a linear trend, the projected adjustment is:

$$\text{Dollar Adjustment} = \text{Annual Dollar Trend} \times \text{Trend Period (years)}$$

Linear trends can project negative values when the trend is negative — a shortcoming that limits their use for frequency and severity (which cannot be negative in practice).

#### Practical considerations
- Catastrophe losses should be **excluded** from trend data to avoid distortion.
- Shock losses may be removed or the actuary may analyze trends on basic limits losses.
- If internal data is too sparse or volatile, external indices (CPI, medical components, Bureau of Labor Statistics wage data) can supplement the analysis.

---

### Trend Period Calculation

The **trend period** is the length of time from the midpoint of the historical experience period to the midpoint of the forecast period.

| Term | Definition |
|---|---|
| **Experience period** | The historical period from which loss data is drawn (e.g., Accident Year 2022) |
| **Forecast period** | The future period in which the new rates will be in effect |
| **"Trend from" date** | Midpoint of the experience period losses (for AY data: July 1 of the accident year) |
| **"Trend to" date** | Midpoint of the forecast period losses |

**For a 12-month policy, annual rate effective date, rates in effect for one year:**
- Policies written Jan 1–Dec 31 cover claims through Dec 31 of the following year → 2-year window
- Midpoint of forecast period = December 31 of the year rates take effect

**Example:** AY 2022 losses, rates effective January 1, 2026, in effect for one year (annual policies):
- "Trend from": July 1, 2022
- "Trend to": December 31, 2026
- **Trend period: 4.5 years**
- **Trend factor**: $(1 + t)^{4.5}$ where $t$ is the selected annual pure premium trend

> **Formula sheet note** — Trend period length depends on: (1) type of data (AY vs. PY), (2) policy term (6-month vs. 12-month), (3) how long rates are expected to be in effect. For **6-month policies**, the forecast window is only 18 months, so the "trend to" date is 3 months earlier than for 12-month policies with the same effective date.

---

### Earned vs. Written Premium: Effect on Trend Dates

The data basis — whether losses are organized by **accident date** (paired with earned premium) or by **written/policy date** (paired with written premium) — changes where the "trend to" date falls. The "trend from" date is numerically the same either way for a given calendar year, but the "trend to" date is not.

#### "Trend from" date

| Data Basis | "Trend From" Date | Reasoning |
|---|---|---|
| **Accident Year (earned premium)** | July 1 of the accident year | Average accident date for losses earned uniformly over a CY |
| **Policy Year (written premium)** | July 1 of the policy year | Average *written* date for policies issued uniformly over a CY |

For the same calendar year of data, these two dates are equal — the distinction does not affect the "trend from" date in practice.

#### "Trend to" date — where the difference emerges

The "trend to" date is the midpoint of the **forecast period**, and the forecast period is defined differently depending on the data basis.

**Written premium basis:**
The forecast period covers all policies that will be *written* under the new rates. For rates effective Jan 1, YYYY, in effect for one year with uniform writings:
- Writing window: Jan 1, YYYY → Dec 31, YYYY
- Average written date → **July 1, YYYY**

**Earned premium (accident year) basis:**
The forecast period covers all *accidents* that will occur on new-rate policies. For 12-month policies written uniformly throughout year YYYY:
- Average written date = July 1, YYYY
- Each policy earns for 12 months, so the average accident midpoint = July 1, YYYY + 6 months → **January 1, YYYY+1**

The "trend to" date is **6 months later** on an earned premium basis than on a written premium basis (for 12-month policies). For **6-month policies**, the difference shrinks to **3 months**.

#### Side-by-side example

Same scenario as above: AY/PY 2022 data, rates effective January 1, 2026, in effect for one year:

| Basis | "Trend From" | "Trend To" | Trend Period |
|---|---|---|---|
| **Earned (AY)** | July 1, 2022 | January 1, 2027 ≈ Dec 31, 2026 | **4.5 years** |
| **Written (PY)** | July 1, 2022 | July 1, 2026 | **4.0 years** |

The written premium trend period is shorter because writing precedes earning — you write a policy before you earn the premium on it. The accident (and therefore the loss cost) occurs somewhere in the middle of the policy term, so it lags the written date by half the policy term.

> **W&M reference** — This distinction is covered in the Trending chapter. The general rule: *trend period = (trend-to date) − (trend-from date)*, where trend-to is the average **accident** date in the forecast period for earned/AY data, and the average **written** date in the forecast period for written/PY data.

---

### One-Step vs. Two-Step Trending

#### One-Step Trending

A single trend rate is applied to adjust historical data directly to the forecast period. This is the standard approach when the trend is expected to be **consistent across both the historical and future periods**.

#### Two-Step Trending

Used when the **historical trend rate is expected to differ from the future trend rate** — for example, when a recent law change has distorted the historical trend, or when the actuary believes the underlying trend is shifting.

**Step 1 — Current trend step:** Adjust each historical data point from the experience period midpoint to the average date of the *latest available trend data*. This brings everything to a common "current" base.

**Step 2 — Projected trend step:** Adjust from the latest data average date to the forecast period midpoint using the selected *future* trend rate.

$$\text{Total Trend Factor} = \text{Current Trend Factor} \times \text{Projected Trend Factor}$$

---

### Overlap Fallacy

A common misconception: *"Developing losses to ultimate and then trending them double-counts inflation."*

**This is false.** The two adjustments address completely different time intervals:

- **Loss development** adjusts immature losses to their expected *ultimate* value. It accounts for payments that will be made in the future on claims that *have already occurred* — all before the rate effective date.
- **Loss trend** adjusts those ultimate losses from the historical cost level to the *future* cost level expected during the period the rates will be in effect.

The development period ends around the current valuation date. The trend period begins at the historical accident date midpoint and ends at the future forecast midpoint. There is no overlap.

> **Formula sheet note** — The overlap fallacy is a named concept explicitly tested on the exam. The correct answer is always that development and trend do not overlap.

---

### Leveraged Effect of Limits on Severity Trend

When losses are subject to a **per-occurrence limit**, the impact of a severity trend differs by layer. For a **positive** severity trend:

$$\text{Basic Limits Trend} \leq \text{Total Limits Trend} \leq \text{Excess Limits Trend}$$

For a **negative** severity trend, the inequality reverses:

$$\text{Basic Limits Trend} \geq \text{Total Limits Trend} \geq \text{Excess Limits Trend}$$

**Intuition:** A positive trend pushes some claims above the basic limit for the first time — the full increase is realized in the excess layer but only a partial increase is captured in the basic limit layer. Claims already above the limit see no change in the basic-limit amount at all.

#### Quick Example

Basic limit = $25,000. A $24,000 claim increases by 10% to $26,400.

| Layer | Before | After | Change |
|---|---|---|---|
| Total limits | $24,000 | $26,400 | **+10.0%** |
| Basic limits (capped at $25,000) | $24,000 | $25,000 | **+4.2%** |
| Excess losses | $0 | $1,400 | *undefined (from zero)* |

The 10% total-limits trend is damped to only 4.2% in the basic-limits layer for this claim.

> **Exam tip** — When trending data subject to per-occurrence limits, either (1) use basic limits loss data for the trend analysis, or (2) apply a limits adjustment to convert from total limits to basic limits trend before applying to basic limits ratemaking data.

## Expenses and Profit Load

After projecting ultimate losses and trending them to the future policy period, the actuary must determine provisions for **loss adjustment expenses, underwriting expenses, and profit** to complete the fundamental insurance equation.

---

### Loss Adjustment Expenses (LAE)

#### Allocated LAE (ALAE)

ALAE is directly attributable to specific claims (e.g., outside legal defense fees, expert witness costs). Two treatment options in ratemaking:

1. **Include ALAE with losses** — develop and trend ALAE together with losses. Simpler; assumes ALAE tracks loss development in the same pattern.
2. **Study ALAE separately** — develop and trend ALAE independently. More accurate if ALAE patterns differ from loss patterns (e.g., if attorney involvement is growing, ALAE may trend faster than losses).

#### Unallocated LAE (ULAE)

ULAE (e.g., claims department salaries, overhead) cannot be assigned to a specific claim. The standard ratemaking approach:

1. Calculate the ratio of **calendar year paid ULAE** to **calendar year paid loss + ALAE** for several historical years.
2. Select a ratio based on the historical pattern.
3. Apply the selected ratio to each accident year's **reported loss + ALAE** to estimate the ULAE provision.

*Alternative approaches: count-based allocation methods (allocating ULAE by claim count), or time studies of how claims adjusters spend their time.*

---

### Underwriting Expenses

Underwriting (UW) expenses fall into four categories, each typically measured against different premium bases:

| Category | Nature | Typical Basis |
|---|---|---|
| **General expenses** | Fixed — roughly constant per policy regardless of size | Earned premium (or exposure/policy counts) — countrywide |
| **Other acquisition** | Variable — scale with premium | Written premium (or exposure/policy counts) — countrywide |
| **Commissions & Brokerage** | Variable | Written premium (or exposure/policy counts) — countrywide/statewide |
| **Taxes, Licenses & Fees** | Variable | Written premium — statewide |

**Fixed expenses** are roughly constant per policy (e.g., billing costs, policy issuance, certain overhead). **Variable expenses** scale proportionally with premium (e.g., commissions as a % of premium written). The distinction matters because fixed expenses should not be loaded as a flat percentage of premium — that would undercharge large risks and overcharge small ones.

---

### Three Expense Projection Methods

#### 1. All Variable Expense Method

All UW expenses are treated as a constant percentage of premium.

**Steps:**
1. Calculate the expense ratio for each year and expense category.
2. Select a ratio for each category.
3. Sum selected ratios to get a total UW expense provision.

**Limitation:** When some expenses are truly fixed per policy (not proportional to premium size), this method undercharges risks with below-average premium and overcharges risks with above-average premium. A $1M commercial policy should not be charged 10× the billing cost of a $100K policy just because premiums scale 10×.

#### 2. Premium-Based Projection Method

Fixed and variable expenses are treated separately, but both are expressed as ratios to premium.

**Steps:**
1. Derive expense ratios for each year and category.
2. Select a ratio for each type.
3. Split each selected ratio into fixed and variable components.
4. Sum fixed ratios for a total fixed expense provision; sum variable ratios for a variable provision.

If needed, convert the fixed expense ratio to a per-exposure amount:
$$\text{Fixed Expense per Exposure} = \text{Fixed Expense Ratio} \times \text{Projected Average Premium}$$

**Limitation:** If historical and projected average premium levels differ materially, the implied fixed dollar amount per policy will be distorted (because fixed dollars are being divided by different premium denominators).

#### 3. Exposure/Policy-Based Projection Method

Variable expenses are related to premium; fixed expenses are related to exposure or policy count.

**Steps:**
1. Split each expense into fixed and variable components.
2. Calculate expense ratios: divide fixed expenses by exposures (or policy counts); divide variable expenses by premium. Trend fixed per-exposure amounts if needed.
3. Select a ratio for each type.
4. Sum fixed per-exposure selections for the total fixed provision; sum variable ratios for the variable provision.

This method properly recognizes that fixed expenses grow with the *number of policies*, not with premium levels, avoiding the distortion problem of the premium-based method.

This method can be enhanced by:
- Improving the methodology for splitting expenses into fixed and variable
- Collecting data at a finer level (e.g., by state rather than countrywide)
- Accounting for expenses that vary by business characteristics (new vs. renewal, large vs. small account)
- Reflecting economies of scale

---

### Underwriting Profit Provision

$$\text{Total Profit} = \text{Investment Income} + \text{UW Profit}$$

- **Investment income** arises from investing policyholder-supplied funds (unearned premium reserves and loss reserves) and shareholder capital.
- **Underwriting profit** is set to achieve the **target total return** after investment income has been accounted for. If the insurer earns significant investment income, the target UW profit may be small or even negative.

> In practice, the exact derivation of the target UW profit in consideration of investment income and return on equity is beyond the scope of CAS Exam 5.

---

### PLR and VPLR

Two key permissible loss ratios appear in the overall indication formulas:

$$\text{VPLR (Variable Permissible Loss Ratio)} = 1 - V - Q'$$

$$\text{PLR (Total Permissible Loss Ratio)} = 1 - F - V - Q'$$

where:
- $F$ = fixed UW expense ratio (fixed expenses / premium)
- $V$ = variable UW expense ratio (variable expenses / premium)
- $Q'$ = target UW profit provision (% of premium)

**VPLR** represents the share of each premium dollar available to pay losses, LAE, *and* fixed expenses.
**PLR** represents the share available to pay losses and LAE only (after all expense provisions are subtracted).

*Example:* If $F = 5\%$, $V = 25\%$, $Q' = 5\%$:
$$\text{VPLR} = 1 - 0.25 - 0.05 = 0.70 \quad \text{PLR} = 1 - 0.05 - 0.25 - 0.05 = 0.65$$

## Overall Rate Indications

An **overall rate indication** answers the question: at the current average rate level, is the existing premium adequate, excessive, or inadequate? After all the adjustments covered in prior sections — development, trend, large loss provisions, on-level premium, benefit changes — the actuary combines everything into a single indication using one of two equivalent methods.

---

### Pure Premium Method

The pure premium method derives the **indicated average rate per exposure** directly from projected costs:

$$\boxed{\text{Indicated Avg Rate} = \frac{\hat{L} + \hat{E}_F + \hat{E}_V}{1 - V - Q'}}$$

where:
- $\hat{L}$ = projected losses + LAE per exposure (developed to ultimate, trended to the forecast period, adjusted for large losses and benefit changes)
- $\hat{E}_F$ = projected fixed expenses per exposure
- $\hat{E}_V$ = projected variable expenses per exposure (= $V \times \text{Indicated Rate}$, embedded in the denominator)
- $V$ = variable expense ratio
- $Q'$ = target UW profit provision

The **indicated rate change** = Indicated Avg Rate / Current Avg Rate − 1.

This method does **not** require on-level premium — it works directly from loss and exposure data.

---

### Loss Ratio Method

The loss ratio method derives the **indicated percent change** in the current rate level:

$$\boxed{\text{Indicated Avg Rate Change} = \frac{\text{Projected Loss \& LAE Ratio} + F}{1 - V - Q'} - 1}$$

where the Projected Loss & LAE Ratio uses **on-level earned premium** as the denominator. This is why the loss ratio method requires historical premium to be adjusted to the current rate level first.

The output is a multiplicative rate change factor (subtract 1 to get the percentage change).

---

### Comparison of the Two Methods

| Feature | Pure Premium Method | Loss Ratio Method |
|---|---|---|
| **Loss measure** | Pure premium (losses per exposure) | Loss ratio (losses / on-level premium) |
| **Output** | Indicated average rate | Indicated average rate change |
| **Requires on-level premium?** | No | Yes |
| **Requires earned exposures?** | Yes | Not directly |
| **Results equivalent?** | Yes — when consistent inputs and assumptions are used across both |

Neither method is inherently superior. The choice depends on data availability and how the problem is structured. On the exam, either may be given or required.

> **Formula sheet note** — The methods produce identical results when the same data and assumptions are used. The most common exam error is using unadjusted (not on-level) premium in the loss ratio denominator. Always check whether the premium provided is already on-level or needs adjustment.

---

### Worked Example

**Given data (after all adjustments):**

| Item | Value |
|---|---|
| Projected ultimate losses + LAE | $9,500,000 |
| Earned exposures | 100,000 |
| On-level earned premium | $12,000,000 |
| Fixed expense ratio ($F$) | 5% |
| Variable expense ratio ($V$) | 25% |
| Target UW profit ($Q'$) | 5% |

**Pure Premium Method:**

- Projected pure premium = $9,500,000 / 100,000 = **$95.00 per exposure**
- Current average premium = $12,000,000 / 100,000 = $120.00
- Fixed expenses per exposure = 5% × $120.00 = **$6.00**

$$\text{Indicated Avg Rate} = \frac{95.00 + 6.00}{1 - 0.25 - 0.05} = \frac{101.00}{0.70} = \mathbf{\$144.29}$$

Indicated rate change = $144.29 / $120.00 − 1 = **+20.2%**

**Loss Ratio Method:**

- Projected loss & LAE ratio = $9,500,000 / $12,000,000 = **79.2%**

$$\text{Indicated Rate Change} = \frac{0.792 + 0.05}{1 - 0.25 - 0.05} - 1 = \frac{0.842}{0.70} - 1 = 1.203 - 1 = \mathbf{+20.3\%}$$

Both methods agree (minor rounding difference). Rates are currently inadequate by approximately **20%** — the indicated change is a significant increase.

> **Interpretation:** The projected loss & LAE ratio of 79.2% far exceeds the PLR of 65% (= 1 − 0.05 − 0.25 − 0.05). This gap drives the indication. The PLR is the maximum loss ratio consistent with covering all costs and hitting the profit target — any ratio above it means rates need to increase.
