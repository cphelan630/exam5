Accurate, well-organized data is the foundation of every actuarial analysis. Before rates can be set or reserves estimated, an actuary must understand where data comes from, how it is structured, and what each figure represents. This notebook covers the foundational insurance data concepts that underpin both ratemaking and reserving — the "basics" that every CAS Exam 5 candidate must know cold.

**Learning goals:**
1. Describe the policy lifecycle and identify the exposure types used in ratemaking
2. Explain the differences between written, earned, unearned, and in-force premium and exposures
3. Understand how data is aggregated (calendar year, accident year, policy year, report year) and the trade-offs of each method
4. Describe the claims lifecycle and the progression from paid losses to case reserves to incurred to ultimate
5. Explain the structure of insurer accounting data — the annual statement, SAP vs. GAAP, income statement, and balance sheet
6. Calculate and interpret key insurer performance metrics: combined ratio, operating ratio, and related financial measures

## Formula Sheet Quick Reference

| Concept | Formula / Definition |
|---|---|
| **Frequency** | Claims / Exposures |
| **Severity** | Losses / Claims |
| **Pure Premium** | Losses / Exposures = Frequency × Severity |
| **Loss Ratio** | Losses / Earned Premium = Pure Premium / Avg Premium |
| **LAE Ratio** | LAE / Losses |
| **UW Expense Ratio** | UW Expenses / Written Premium |
| **Combined Ratio** | Loss Ratio + LAE Ratio + UW Expense Ratio |
| **Operating Ratio** | Combined Ratio − Investment Income Ratio |
| **Investment Income Ratio** | Net Investment Income / Earned Premium |
| **Premium-to-Surplus Ratio** | Net Written Premium / Policyholder Surplus |
| **Retention Ratio** | Policies Renewed / Potential Renewal Policies |
| **Close Ratio** | Accepted Quotes / Total Quotes |
| **EOY Unearned Exposure** | Written Exposure − Earned Exposure + BOY Unearned Exposure |
| **EOY Unearned Premium** | Written Premium − Earned Premium + BOY Unearned Premium |
| **Reported (Case Incurred) Loss** | Paid Losses + Case Reserve |
| **IBNER** | Estimated Ultimate on Reported Claims − Aggregate Reported Losses at Valuation Date |
| **Ultimate Losses** | Reported Losses + IBNR |
| **Total Reserve** | Case Reserves + IBNR (Pure) + IBNER |

## Insurance Company Overview

### Lines of Business

Property and casualty insurance is divided into **personal lines** (coverage for individuals and families) and **commercial lines** (coverage for businesses). Lines are also characterized by how quickly claims are reported and settled.

| Category | Examples | Typical Development |
|---|---|---|
| **Personal Lines** | Private passenger auto, homeowners, personal umbrella | Short to medium tail |
| **Commercial Lines** | Commercial auto, general liability, workers comp, commercial property, professional liability | Medium to long tail |
| **Short-tailed** | Auto physical damage, homeowners property, inland marine | Most claims settle within 12–18 months of the accident |
| **Long-tailed** | Medical malpractice, workers comp, general liability, D&O | Claims can develop over 10+ years |

The tail length matters enormously for reserving (longer development → more uncertainty) and for investment income (longer reserves stay invested → more float income).

---

### Insurance Company Departments and Data Flow

Each department in an insurance company generates data that the actuary must understand and use correctly:

| Department | Primary Function | Key Data Produced |
|---|---|---|
| **Underwriting** | Evaluate and accept/reject risks; set pricing guidelines | Policy characteristics, rating variables, coverage terms |
| **Policy Issuance / Administration** | Issue policies, process endorsements and cancellations | Written premium, exposure, policy counts |
| **Claims** | Investigate, reserve, and settle claims | Paid losses, case reserves, claim counts, closure dates |
| **Accounting** | Financial reporting, reserving to regulators | Calendar year income, balance sheet, statutory filings |
| **Actuarial** | Rate setting, reserve estimation, financial modeling | Uses all of the above |

Breakdowns in data consistency between departments are a primary source of actuarial data quality problems.

---

### Occurrence vs. Claims-Made Policies

One of the most important structural distinctions in insurance data is whether a policy is written on an **occurrence** or **claims-made** basis:

| Feature | Occurrence Policy | Claims-Made Policy |
|---|---|---|
| **Trigger** | Loss *occurs* during the policy period, regardless of when claimed | Claim must be *filed* during the policy period |
| **IBNR tail** | Long — claims can emerge years after the accident | Short — once the policy expires, no new claims attach |
| **Common lines** | Most P&C lines, including GL and auto | Medical malpractice, D&O, E&O, cyber (increasingly) |
| **Retroactive date** | N/A | First inception date; claims arising before this date are excluded |
| **Tail coverage** | Not needed | "Extended reporting period" endorsement needed if switching insurers |

> **Exam tip** — Occurrence policies are far more common. Medical malpractice and professional liability are the classic exam examples of claims-made lines. The key impact on actuarial work: claims-made data has a *truncated* IBNR tail, making development factors lower and reserves less uncertain than occurrence data for the same line. In other words, claims-made exists to reduce the pricing risk related to the report lag on long-tail lines.

## Policy Data and the Policy Lifecycle

### The Policy Lifecycle

A policy passes through several stages, each generating data that actuaries use:

1. **Application** — Prospective insured submits risk information; rating variables are recorded
2. **Underwriting** — Insurer evaluates the risk; accepts, declines, or modifies terms
3. **Policy Issuance** — Policy is bound and issued; **written premium** and **written exposure** are recorded
4. **In-Force Period** — Policy is active; coverage is being provided; **earned premium** accumulates
5. **Mid-Term Changes** — Endorsements, cancellations, or audits alter the original policy
6. **Expiration / Renewal** — Policy expires; insurer offers renewal; the **retention ratio** reflects how many policyholders renew

**Mid-term changes** that affect premium and exposure data:

| Change | Description | Data Impact |
|---|---|---|
| **Endorsement** | Modifies coverage, limits, or rating info mid-term | Adjusts written premium; changes future earned premium |
| **Flat cancellation** | Policy voided as if never issued | Removes written premium entirely |
| **Pro-rata cancellation** | Policy terminated; premium refunded proportionally to unused time | Reduces written premium by the unearned portion |
| **Short-rate cancellation** | Insured-initiated cancellation; refund less than pro-rata (penalty) | Reduces written premium, but insurer retains more than pro-rata |
| **Audit** | Final premium adjusted based on actual exposure (payroll, sales) | Creates additional written premium or return premium after policy expires |

---

### Exposure Bases

An **exposure base** is the unit of risk to which a rate is applied. The rate × exposure = the policy premium. Selecting the right exposure base is a foundational actuarial decision.

**Three criteria for a good exposure base:**
1. **Proportional to expected loss** — As the exposure unit increases, expected loss should increase proportionally. A risk with 2× the exposures should expect roughly 2× the losses.
2. **Practical to measure** — The exposure must be observable, verifiable, and collectable at policy issuance (or after audit).
3. **Difficult for the insured to manipulate** — The insured should not be able to reduce their exposure unit without actually reducing their risk.

**Common exposure bases by line of business:**

| Line of Business | Exposure Base | Rationale |
|---|---|---|
| Private passenger auto liability | Car-years (vehicle-years) | Each insured vehicle faces road exposure independently |
| Homeowners | House-years | Each insured home is an independent risk unit |
| General liability | Payroll, sales, square footage, or units | Scales with the level of business activity generating loss |
| Workers compensation | Payroll ($100 of payroll) | Injury exposure scales with the number of hours worked |
| Commercial auto | Vehicle-years | Same as personal auto |
| Medical malpractice | Physician-years (by specialty) | Each physician is an independent risk unit |
| Products liability | Sales revenue | More product sold → more exposure to product liability claims |
| Inland marine | Value of property | Expected loss scales directly with the value at risk |

> **Exam tip** — The exam may ask you to evaluate whether a proposed exposure base meets the three criteria. The most common failure is the first criterion: for example, using *number of policies* as an exposure base for workers comp ignores that a large employer has far more exposure than a small one.

---

### Written, Earned, Unearned, and In-Force

These four measures describe *how much* premium or exposure is associated with a given period, depending on what "point in time" question you are answering:

| Measure | Timing | Definition |
|---|---|---|
| **Written** | When the policy is *issued* | Total premium/exposure of all policies issued during the period |
| **Earned** | As coverage is *provided* | Portion of written for which coverage has already been delivered as of the valuation date |
| **Unearned** | Remaining coverage not yet provided | Written minus earned; the insurer's obligation to provide future coverage |
| **In-Force** | At a specific *point in time* | Full-term premium/exposure for all policies currently active at that moment |

**Accounting identity:**

$$\text{EOY Unearned} = \text{Written} - \text{Earned} + \text{BOY Unearned}$$

This works for both premium and exposures.

**Earning pattern:** For most lines, premium earns *pro-rata* (uniformly over the policy term). A 12-month policy written January 1 earns 1/12 of its premium per month. A 12-month policy written July 1 earns 6/12 by year-end.

**Worked example — earning and unearned premium:**

A company writes the following annual ($1,200 premium each) policies in 2024:
- Policy A: Written January 1, 2024 — expires December 31, 2024
- Policy B: Written July 1, 2024 — expires June 30, 2025

As of December 31, 2024:

| Policy | Written | Earned (as of 12/31/24) | Unearned |
|---|---|---|---|
| A | $1,200 | $1,200 (12/12) | $0 |
| B | $1,200 | $600 (6/12) | $600 |
| **Total** | **$2,400** | **$1,800** | **$600** |

**In-force premium** as of December 31, 2024: only Policy B is still active → in-force = $1,200 (full annual premium).

---

### Premium Development for Auditable Lines

For lines where the final premium depends on actual experience measured *after* policy inception (workers compensation, payroll-based general liability), the **written premium at inception** is an estimate based on projected payroll or receipts. After the policy expires, the insurer conducts an **audit** to determine the actual exposure.

- If actual payroll > estimated payroll → **additional (audit) premium** is billed
- If actual payroll < estimated payroll → **return premium** is issued

Because audit premiums are not recorded until after the policy period, the written premium for recent policy years is still developing. Actuaries must **develop written premium to ultimate** before using it in a loss ratio ratemaking analysis — otherwise the denominator is understated and the loss ratio appears inflated.

> **Exam tip** — Premium development (also called "on-leveling for audit") is an additional complication beyond the parallelogram rate-level adjustment. For workers comp ratemaking, always check whether the written premium needs to be developed before applying an on-level factor. The on-level and audit-development adjustments are multiplicative and both necessary.

## Data Aggregation Methods

When organizing premium and loss data for ratemaking or reserving, every transaction must be assigned to a time period. The four standard methods differ in *which date* is used to assign a record to a period. Each creates different matching properties between premium and losses.

### Summary Comparison

| Method | Grouping Criterion | Premium-Loss Match | Development Needed? | Primary Uses |
|---|---|---|---|---|
| **Calendar Year** | Transaction date falls in the CY | Poor | No — data is final at year-end | Statutory accounting, financial reporting |
| **Accident Year** | Date of loss/accident falls in the AY | Good | Yes | Most ratemaking analyses, most reserving |
| **Policy Year** | Date policy is written falls in the PY | Best (perfect) | Yes — longest to mature | Underwriting analysis, pricing audits |
| **Report Year** | Date claim first reported falls in the RY | Moderate | Less than AY | Claims-made lines, some GL reserving |

---

### Calendar Year (CY)

All cash transactions (premium payments received, loss payments made) and reserve movements (changes in case reserves) that occur during a given 12-month calendar period are grouped together, regardless of when the accident happened or when the policy was written.

- **Calendar year data is final** at year-end: no future activity will be attributed to a past CY. This is why CY data requires no development.
- **Disadvantage**: A CY includes payments and reserve changes on claims from *many* different accident years simultaneously. You cannot study the loss development pattern for a specific group of risks using CY data alone.
- **Primary use**: Statutory financial reporting. The annual statement filed with regulators is on a CY basis.

---

### Accident Year (AY)

Losses are grouped by the year in which the loss *event* (accident) occurred. All payments and reserve changes for claims arising from AY 2022 accidents, regardless of when those payments are made, are assigned to AY 2022.

- **Advantage**: Premium earned on AY 2022 policies is a good match for AY 2022 losses — both relate to risk exposed during 2022.
- **Advantage**: Allows identification of accident-year trends (e.g., 2022 had an unusually bad cat season).
- **Disadvantage**: At any valuation date, recent accident years are immature — their ultimate losses are unknown and must be estimated using development factors.
- **Primary use**: The default method for most P&C ratemaking and reserving analyses.

---

### Policy Year (PY)

All premium and losses associated with policies *written* during a given calendar year are grouped together, regardless of when claims occur or payments are made.

- **Advantage**: Perfect premium-loss match — every dollar of written premium for PY 2022 is matched with every claim generated by those exact policies.
- **Disadvantage**: PY data takes the *longest* time to mature. A policy written December 31, 2022 can generate claims that settle a decade later — all still assigned to PY 2022.
- **Primary use**: Underwriting-year analysis, commercial lines pricing where the underwriter wants to evaluate the profitability of a specific book of business.

**Maturity comparison for a 12-month policy:**

| Aggregation | First "complete" accident exposure | Losses final by approximately... |
|---|---|---|
| Calendar Year | Immediately at year-end | Year-end (no development) |
| Accident Year | End of the accident year | Several years later (line-dependent) |
| Policy Year | Policies expire 12 months after inception | Many more years later (longer tail) |

---

### Report Year (RY)

Losses are grouped by the year in which the claim is *first reported* to the insurer. This method is most useful when IBNR (unreported claims) is not a major concern — typically for claims-made policies, where the policy itself limits which claims can be reported.

- **Advantage**: Eliminates pure IBNR (unreported claims); development on RY data comes only from IBNER (changes on already-reported claims). Data stabilizes faster than AY data for the same line.
- **Disadvantage**: Does not match well to written or earned premium; not appropriate for long-tail occurrence lines.
- **Primary use**: Claims-made medical malpractice; some GL analyses; excess/umbrella lines.

---

### Practical Notes on Selecting an Aggregation Method

1. **Ratemaking**: Accident year is standard. It provides a good premium-loss match while maturing in a reasonable time. Policy year is used for commercial lines where underwriting discipline is being evaluated.

2. **Reserving**: Accident year is most common. Calendar year is used only for financial reporting purposes. Report year may be used for claims-made lines.

3. **Development hierarchy**: PY takes longest to mature, then AY, then RY, then CY (no development).

> **Exam tip** — Know the pros and cons of each method cold. A common exam question gives a scenario (e.g., "the insurer wants to study the impact of a workers comp benefit change that took effect on January 1, 2022, on a specific group of policies") and asks which aggregation method best isolates the effect. Policy year will give the cleanest answer because it perfectly matches the premiums and losses of the affected policies.

## Claims Data and the Claims Lifecycle

### The Claims Lifecycle

A claim progresses through a predictable lifecycle. Each stage produces data that actuaries use in reserving and ratemaking:

1. **Loss event** — The accident or occurrence that gives rise to a potential claim
2. **First Notice of Loss (FNOL)** — The insured (or claimant) notifies the insurer; the claim is opened in the system
3. **Investigation** — The insurer investigates coverage, liability, and damages; an adjuster is assigned
4. **Initial Case Reserve Set** — The adjuster establishes an initial **case reserve** estimating the expected cost to settle
5. **Ongoing Adjustment** — As more information emerges, the case reserve is updated upward or downward
6. **Payment(s)** — The insurer makes loss payments (partial or final settlements, medical payments, etc.)
7. **Settlement / Closure** — The claim is settled and closed; no further payments expected
8. **Re-opening** — A closed claim may re-open if the insured's condition worsens or additional damages are discovered

---

### Claim Status

| Status | Description |
|---|---|
| **Open** | Claim is active; reserves may change; future payments expected |
| **Closed** | Settlement finalized; file closed; no further payments anticipated |
| **Closed Without Payment (CWP)** | Claim reported but determined to be non-covered, fraudulent, or otherwise without indemnity |

---

### Loss Measures

Understanding the hierarchy of loss measures is essential for both ratemaking and reserving:

**Paid losses**: amounts actually disbursed to claimants to date. Conservative measure; only reflects cash paid.

**Case reserve** (case outstanding): the adjuster's current estimate of the *remaining* amount needed to fully settle the open claim (excluding payments already made).

**Reported loss** (case incurred): the total expected cost of all *known* claims at the current valuation:

$$\text{Reported Loss} = \text{Paid Losses} + \text{Case Reserves}$$

**IBNR** (Incurred But Not Reported): represents claims that have *already occurred* but have not yet been reported to the insurer. These are claims the insurer does not yet know about. This is a pure "count" uncertainty — how many accidents have happened that we haven't heard about?

**IBNER** (Incurred But Not Enough Reported): the expected deficiency in the case reserves on *already-reported* claims. Even for known claims, case reserves may be set too low (especially early in the claim's life). IBNER = the reserve development that will occur on reported claims before they close.

$$\text{IBNER} = \text{Estimated Ultimate on Reported Claims} - \text{Aggregate Reported Losses at Valuation Date}$$

**Ultimate losses**: the total amount the insurer will pay once all claims — known and unknown — are fully settled:

$$\text{Ultimate Losses} = \text{Reported Losses} + \text{IBNR (Pure)}$$

**Total reserve** held by the insurer on the balance sheet covers all components:

$$\text{Total Reserve} = \text{Case Reserves} + \text{IBNR (Pure)} + \text{IBNER}$$

> **Key distinction** — IBNR (pure) is a count uncertainty (unreported events); IBNER is a reserve adequacy uncertainty (known claims with inadequate reserves). In common usage, actuaries often use "IBNR" loosely to refer to both. Regulators use "IBNR" in statutory filings to mean the entire reserve for unreported + underreserved claims.

---

### Introduction to Loss Development Triangles

A **loss development triangle** is the primary tool for studying how losses mature over time. Rows represent **accident years** (or policy years); columns represent the **age of evaluation** (months from the start of the accident year to the valuation date).

**Example — Paid Loss Triangle ($000s):**

| AY | 12 mo | 24 mo | 36 mo | 48 mo |
|---|---|---|---|---|
| 2021 | $1,200 | $1,500 | $1,620 | $1,680 |
| 2022 | $1,350 | $1,680 | $1,790 | — |
| 2023 | $1,450 | $1,820 | — | — |
| 2024 | $1,520 | — | — | — |

The rightmost observed value in each row forms the **latest diagonal** (the current valuation date). Values to the right of and below the diagonal are unknown — these are what actuaries estimate using development factors.

**Key observations:**
- **Paid triangles** grow monotonically (paid losses never decrease). Development is always positive or zero.
- **Incurred (reported) triangles** can move in either direction: upward when reserves are strengthened; downward when claims close for less than reserved or reserves are released.
- **Short-tailed lines** show rapid convergence to ultimate (most development occurs in the first 12–24 months).
- **Long-tailed lines** show slow, continued development over many years.

**Age-to-age factor** (link ratio): the ratio of losses at consecutive development ages for the same accident year, used to project forward:

$$f_{12 \to 24} = \frac{\sum_{\text{AY}} L_{24}}{\sum_{\text{AY}} L_{12}}$$

*(Full development methodology — chain ladder, BF, Benktander — is covered in the reserving notebooks.)*

> **Exam tip** — Triangles are tested in both the ratemaking and reserving domains of Exam 5. For ratemaking, you may be asked to read off the latest diagonal values and develop each accident year to ultimate as part of a pure premium or loss ratio calculation. For reserving, the same triangle becomes the starting point for IBNR estimation.

---

### Paid vs. Incurred Triangles: Which to Use?

| Consideration | Paid | Incurred (Reported) |
|---|---|---|
| **Volatility** | Lower — only cash payments | Higher — affected by reserving judgment |
| **Bias risk** | None from reserve changes | May be distorted by reserve strengthening/weakening trends |
| **Development speed** | Slower (payments lag reserves) | Faster (reserves set early) |
| **Preferred when** | Reserving philosophy has changed; reserve adequacy is in question | Stable reserving practices; want faster signal of ultimate |

In practice, actuaries examine both triangles and compare them. A divergence between paid and incurred development patterns is a key diagnostic signal (see triangle diagnostics notebook).

## Loss Adjustment Expenses (LAE)

**Loss Adjustment Expenses (LAE)** are the costs the insurer incurs in investigating, managing, and settling claims. They are treated separately from losses in actuarial analyses because they often develop at different rates and are driven by different forces.

---

### ALAE vs. ULAE

The traditional actuarial classification:

**Allocated LAE (ALAE)**: costs directly attributable to a *specific* claim. Each dollar of ALAE can be assigned to a particular claim file.
- Examples: outside defense counsel fees, court costs, expert witness fees, independent medical exams, private investigators
- Tracked at the claim level; developed and trended alongside losses
- Attorney involvement trends may cause ALAE to trend at a different rate than losses themselves

**Unallocated LAE (ULAE)**: general claims department costs that *cannot* be traced to an individual claim.
- Examples: claims adjuster salaries, claims department overhead, systems and technology costs, quality review staff
- Not tracked at the claim level; allocated to losses in aggregate using a ratio method

| Feature | ALAE | ULAE |
|---|---|---|
| Assignable to specific claim | Yes | No |
| Tracked per claim | Yes | No |
| Development pattern | Similar to losses | Needs separate estimation |
| Driver of trends | Litigation rates, attorney involvement | Staffing, overhead |

---

### NAIC Statutory Classification

The NAIC (National Association of Insurance Commissioners) annual statement uses a different breakdown that actuaries must also know:

| NAIC Term | Relationship to ALAE/ULAE | Description |
|---|---|---|
| **DCC** (Defense and Cost Containment) | Subset of ALAE | Costs of defending claims — outside legal counsel, cost containment services (nurse case managers, utilization review) |
| **A&O** (Adjusting and Other) | Remainder of ALAE + ULAE | Claims adjusting expenses, investigation costs, and all other LAE not in DCC |

The mapping: **ALAE** splits into DCC (defense portion) + part of A&O (adjusting portion). **ULAE** flows into A&O.

> **Exam tip** — Know both classification systems. ALAE/ULAE is used in actuarial analysis (development, trending). DCC/A&O is used in statutory financial reporting. A common exam trap: confusing which classification belongs to which framework.

---

### ULAE Allocation: Calendar Year Ratio Method

Since ULAE cannot be assigned to individual claims, a ratio method is used to estimate the ULAE provision for each accident year:

**Step-by-step:**
1. Gather **calendar year paid ULAE** for several recent years (from statutory filings)
2. Gather **calendar year paid losses + ALAE** for the same years
3. Compute the annual ratio: ULAE / (Losses + ALAE)
4. Select a ratio based on trend and judgment
5. Apply the selected ratio to each accident year's **reported losses + ALAE** to produce the ULAE provision

**Worked example:**

| CY | Paid ULAE | Paid Losses + ALAE | Ratio |
|---|---|---|---|
| 2022 | $580,000 | $5,200,000 | 11.2% |
| 2023 | $610,000 | $5,500,000 | 11.1% |
| 2024 | $650,000 | $5,800,000 | 11.2% |
| **Selected** | | | **11.2%** |

For AY 2024 with reported losses + ALAE = $4,800,000:

ULAE provision = $4,800,000 × 11.2% = **$537,600**

**Alternative: count-based methods.** More sophisticated methods allocate ULAE based on the number of open and closed claims, reflecting that about half of ULAE is incurred when a claim is opened and half when it is closed. These methods are more accurate but require claim count data.

---

### LAE in Ratemaking

In ratemaking, the actuary must decide how to treat LAE in the loss projection:

1. **Combine ALAE with losses** and develop/trend them together. Simple; appropriate when ALAE patterns closely track loss patterns.
2. **Study ALAE separately** from losses. More accurate when attorney involvement is trending up (pushing ALAE trend faster than loss trend) or when a line has a significant defense cost component.
3. **ULAE** is typically added as a ratio to losses + ALAE after the primary analysis, using the calendar year ratio method described above.

The standard ratemaking ratio often used is:

$$\text{Loss \& LAE Ratio} = \frac{\text{Losses} + \text{ALAE} + \text{ULAE}}{\text{Earned Premium}}$$

## Accounting Data and the Annual Statement

Insurance companies operate under a specialized accounting framework. Understanding the structure of insurer financials is essential for interpreting the data that feeds actuarial analyses.

---

### SAP vs. GAAP

Insurance companies file two sets of financial statements:

1. **Statutory Accounting Principles (SAP)**: filed with state insurance regulators; focuses on *solvency protection* for policyholders
2. **Generally Accepted Accounting Principles (GAAP)**: filed by publicly-traded insurers with the SEC; focuses on *periodic income measurement* for investors

The key differences:

| Feature | SAP | GAAP |
|---|---|---|
| **Primary purpose** | Policyholder solvency protection | Accurate periodic earnings for investors |
| **Policy acquisition costs** | Expensed immediately when the policy is written (conservative) | Deferred asset, amortized over the policy period |
| **Unearned premium reserve** | 100% of unearned written premium as a liability | Net of deferred acquisition costs |
| **Bond valuation** | Amortized cost (smoothes market fluctuations) | Fair value for trading/available-for-sale bonds |
| **Surplus concept** | **Policyholder surplus** = Assets − Liabilities | Shareholder equity = Assets − Liabilities |
| **Effect on surplus** | Lower than GAAP surplus (expensing acquisition costs immediately) | Higher than SAP surplus |

The conservative SAP treatment of acquisition costs reduces reported surplus — this is intentional. Regulators want the insurer to show its obligations prominently. An insurer that appears solvent under SAP should have adequate resources to pay claims.

> **Exam tip** — SAP surplus is lower than GAAP equity because acquisition costs are expensed immediately rather than deferred. Understand the direction of the difference and the reason behind it.

---

### The Statutory Annual Statement (The "Yellow Book")

The primary regulatory filing. Key schedules relevant to actuaries:

| Schedule | Content |
|---|---|
| **Part 1 (Underwriting and Investment Exhibit)** | Earned premium, incurred losses, and LAE by line of business |
| **Schedule P** | Loss and LAE development triangles by line (10 years of history) |
| **Schedule F** | Reinsurance recoverables and potential uncollectible reinsurance |
| **Schedule H** | Health-related lines (for health insurers) |
| **Annual Statement Page 14** | Summary of losses and loss adjustment expenses |

**Schedule P** is the annual statement source most used by actuaries — it provides the paid and incurred development triangles that form the backbone of any reserve analysis.

---

### The Balance Sheet

The balance sheet shows the insurer's financial position at a point in time:

$$\text{Assets} = \text{Liabilities} + \text{Policyholder Surplus}$$

**Key assets:**

| Asset | Description |
|---|---|
| **Bonds (fixed income)** | Typically 60–70% of invested assets; valued at amortized cost under SAP |
| **Equities** | Stocks; valued at market |
| **Premiums receivable** | Premium billed but not yet collected from policyholders |
| **Reinsurance recoverables** | Amounts due from reinsurers on ceded claims |
| **Cash and short-term investments** | Liquid assets |

**Key liabilities:**

| Liability | Description |
|---|---|
| **Unearned Premium Reserve (UPR)** | The insurer's obligation to provide future coverage on in-force policies; = unearned written premium |
| **Loss Reserve** | Estimated liability for all outstanding claims (reported losses + IBNR) |
| **LAE Reserve** | Estimated liability for future loss adjustment expenses on outstanding claims |
| **Ceded reinsurance premiums payable** | Amounts owed to reinsurers |
| **Other liabilities** | Taxes payable, accrued expenses, etc. |

**Policyholder surplus** is the residual:

$$\text{Policyholder Surplus} = \text{Total Assets} - \text{Total Liabilities}$$

Surplus is the insurer's financial cushion — what's left after all obligations are theoretically paid. It absorbs unexpected adverse loss development, catastrophe events, and investment losses. Regulators set minimum surplus requirements by line of business and size.

---

### The Income Statement

Two components make up total insurer income:

$$\text{Total Income} = \text{Underwriting Income (or Loss)} + \text{Net Investment Income} + \text{Realized Capital Gains}$$

**Underwriting income (loss):**

$$\text{UW Income} = \text{Earned Premium} - \text{Incurred Losses} - \text{Incurred LAE} - \text{UW Expenses}$$

A **negative underwriting income** (underwriting loss) means the combined ratio exceeded 100% — premiums collected were insufficient to cover losses and expenses.

**Net investment income**: interest and dividends earned on the invested portfolio (the "float").

**Realized capital gains**: gains (or losses) from selling investments above (below) cost.

> **Key insight** — A well-run insurer can sustain a moderate underwriting loss as long as investment income is sufficient. The classic example: workers compensation has a combined ratio that often exceeds 100% for individual insurers, but the long development tail generates substantial investment income on held reserves. The *total* return (underwriting + investment) can still be positive.

---

### The "Float" and Investment Income

One of the unique economics of insurance is the **float**: insurers collect premium *before* paying losses. During the time between premium collection and loss payment, the insurer invests those funds and earns investment income. The longer the development tail, the larger and longer-lasting the float.

| Component | Creates Float? |
|---|---|
| Unearned premium reserve | Yes — premium collected, coverage not yet provided |
| Loss reserves | Yes — losses incurred, payments not yet made |
| Policyholder surplus | Yes — surplus capital is invested too |

Longer-tailed lines (WC, GL) create more float than short-tailed lines (auto physical damage), which is why long-tail lines can price at combined ratios above 100% and still be profitable. Actuaries must account for this when setting target underwriting profit provisions.

## Performance Metrics and Ratios

Understanding the key financial metrics used to evaluate insurer performance is essential for interpreting actuarial analyses and answering exam questions about profitability.

---

### Basic Loss and Premium Metrics

| Metric | Formula | Interpretation |
|---|---|---|
| **Frequency** | Claims / Exposures | Number of claims per unit of exposed risk |
| **Severity** | Losses / Claims | Average cost per claim |
| **Pure Premium** | Losses / Exposures | Average loss cost per exposure unit; = Frequency × Severity |
| **Average Premium** | Premium / Exposures | Average amount charged per exposure unit |
| **Loss Ratio** | Losses / Earned Premium | Losses as a % of premium; = Pure Premium / Avg Premium |

These metrics decompose performance and help identify whether problems stem from frequency (more claims), severity (costlier claims), or pricing (inadequate premium).

---

### Expense Ratios

**Underwriting Expense Ratio** :

$$\text{UW Expense Ratio} = \frac{\text{Underwriting Expenses}}{\text{Written Premium}}$$

*Written* premium is the denominator because most underwriting expenses (commissions, taxes) are incurred when the policy is *written*, not as it earns.

**Underwriting expense categories:**

| Category | Nature | Typical Basis |
|---|---|---|
| Commissions and brokerage | Variable — % of premium | Written premium |
| Other acquisition costs | Variable | Written premium |
| Taxes, licenses, and fees | Variable | Written premium |
| General expenses | Mixed (fixed + variable) | Written or earned premium |

---

### The Combined Ratio

The **combined ratio** is the single most widely used measure of underwriting profitability:

$$\text{Combined Ratio} = \text{Loss Ratio} + \text{LAE Ratio} + \text{UW Expense Ratio}$$

**Trade basis** (industry standard):
$$\text{CR} = \frac{\text{Losses}}{\text{EP}} + \frac{\text{LAE}}{\text{EP}} + \frac{\text{UW Expenses}}{\text{WP}}$$

The "trade basis" mixes denominators: losses and LAE use *earned* premium; expenses use *written* premium. This matches how costs are incurred.

**Statutory basis**: all three components use *earned* premium as denominator. Less common in practice but sometimes tested.

| Component | Trade Basis Denominator | Statutory Basis Denominator |
|---|---|---|
| Loss ratio | Earned premium | Earned premium |
| LAE ratio | Earned premium | Earned premium |
| Expense ratio | **Written** premium | Earned premium |

**Interpretation:**
- Combined ratio = 100% → breaking even on underwriting (no profit, no loss)
- Combined ratio < 100% → underwriting profit
- Combined ratio > 100% → underwriting loss

Note: a combined ratio above 100% does *not* necessarily mean the company is unprofitable — investment income may more than offset the underwriting loss (see operating ratio below).

---

### Operating Ratio

The **operating ratio** incorporates investment income to give a more complete picture of profitability:

$$\text{Operating Ratio} = \text{Combined Ratio} - \text{Investment Income Ratio}$$

$$\text{Investment Income Ratio} = \frac{\text{Net Investment Income}}{\text{Earned Premium}}$$

- Operating ratio < 100% → company is profitable overall
- Operating ratio > 100% → company is losing money even after investment income

A combined ratio of 103% with an investment income ratio of 8% gives an operating ratio of **95%** — profitable overall despite an underwriting loss.

---

### Premium-to-Surplus Ratio

$$\text{Premium-to-Surplus Ratio} = \frac{\text{Net Written Premium}}{\text{Policyholder Surplus}}$$

Measures leverage — how much premium the insurer is writing relative to its capital cushion.
- **Higher ratio** → more leverage → more risk; insurer is stretched thin relative to its capacity to absorb unexpected losses
- **Lower ratio** → more conservative; capital is underdeployed
- Traditional regulatory benchmark: no higher than **3:1**

---

### Return on Surplus / Return on Equity

$$\text{Return on Surplus} = \frac{\text{Net Income}}{\text{Policyholder Surplus}}$$

Reflects overall profitability relative to the capital base. Combines underwriting income, investment income, and any realized gains. Rating agencies and regulators monitor this alongside the combined ratio.

---

### Marketing / Distribution Metrics

| Metric | Formula | What It Measures |
|---|---|---|
| **Retention Ratio** | Policies Renewed / Potential Renewal Policies | % of expiring policyholders who renew; measure of customer loyalty and competitive pricing |
| **Close Ratio** | Accepted Quotes / Total Quotes | % of quotes that convert to policies; measures sales effectiveness |

A low retention ratio may signal pricing is too high relative to competitors, poor service quality, or adverse selection. A low close ratio may indicate rates are uncompetitive in the new business market.

---

### Worked Example — All Ratios

**Given for Calendar Year 2024:**

| Item | Amount |
|---|---|
| Earned Premium | $50,000,000 |
| Written Premium | $52,000,000 |
| Incurred Losses | $31,000,000 |
| Incurred LAE | $5,000,000 |
| Underwriting Expenses | $13,000,000 |
| Net Investment Income | $4,000,000 |
| Net Income | $5,200,000 |
| Policyholder Surplus (year-end) | $40,000,000 |

**Compute:**

| Metric | Calculation | Result |
|---|---|---|
| Loss ratio | $31M / $50M | **62.0%** |
| LAE ratio | $5M / $50M | **10.0%** |
| UW expense ratio (trade) | $13M / $52M | **25.0%** |
| **Combined ratio** | 62.0% + 10.0% + 25.0% | **97.0%** |
| Investment income ratio | $4M / $50M | **8.0%** |
| **Operating ratio** | 97.0% − 8.0% | **89.0%** |
| Premium-to-surplus | $52M / $40M | **1.30:1** |
| Return on surplus | $5.2M / $40M | **13.0%** |

**Interpretation**: The insurer is in excellent shape. Combined ratio of 97% shows a modest underwriting profit. After investment income, the operating ratio of 89% reflects strong total profitability. Leverage is conservative at 1.3:1. Return on surplus of 13% is solid.

> **Exam tip** — The most common combined ratio error is using earned premium as the denominator for the expense ratio on the trade basis. Always use written premium for expenses. Also watch for questions that present the combined ratio as a single number and ask you to determine whether the company was profitable — remember that a combined ratio above 100% still allows profitability if investment income is high enough.
