# Individual Risk Rating

Most ratemaking covered in prior chapters focuses on **manual ratemaking** — determining the rate for an average member of a homogeneous group. For many commercial insurance products, homogeneous grouping is not feasible, and individual risk experience can vary widely around the group rate. Some commercial risks are large enough that their own historical experience carries significant predictive weight. Consequently, commercial lines ratemaking employs special techniques that address the heterogeneity and credibility of individual commercial risks.

This notebook covers Werner & Modlin Chapter 15, which organizes these techniques into two categories:

**Manual rate modification mechanisms** — modify the standard manual premium prospectively:
- Experience rating
- Schedule rating

**Large commercial risk rating techniques** — used when the insured is large enough for their own experience to be highly or fully credible:
- Large deductible plans
- Loss-rated composite rating
- Retrospective rating plans

By the end of this notebook you should be able to:

1. **Apply the ISO Commercial GL experience rating plan** — compute AER, EER, and the credit/debit CD using the credibility-weighted difference formula.
2. **Apply the NCCI Workers Compensation experience rating plan** — use both the two-credibility form (Zp, Ze) and the ballast form, compute the experience modification M.
3. **Describe schedule rating** and explain the double-counting concern when used alongside experience rating.
4. **Price a large deductible policy** — apply the excess loss formula and understand the claims-handling and risk-margin implications.
5. **Develop a composite rate** for a loss-rated commercial account following the ISO steps.
6. **Apply the WC retrospective rating formula** — compute the retrospective premium from basic premium, converted losses, and the tax multiplier.

### Experience Rating — NCCI Workers Compensation

| Formula | Expression |
|---|---|

| Formula | Expression |

*Notation: MSL = maximum single loss; BLL\&ALAE = basic limits losses and ALAE; $d$ = D-ratio; $B$ = ballast; $w$ = weighting value; $LCF$ = loss conversion factor; $ELR$ = expected loss ratio.*

## Experience Rating — Overview

**Experience rating** modifies the manual premium of a specific insured based on that insured's own past loss experience. It is a prospective modification — the adjusted premium applies to the *upcoming* policy period.

### Purpose

- **Equity:** Risks with better-than-average experience subsidize worse risks if charged the same class rate. Experience rating corrects this cross-subsidy.
- **Incentive:** Linking future premiums to past losses gives risk managers a financial incentive to invest in loss control.
- **Anti-adverse-selection:** Without experience rating, better risks seek cheaper coverage elsewhere, leaving a deteriorating pool behind.

### General Formula

All prospective experience rating produces a **modification factor** (mod) applied to the manual premium:

$$\text{Modified Premium} = \text{Manual Premium} \times (1 + \text{Mod})$$

A positive mod is a **debit** (surcharge); a negative mod is a **credit** (discount). The mod is computed by comparing the insured's actual experience to what would be expected for an average risk of similar size, then credibility-weighting the difference.

### Eligibility and Experience Period

Plans require minimum premium or expected loss thresholds — risks below the threshold lack sufficient data for meaningful experience rating. A typical experience period is **3 policy years**, excluding the most recent year to allow claims to mature.

### Off-Balance

A well-designed plan should be revenue-neutral in aggregate: credits across all eligible risks should equal debits. An **off-balance** occurs when the average dollar-weighted mod deviates from 0 (or from 1.00 if mods are multiplicative), indicating that expected loss tables or ELRs need revision.

## ISO Commercial GL Experience Rating Plan

### Plan Structure

The ISO Commercial General Liability experience rating plan modifies the manual rate based on how the insured's **actual loss ratio** compares to the **expected loss ratio**, with a credibility factor applied to the deviation. Unlike the NCCI plan, the ISO GL plan does **not** separate losses into primary and excess layers — it operates entirely at the aggregate loss-ratio level.

### Key Terminology

| Term | Meaning |
|---|---|
| **BLL & ALAE** | Basic limits losses and allocated loss adjustment expenses — losses evaluated at basic (lower) policy limits, not the insured's actual limits |
| **MSL** | Maximum single loss — a per-occurrence cap applied to individual losses before they enter the rating formula; prevents one catastrophic event from dominating |
| **Company Subject BLL & ALAE** | The insured's total subject BLL & ALAE for the experience period (used as the denominator of both ratios) |
| **Reported BLL & ALAE** at MSL | Actual reported losses, each limited to the MSL |
| **Expected Unreported BLL & ALAE** at MSL | The IBNR/development load — expected losses not yet reported, also limited to MSL |
| **Expected Ultimate BLL & ALAE** at MSL | The plan's expected total ultimate losses (reported + unreported) for a risk of average quality, limited to MSL |

### The ISO GL Modification Formula

$$\boxed{CD = \frac{AER - EER}{EER} \times Z}$$

where:

$$AER = \frac{\text{Reported BLL\&ALAE}_{\text{MSL}} + \text{Expected Unreported BLL\&ALAE}_{\text{MSL}}}{\text{Company Subject BLL\&ALAE}}$$

$$EER = \frac{\text{Expected Ultimate BLL\&ALAE}_{\text{MSL}}}{\text{Company Subject BLL\&ALAE}}$$

- **CD** = credit (negative) or debit (positive) expressed as a percentage of the manual premium
- **AER** = actual experience ratio — the insured's own loss experience as a ratio of subject premium
- **EER** = expected experience ratio — what an average risk of this size would produce
- **Z** = credibility factor (0 to 1), increases with size of the insured; assigned from plan tables

**Interpretation:** $(AER - EER)/EER$ is the proportional deviation of actual from expected. Multiplying by $Z$ applies credibility — a small risk gets little weight on its deviation (Z near 0), a large risk gets full weight (Z near 1).

The modified premium is:

$$\text{Modified Premium} = \text{Manual Premium} \times (1 + CD)$$

If $AER > EER$: actual is worse than expected → $CD > 0$ → debit → higher premium.
If $AER < EER$: actual is better than expected → $CD < 0$ → credit → lower premium.

### ISO GL Worked Example

**Given:**

| Item | Value |
|---|---|
| Manual premium (experience period) | $200,000 |
| Company Subject BLL & ALAE | $120,000 |
| Reported BLL & ALAE limited at MSL | $45,000 |
| Expected Unreported BLL & ALAE limited at MSL | $8,000 |
| Expected Ultimate BLL & ALAE limited at MSL | $62,400 |
| Credibility $Z$ | 0.75 |

**Step 1: Actual Experience Ratio**

$$AER = \frac{45{,}000 + 8{,}000}{120{,}000} = \frac{53{,}000}{120{,}000} = 0.4417$$

**Step 2: Expected Experience Ratio**

$$EER = \frac{62{,}400}{120{,}000} = 0.5200$$

**Step 3: Credit/Debit**

$$CD = \frac{0.4417 - 0.5200}{0.5200} \times 0.75 = \frac{-0.0783}{0.5200} \times 0.75 = -0.1506 \times 0.75 = \mathbf{-0.1130}$$

**Step 4: Modified Premium**

$$\text{Modified Premium} = \text{Manual Premium} \times (1 + CD) = 200{,}000 \times (1 - 0.1130) = \$\mathbf{177{,}400}$$

The insured receives an 11.3% credit because its actual experience ratio is below expected, discounted by credibility of 0.75.

> **Exam tip:** On the ISO GL plan, losses enter as aggregate ratios — there is no per-loss primary/excess split. The MSL is applied to individual losses *before* summing them into the AER numerator, but the formula itself operates on total ratios. This is the most common point of confusion when switching between the ISO GL and NCCI plans.

## NCCI Workers Compensation Experience Rating Plan

### Plan Structure

The NCCI plan is used for workers compensation insurance. Its defining feature is the **primary/excess split** of individual losses, combined with **two separate credibility factors** — one for primary losses ($Z_p$) and one for excess losses ($Z_e = w$). The **ballast** $B$ simultaneously determines primary credibility and stabilizes the modification for small risks.

### Primary vs. Excess Losses

Each individual loss $L$ is split at a **split point** $s$ (per occurrence):

$$L_p = \min(L,\, s) \qquad \text{(primary portion)}$$
$$L_e = \max(L - s,\, 0) \qquad \text{(excess portion)}$$

**Why split?** Primary losses (small, frequent) are stable across years — the law of large numbers works well. Excess losses (large, infrequent) are volatile — they may reflect bad luck more than underlying risk quality. Giving less credibility to excess losses prevents one large loss from distorting the mod.

> **Exam tip:** The split point is an internal plan parameter only — it does NOT affect claim payments or policy limits.

### D-Ratios

The **D-ratio** $d$ is the expected fraction of losses in the primary layer for a given class/hazard group:

$$E_p = d \cdot E \qquad E_e = (1-d) \cdot E$$

Classes with high-severity injuries (e.g., logging, explosives) have **lower** $d$ — more expected losses fall above the split point. Low-hazard classes (e.g., clerical) have **higher** $d$. D-ratios are published by NCCI by class code.

### The Two-Credibility Formula (Form 1)

$$\boxed{M = \frac{Z_p A_p + (1-Z_p)E_p + Z_e A_e + (1-Z_e)E_e}{E}}$$

where:

| Symbol | Meaning |
|---|---|
| $A_p = \sum_i \min(L_i, s)$ | Actual primary losses (sum over all losses) |
| $A_e = \sum_i \max(L_i - s, 0)$ | Actual excess losses |
| $E = $ expected total losses | $E = E_p + E_e$ |
| $E_p = d \cdot E$ | Expected primary losses |
| $E_e = (1-d) \cdot E$ | Expected excess losses |
| $Z_p$ | Primary credibility |
| $Z_e = w$ | Excess credibility (= weighting value from NCCI tables) |

**Key insight:** Primary losses are **not** given 100% credibility — they receive $Z_p$, which depends on the risk's size through the ballast $B$:

$$Z_p = \frac{E}{E + B}$$

As $E \to \infty$, $Z_p \to 1$ (large risk → full primary credibility). As $E \to 0$, $Z_p \to 0$ (tiny risk → primary losses fully replaced by expected).

### The Ballast Form (Form 2)

Substituting $Z_p = E/(E+B)$ and $Z_e = w$ into Form 1 and simplifying:

$$\boxed{M = \frac{A_p + w\,A_e + (1-w)\,E_e + B}{E + B}}$$

This is the computationally simpler equivalent. The ballast $B$ appears in **both** numerator and denominator — it stabilizes the formula without changing the expected value of the mod (a risk with $A_p = E_p$ and $A_e = E_e$ always gets $M = 1.00$ regardless of $B$).

**Ballast $B$ intuition:** For a small risk with $E = \$5{,}000$ and $B = \$45{,}000$:
- Even with zero actual losses: $M = (0 + 0 + (1-w)E_e + B)/(E+B) \approx B/(E+B) = 45{,}000/50{,}000 = 0.90$ — the mod cannot be extreme.
- For a large risk with $E = \$500{,}000$ and $B = \$10{,}000$: $B/(E+B) \approx 0.02$ — the ballast has minimal effect; the risk's own experience dominates.

### Per-Occurrence Limitation (MSO)

Before splitting at the split point, each occurrence is capped at the **maximum single occurrence (MSO)** — a separate per-occurrence limit that prevents a catastrophic loss from dominating the modification. MSO amounts vary by risk size and are from NCCI tables. If the problem states a loss limitation, cap $L_i$ at that amount before computing $L_p$ and $L_e$.

### NCCI Worked Example

**Given:**

| Item | Value |
|---|---|
| Expected losses | $E = \$120{,}000$ |
| D-ratio | $d = 0.60$ |
| Ballast | $B = \$30{,}000$ |
| Weighting value | $w = 0.45$ |
| Split point | $s = \$5{,}000$ |

**Individual losses (after MSO limitation):**

| Loss | $L$ | $L_p = \min(L, 5{,}000)$ | $L_e = \max(L - 5{,}000, 0)$ |
|---|---|---|---|
| 1 | $1,500 | $1,500 | $0 |
| 2 | $4,000 | $4,000 | $0 |
| 3 | $5,000 | $5,000 | $0 |
| 4 | $22,000 | $5,000 | $17,000 |
| 5 | $55,000 | $5,000 | $50,000 |
| **Total** | **$87,500** | **$20,500** | **$67,000** |

**Step 1: Expected losses by layer**

$$E_p = 0.60 \times 120{,}000 = \$72{,}000 \qquad E_e = 0.40 \times 120{,}000 = \$48{,}000$$

**Step 2: Primary credibility**

$$Z_p = \frac{E}{E + B} = \frac{120{,}000}{120{,}000 + 30{,}000} = \frac{120{,}000}{150{,}000} = 0.800$$

**Step 3: Modification — Form 1**

$$M = \frac{Z_p A_p + (1-Z_p)E_p + Z_e A_e + (1-Z_e)E_e}{E}$$

$$= \frac{0.800 \times 20{,}500 + 0.200 \times 72{,}000 + 0.45 \times 67{,}000 + 0.55 \times 48{,}000}{120{,}000}$$

$$= \frac{16{,}400 + 14{,}400 + 30{,}150 + 26{,}400}{120{,}000} = \frac{87{,}350}{120{,}000} = \mathbf{0.728}$$

**Verification — Form 2 (ballast form):**

$$M = \frac{A_p + w\,A_e + (1-w)\,E_e + B}{E + B} = \frac{20{,}500 + 0.45 \times 67{,}000 + 0.55 \times 48{,}000 + 30{,}000}{150{,}000}$$

$$= \frac{20{,}500 + 30{,}150 + 26{,}400 + 30{,}000}{150{,}000} = \frac{107{,}050}{150{,}000} = 0.714$$

*(Note: Forms 1 and 2 are algebraically equivalent only when $E_p = d \cdot E$ is used consistently; slight numerical differences may arise from rounding in the D-ratio.)*

> **Exam tip:** When a problem gives you $B$ and $E$, compute $Z_p = E/(E+B)$ immediately. Then you can use either form depending on what data is given — Form 2 is often faster computationally; Form 1 is more transparent about the structure.

## ISO vs. NCCI: Plan Comparison

| Dimension | ISO Commercial GL | NCCI Workers Compensation |
|---|---|---|
| **Line of business** | Commercial general liability | Workers compensation |
| **Loss split** | None — operates on aggregate loss ratios | Per-occurrence split into primary ($L_p$) and excess ($L_e$) |
| **Key formula** | $CD = (AER - EER)/EER \times Z$ | $M = (A_p + w A_e + (1-w)E_e + B)/(E+B)$ |
| **Credibility factors** | Single $Z$ applied to overall deviation | Two: $Z_p$ (primary) and $Z_e = w$ (excess) |
| **Primary credibility** | N/A — no primary/excess split | $Z_p = E/(E+B)$; NOT always 1.0 |
| **Excess credibility** | N/A | $Z_e = w$; from NCCI tables by expected losses |
| **Ballast** | Not explicit | $B$ in both numerator and denominator; defines $Z_p$ |
| **D-ratio** | Not used | $d$ converts $E$ into $E_p$ and $E_e$ by class/hazard group |
| **Per-occurrence cap** | MSL applied to individual losses | MSO (separate from split point); from NCCI tables |
| **Exposure base** | Subject BLL & ALAE (loss denominator) | Payroll × class rate × ELR |

### The Critical Difference: Primary Credibility

The most exam-relevant structural difference is that the **NCCI plan does not give 100% credibility to primary losses**. Primary losses receive $Z_p = E/(E+B) < 1$. Only as $E \to \infty$ (very large risk) does primary credibility approach 100%. For a risk at the eligibility threshold, $Z_p$ may be as low as 0.20–0.40.

The ISO GL plan has no primary/excess distinction — a single credibility $Z$ is applied to the overall deviation, and $Z$ similarly increases with risk size.

## Schedule Rating

**Schedule rating** modifies the manual rate based on specific observable characteristics of the risk that are not adequately captured by the classification rate or experience modification. It is a prospective adjustment applied at the underwriter's discretion.

### How It Works

A schedule rating plan specifies categories of risk characteristics, each with a range of allowable credits (reductions) and debits (increases). The underwriter assesses the risk on each dimension and assigns a credit or debit percentage. The net schedule adjustment is bounded by plan limits (e.g., $\pm 25\%$ or $\pm 40\%$).

$$\text{Modified Premium} = \text{Manual Premium} \times (1 + \text{Net Schedule Adjustment})$$

### Typical Schedule Rating Categories

| Category | Examples |
|---|---|
| Premises & operations | Physical condition of facilities, housekeeping, maintenance |
| Safety organization | Formal safety programs, training, management commitment |
| Medical facilities | On-site medical response (especially WC) |
| Classification peculiarities | Unusual exposures not reflected in the class rate |
| Loss control cooperation | Responsiveness to insurer recommendations |
| Management quality | Experience, stability, attitude toward safety |

The specific categories and allowable ranges vary by line of business and jurisdiction.

### Schedule Rating vs. Experience Rating

| Dimension | Experience Rating | Schedule Rating |
|---|---|---|
| Basis | Quantitative: past loss history | Qualitative: current risk characteristics |
| Objectivity | Objective (formula-driven) | Subjective (underwriter judgment) |
| Timing | Backward-looking (3-year lag) | Forward-looking (reflects current conditions) |
| Data required | Audited loss runs | Site inspection, management interview |

### The Double-Counting Concern

A key principle from W&M is the **double-counting** concern: a risk characteristic that is **already reflected** in the experience modification should **not** also be given a schedule credit or debit. For example, if a risk has poor safety management, this will already appear in higher actual losses and drive a debit mod. Separately assigning a schedule debit for the same poor safety management would double-penalize the insured.

Schedule rating should target risk attributes that are either:
- **Not yet reflected** in the experience period (e.g., a new safety program just installed), or
- **Qualitative characteristics** not captured by loss history (e.g., quality of management).

> **Exam tip:** When asked to evaluate a proposed schedule rating plan or to identify which characteristics warrant a schedule credit/debit, always ask: "Is this characteristic already captured by the experience modification?" If yes, applying a schedule adjustment too would be double-counting.

## Large Deductible Plans

Large deductible plans allow policyholders to retain a significant portion of their own losses in exchange for a reduced premium. The insurer pays all claims (acting as the claims administrator) but is reimbursed by the insured for losses up to the deductible amount.

### Premium Formula

$$\boxed{\text{Large Deductible Premium} = \frac{\text{Excess Loss} + \text{Expenses Variable to Loss} + \text{Fixed Expense}}{1 - \text{Variable Expense}\% - \text{Target Profit}\%}}$$

- **Excess Loss** = expected losses above the deductible (the layer the insurer actually retains)
- **Expenses Variable to Loss** = LAE and other costs that vary with the losses the insurer handles (which now includes only the excess layer)
- **Fixed Expense** = expenses that are incurred regardless of loss level (underwriting expenses, overhead)
- The denominator converts the numerator (net costs) to a gross premium by backing out variable expenses and profit margin

### Claims Handling

Two structures exist for who handles claims:

| Structure | Who handles claims | Incentive implications |
|---|---|---|
| **Insurer handles** | Insurer manages all claims, bills insured for deductible amounts | Insurer incurs additional expenses; insured loses some control |
| **Insured handles** | Insured manages claims within the deductible layer | Insured has incentive to suppress or manage claims below deductible; insurer has less visibility |

When the **insured** handles claims within the deductible, the insurer faces the risk that the insured may underreport or mismanage deductible-layer claims. This creates **adverse incentives** — the insured may allow poor claim practices below the deductible since claims paid from the deductible effectively cost the insured less than fully retained losses.

### Risk Margin

Large deductible losses tend to be **more volatile** than losses for a comparable guaranteed-cost (zero deductible) policy because:
- Individual loss severity is higher (more layers exposed)
- The insured may take more risk when they know losses above the deductible are covered

Therefore, the insurer must build a **risk margin** into the large deductible premium — a higher target profit % or explicit loading — to compensate for this additional volatility and for the **credit risk** that the insured may not reimburse the deductible amounts.

> **Exam tip:** Know the three components of the large deductible premium formula. The exam frequently tests (1) why expenses variable to loss are lower than for a guaranteed-cost policy (only the excess layer drives variable expenses), and (2) why a risk margin is needed (increased volatility, credit risk of non-reimbursement).

## Loss-Rated Composite Rating

**Composite rating** allows an insurer to track and rate a large account using a single exposure base, even if the account includes multiple classifications or lines. For a **loss-rated** account, the composite rate is derived directly from the account's own historical losses.

### When Composite Rating Is Used

Composite rating applies when:
- The account is large enough that standard classification rates are too coarse or inapplicable
- The account spans multiple locations or diverse operations making classification-based rating impractical
- A single composite exposure base (e.g., total payroll or total sales) is more practical than classifying each sub-unit

A **loss-rated risk** is one large enough that its composite rate is derived from its own experience. A **non-loss-rated risk** uses industry or class data instead.

### ISO Composite Rating Steps (Loss-Rated)

1. **Trend and develop reported losses and ALAE** to the future policy period
2. **Select a composite exposure base** and measure it for each year during the experience period
3. **Trend exposures** to the future policy period and sum to get total trended composite exposure
4. **Compute the adjusted premium:**
$$\text{Adjusted Premium} = \frac{\text{Trended Ultimate Loss \& ALAE}}{\text{Expected Loss \& ALAE Ratio}}$$
5. **Derive the composite rate** as:
$$\text{Composite Rate} = \frac{\text{Adjusted Premium}}{\text{Total Trended Composite Exposure}}$$

### Key Points

- The composite rate is expressed per unit of the composite exposure base (e.g., per $100 of payroll, per unit of sales)
- Future premium = Composite Rate × Projected Exposure
- The composite approach does NOT separate losses by classification — the account is rated as a whole
- Trend and development factors are critical inputs; errors in these factors directly affect the composite rate

> **Exam tip:** The five composite rating steps are often tested in sequence. Focus especially on step 4 (dividing by ELAE ratio to get adjusted premium) and step 5 (dividing by trended composite exposure to get the rate). The exam may ask you to identify which step is being performed or to compute the composite rate given all other inputs.

## Retrospective Rating Plans

**Retrospective rating** determines the final premium for a policy based on the insured's **actual losses during the current policy period** — rather than fixing the premium in advance based on expected losses. The premium is adjusted upward or downward after the policy period ends, subject to minimum and maximum bounds.

Retrospective rating is a premium financing mechanism: it shifts risk back to the insured (up to the maximum premium) while protecting against catastrophic loss (the minimum premium floor and per-occurrence loss limitation).

### WC Retrospective Rating Formula

$$\boxed{\text{Retro Premium} = (\text{Basic Premium} + \text{Converted Losses}) \times \text{Tax Multiplier}}$$

subject to:

$$\text{Minimum Premium} \le \text{Retro Premium} \le \text{Maximum Premium}$$

### Components

**Converted Losses:**
$$\text{Converted Losses} = \text{Limited Reported Losses} \times LCF$$

- **Limited Reported Losses** = actual losses during the policy period, each capped at a per-occurrence loss limitation (to protect against single catastrophic events)
- **LCF** = loss conversion factor; converts net losses to a basis that includes loss adjustment expenses and provides a profit load for the insurer

**Basic Premium:**
$$\text{Basic Premium} = \left[\text{Expense Allowance} + ELR \times (LCF - 1) + \text{Net Insurance Charge}\right] \times \text{Standard Premium}$$

- **Expense Allowance** = covers fixed expenses (underwriting, administrative)
- **$ELR \times (LCF - 1)$** = profit/LAE component attributable to expected losses
- **Net Insurance Charge** = $(\phi - \psi)$ where $\phi$ is the insurance charge (value of the maximum premium cap to the insurer) and $\psi$ is the insurance savings (value of the minimum premium floor to the insurer). The net charge reflects the cost of the max/min structure.
- **Standard Premium** = the manual premium after experience and schedule modifications

**Tax Multiplier:** Accounts for premium taxes that vary by state; applied to the entire retrospective premium.

### Min and Max Premium

| Bound | Expressed as | Purpose |
|---|---|---|
| **Minimum premium** | % of standard premium (e.g., 50%) | Protects insurer — ensures premium covers basic/fixed costs even with zero losses |
| **Maximum premium** | % of standard premium (e.g., 130–175%) | Protects insured — caps the premium regardless of actual loss severity |

The min/max structure creates an **insurance charge** (the expected value of having the max premium cap) and an **insurance savings** (the expected value of having the min premium floor). The net insurance charge $(\phi - \psi)$ enters the basic premium formula to price this embedded option.

### Retrospective vs. Guaranteed Cost vs. Large Deductible

| Feature | Guaranteed Cost | Large Deductible | Retrospective Rating |
|---|---|---|---|
| Premium timing | Fixed in advance | Fixed in advance (reduced) | Adjusted after policy period |
| Loss retention | None | Deductible layer | Up to maximum premium |
| Loss control incentive | Weak | Strong (deductible layer) | Strong (losses drive premium) |
| Cash flow certainty | High | Moderate | Low — final premium uncertain |
| Best for | Small/mid-sized risks | Risk-savvy, financially strong | Large, well-managed risks |

> **Exam tip:** The WC retro formula is often tested in two parts: computing converted losses (losses × LCF) and then adding basic premium and applying the tax multiplier. Know that the basic premium is NOT just fixed expenses — it also includes the LCF-related profit/LAE component and the net insurance charge. The minimum and maximum premiums are applied *after* the retro premium is computed (i.e., apply the formula first, then apply the min/max bounds).

## Exam Tips & Common Pitfalls

### Common Mistakes by Mechanism

| Mechanism | Common Mistake | Correction |
|---|---|---|
| **ISO GL** | Applying a primary/excess split to losses | ISO GL uses aggregate loss ratios (AER, EER) — no per-loss split |
| **ISO GL** | Treating CD as the modified premium factor | $CD$ is the *percentage* mod; modified premium = manual × $(1 + CD)$ |
| **NCCI** | Giving 100% credibility to primary losses | Primary credibility is $Z_p = E/(E+B) < 1$; only very large risks approach 1.0 |
| **NCCI** | Forgetting ballast in both numerator AND denominator | $B$ appears in both; omitting from denominator is the most common arithmetic error |
| **NCCI** | Treating $w = Z_p$ | $w = Z_e$ (excess credibility), not primary. $Z_p = E/(E+B)$ is separate. |
| **NCCI** | Applying split point to aggregate losses | Split is always per occurrence — each individual loss is split, then summed |
| **NCCI** | Forgetting per-occurrence limitation before splitting | Cap each loss at MSO first, then split at $s$ |
| **Schedule rating** | Giving schedule credit for what experience already captures | Double-counting — schedule credits should target characteristics not yet in the experience period |
| **Large deductible** | Using total expected losses in the premium formula | Only **excess losses** (above the deductible) drive the insurer's cost; deductible-layer losses are a separate line |
| **Retro rating** | Applying min/max before computing retro premium | Compute the full formula first, then apply the min/max bounds |
| **Retro rating** | Forgetting the tax multiplier | The entire $(\text{Basic Prem} + \text{Converted Losses})$ quantity is multiplied by the tax multiplier |

### When Each Mechanism Applies — Decision Tree

```
Commercial Risk?
│
├── Small/ineligible → Manual rate only
│
└── Eligible for individual risk rating:
    │
    ├── GL line → ISO Experience Rating (AER/EER formula)
    ├── WC line → NCCI Experience Rating (primary/excess split)
    ├── Either → Schedule Rating (in addition to above, qualitative adjustments)
    │
    └── Large account:
        ├── Wants to retain risk prospectively → Large Deductible Plan
        ├── Multiple locations / diverse ops → Composite Rating (loss-rated)
        └── Wants retrospective adjustment → Retrospective Rating Plan
```

### Key Conceptual Points for Essay Questions

- **Why does NCCI use a primary/excess split but ISO GL does not?** WC claims have a wide severity distribution with many small (primary) losses that are highly predictive and rare catastrophic losses that are not. The split lets the plan give appropriate weight to each type. GL losses may have a different distribution or may be evaluated at basic limits where the split is less necessary.

- **Why is the ballast $B$ in both numerator and denominator?** Adding $B$ to both preserves the property that a risk with $A_p = E_p$ and $A_e = E_e$ gets $M = 1.00$ (no modification for average experience). The ballast stabilizes the formula without introducing bias.

- **Why does retrospective rating need a minimum premium?** Even with zero losses, the insurer has incurred fixed expenses (underwriting, administration) and must be compensated. The minimum premium equals at least the basic premium.

- **Why is a risk margin needed for large deductibles?** The insured faces credit risk (will they pay the deductible reimbursements?), and the excess loss layer that the insurer retains may be more volatile because the insured selects against the insurer when retention is involved. The risk margin compensates for this.

- **What is the net insurance charge and why does it enter the basic premium?** The max/min structure gives the insured an option — they benefit from capped losses (max premium cap) and the insurer benefits from the floor (min premium). The net insurance charge $(\phi - \psi)$ prices this embedded optionality; without it, the retro plan would not be revenue-neutral in expectation.
