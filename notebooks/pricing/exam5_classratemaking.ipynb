# Class Ratemaking

**Classification ratemaking** is the process of grouping risks with similar expected loss potential and charging different rates to reflect differences in expected costs among those groups. While overall ratemaking ensures that aggregate premium covers aggregate costs, classification ratemaking ensures that individual risks — or segments of risks — are priced equitably. A company that fails to classify risks appropriately while competitors do so will face adverse selection and deteriorating profitability. Conversely, a company that identifies and prices risk characteristics better than its competitors can gain a lasting competitive advantage.

This notebook covers the classification ratemaking material tested on CAS Exam 5, drawing primarily from Werner & Modlin Chapters 9–12. By the end you should be able to:

1. **Risk classification basics** — Explain adverse selection, favorable selection, and skimming the cream; define key terminology (rating variable, level, relativity).
2. **Criteria for evaluating rating variables** — Apply the four criterion categories (statistical, operational, social, legal) to evaluate proposed rating variables.
3. **Univariate classification methods** — Apply the pure premium, loss ratio, and adjusted pure premium methods to produce indicated relativities; apply credibility blending for each method.
4. **Multivariate classification methods** — Describe the minimum bias procedure and GLM approaches, explain why GLMs are preferred, and identify GLM diagnostics.
5. **Special classifications** — Apply territorial ratemaking (smoothing and clustering), increased limits factors (LAS approach), deductible pricing (LER approach), workers compensation size-of-risk adjustments, and insurance-to-value/coinsurance mechanics.
6. **Credibility** — Apply classical and Bühlmann credibility formulas; evaluate complements of credibility using Boor's six desirable qualities for first dollar and excess ratemaking.

## Formula Sheet Quick Reference

Key formulas for classification ratemaking — scan this before working problems.

| Formula | Expression |
|---|---|
| **PP indicated relativity** (level $i$) | $R_i^{PP} = \dfrac{PP_i}{PP_{base}}$ |
| **LR indicated change factor** (level $i$) | $CF_i = \dfrac{LR_i}{LR_{total}}$ |
| **LR indicated relativity** (level $i$) | $R_i^{LR} = CF_i \times R_i^{current}$, then rebase to base |
| **Adjusted PP** | Adjust exposures by wtd avg current relativity of other variables; then $R_i^{adj} = PP_i^{adj} / PP_{base}^{adj}$ |
| **Classical credibility (Z)** | $Z = \min\!\left(\sqrt{N / E(N)},\ 1\right)$ |
| **Full credibility standard** | $E(N) = \left(\dfrac{z_{(1+p)/2}}{k}\right)^2\!\left(1 + \dfrac{\sigma_s^2}{\mu_s^2}\right)$ |
| **Bühlmann credibility (Z)** | $Z = \dfrac{N}{N + K},\quad K = \dfrac{\text{EVPV}}{\text{VHM}}$ |
| **Credibility-weighted estimate** | $\hat{\mu} = Z \times \text{Observed} + (1-Z) \times \text{Complement}$ |
| **Indicated ILF** | $\text{ILF}(H) = \dfrac{\text{LAS}(H)}{\text{LAS}(B)}$ |
| **LAS (uncensored)** | $\text{LAS}(H) = \dfrac{\text{Total losses capped at } H}{\text{Total claim count}}$ |
| **LAS (censored, layer)** | $\text{LAS}(y) = \text{LAS}(x) + \dfrac{\text{Losses in layer } [x, y]}{\text{Claims from policies with limit} > x}$ |
| **Loss Elimination Ratio** | $\text{LER}(D) = \dfrac{(L+E_L)_B - (L+E_L)_D}{(L+E_L)_B}$ |
| **Indicated deductible relativity** | $1 - \text{LER}(D)$ |
| **Coinsurance apportionment ratio** | $a = \min\!\left(F / cV,\ 1\right)$ |
| **Coinsurance indemnity** | $I = \min(aL,\ F)$ |
| **Coinsurance penalty** | $e = \min(L, F) - I$ |

> **Formula sheet note** — The classical credibility formula simplifies to $E(Y) = (z_{(1+p)/2}/k)^2$ when severity is assumed constant. The full formula including the coefficient of variation of severity is tested less frequently, but know the direction of the adjustment.

## Risk Classification Basics

### Key Terminology

**Classification ratemaking** groups risks into segments with similar expected loss potential and assigns rates accordingly. The underlying rate structure uses **rating variables** — characteristics used to vary rates. Each rating variable has two or more **levels** (its distinct values), with one level designated the **base level** against which all other levels are expressed. A rate differential applied multiplicatively is a **relativity**; one applied additively is an **additive**.

For example, a homeowners insurer might use "Age of Home" as a rating variable with levels: New (0–5 years), Medium (6–20 years), and Old (21+ years). If Medium is the base, rates for New and Old are expressed as relativities to Medium — e.g., 0.85 and 1.15, meaning new homes are charged 85% and old homes 115% of the Medium rate.

> **Exam tip** — Know the difference between a *rating variable* (used to set rates) and an *underwriting variable* (used for risk selection but may not flow through the rate directly). Werner & Modlin use "rating variable" to mean any variable used to vary the rate, regardless of whether it might normally be called an underwriting characteristic.

---

### Adverse Selection

When a company fails to segment risks on a meaningful dimension that competitors are using, it attracts higher-risk insureds and loses lower-risk insureds. This self-reinforcing cycle is **adverse selection**:

1. The insurer fails to implement a rating variable that competitors are using to segment risks.
2. The insurer experiences a distributional shift toward higher-risk insureds.
3. The rate charged becomes inadequate; the insurer becomes unprofitable.
4. The insurer raises the average rate.
5. Steps 2–4 repeat until the insurer improves its rate segmentation, becomes insolvent, or focuses entirely on higher-risk insureds.

*Simple Company vs. Refined Company example (Werner & Modlin Ch. 9):* Suppose high-risk insureds cost $230 and low-risk insureds cost $130; both types are equally split in the market. Simple Company charges $180 (the average) while Refined Company charges $230 and $130 respectively. Because 1 in 10 insureds shops on price, the low-risk insureds migrate to Refined Company (where their rate is lower) and high-risk insureds migrate to Simple Company. By policy year 5, Simple Company's charged rate has risen to $197 while it still cannot cover its increasingly adverse portfolio.

---

### Favorable Selection

Conversely, a company that **identifies a risk characteristic before competitors do** can implement it as a new rating variable and attract more lower-risk insureds at a profit. This is **favorable selection** — the opposite of the adverse selection cycle. The insurer gains a competitive advantage by charging lower rates to the good risks (who flow to the insurer because they can now get better pricing) while charging appropriate rates to the bad risks.

---

### Skimming the Cream

If a company cannot or chooses not to implement a new rating variable but still knows which risks in the population are lower risk, it may focus its marketing and underwriting on attracting those lower-risk insureds without changing the rate. This is called **"skimming the cream."** The company writes only the attractive risks at the average rate and earns a profit, while competitors are left with the remaining (less favorable) mix.

---

### ASOP No. 25

**Actuarial Standard of Practice No. 25** ("Credibility Procedures") defines *credibility* as "a measure of the predictive value in a given application that the actuary attaches to a particular body of data." When selecting a complement of credibility, ASOP 25 requires that the related experience have frequency, severity, or other determinable characteristics reasonably expected to be similar to the subject experience. This standard is directly relevant to the credibility section later in this notebook.

## Criteria for Evaluating Rating Variables

Before implementing a rating variable, the actuary evaluates whether it is appropriate using four categories of criteria (Finger 2001):

| Category | Sub-criteria | Core question |
|---|---|---|
| **Statistical** | Significance, Homogeneity, Credibility | Does the variable actually predict differences in expected cost, reliably? |
| **Operational** | Objective, Inexpensive to administer, Verifiable | Is it practical to use in a rating system? |
| **Social** | Affordability, Causality, Controllability, Privacy | Is it socially acceptable? |
| **Legal** | Compliance with laws and regulations | Is it legally permissible in each jurisdiction? |

---

### Statistical Criteria

1. **Statistical significance** — The expected costs must vary across levels, the differences must be within acceptable statistical confidence, and the results should be stable over time. A variable whose level-by-level experience shows no variation (or only random noise) adds no rating value.

2. **Homogeneity** — Each level should contain risks with *similar* expected costs. The goal is within-group homogeneity and between-group heterogeneity. Note that identical risks may still have different *actual* outcomes in a given year due to randomness — homogeneity is about expected cost, not observed cost.

3. **Credibility** — Each level must have enough exposure to produce a reliable estimate. If a level is too thinly populated, combine similar levels or find additional data. This is the tension between granularity (homogeneity) and stability (credibility).

---

### Operational Criteria

1. **Objective** — The level definitions must be unambiguous. "Skill level of a surgeon" is subjective and unmeasurable; "board certification" or "years of experience" are objective proxies.

2. **Inexpensive to administer** — The cost of collecting and verifying the data must not exceed the ratemaking benefit. Building wind-resistance features into a home may reduce expected losses, but requiring a professional inspection to verify them may be too costly.

3. **Verifiable** — The insurer must be able to confirm the information is accurate and not easily manipulated. Annual mileage driven is a plausible predictor of auto losses, but self-reported estimates are difficult to verify (though telematics devices are changing this).

---

### Social Criteria

1. **Affordability** — If using a variable creates premiums that are unaffordable for a segment of society — especially when insurance is legally required — regulators may require cross-subsidies. Example: 16-year-old drivers may be grouped with 17-year-olds to limit the premium burden on families.

2. **Causality** — Variables directly linked to the mechanism of loss are more socially defensible. Sump pump presence has an obvious causal link to water damage losses. Credit scores have strong predictive power but no obvious causal mechanism, generating public backlash.

3. **Controllability** — It is socially desirable (and societally beneficial) for insureds to be able to improve their risk profile — e.g., implementing loss control programs, adding safety equipment, improving claims history. Uncontrollable characteristics like age or gender face more regulatory scrutiny.

4. **Privacy concerns** — Rating variables that require collection of sensitive personal data (e.g., telematics tracking driving behavior) may raise privacy objections even if voluntarily provided.

---

### Legal Criteria

Rating variables must comply with all applicable laws and regulations in each jurisdiction. Requirements vary widely by state and product:
- Most states require rates to be "not excessive, not inadequate, and not unfairly discriminatory."
- Some states prohibit gender as a rating variable; others allow it.
- Some states permit credit scores for new business but restrict their use at renewal.
- Some states mandate specific ratemaking techniques (e.g., California's sequential analysis requirement for private passenger auto).

> **Exam tip** — Do not confuse **controllability** (social criterion: can the insured improve their risk and lower their premium?) with **verifiability** (operational criterion: can the insurer confirm the information is accurate?). A variable can satisfy both, either, or neither. Annual mileage, for example, is highly controllable (you can drive less) but historically difficult to verify.

## Univariate Classification Methods

The actuary uses historical experience for each level of a rating variable to estimate the appropriate rate differential (relativity) between levels. All three approaches assume a **multiplicative rating structure** (so differentials are relativities) and produce **indicated relativities to base**. The approaches differ in how they handle exposure correlations with other rating variables.

---

### Setup: Two-Variable Worked Example

Suppose a book of business has two rating variables: **Territory** (T1, T2=base, T3) and **Amount of Insurance** (Low, Medium=base, High). Assume all expenses are variable at 30%, profit at 5%, so PLR = 65%.

The exposure and loss data for Territory (collapsing across AOI) are:

| Territory | Exposures | Loss & ALAE | Premium at Current Rate | Current Relativity |
|---|---|---|---|---|
| T1 | 294 | $15,235 | $21,315 | 0.60 |
| T2 (base) | 385 | $26,510 | $40,415 | 1.00 |
| T3 | 309 | $23,255 | $38,270 | 1.30 |
| **Total** | **988** | **$65,000** | **$100,000** | — |

---

### 1. Pure Premium Approach

**Steps:**
1. Calculate the pure premium for each level: $PP_i = \text{Loss \& ALAE}_i / \text{Exposures}_i$
2. Calculate the indicated relativity: $R_i = PP_i / PP_{base}$

**Applied to Territory:**

| Territory | Exposures | Loss & ALAE | Pure Premium | Indicated Relativity |
|---|---|---|---|---|
| T1 | 294 | $15,235 | $51.82 | $51.82 / $68.86 = **0.753** |
| T2 (base) | 385 | $26,510 | $68.86 | 1.000 |
| T3 | 309 | $23,255 | $75.26 | $75.26 / $68.86 = **1.093** |

**Key limitation:** The pure premium for each territory reflects the *combined effect* of territory and all other rating variables. If T1 has a disproportionately high concentration of High-AOI risks (which cost more), the territory pure premium will be inflated — not because T1 itself is riskier, but because of the exposure correlation with AOI. The pure premium approach **does not control for this distributional bias**.

---

### 2. Loss Ratio Approach

**Steps:**
1. Calculate the loss & ALAE ratio for each level: $LR_i = \text{Loss \& ALAE}_i / \text{Premium}_i$
2. Calculate the indicated change factor: $CF_i = LR_i / LR_{total}$
3. Calculate the indicated relativity: $R_i^{indicated} = CF_i \times R_i^{current}$
4. **Rebase** to the base level: divide all indicated relativities by the indicated relativity at the base level.

**Applied to Territory:**

| Territory | LR | CF = LR / LR_total | Indicated Relativity | Rebased to T2 |
|---|---|---|---|---|
| T1 | 71.47% | 71.47/65.00 = 1.100 | 1.100 × 0.60 = 0.660 | 0.660 / 1.100 = **0.600** |
| T2 (base) | 65.60% | 65.60/65.00 = 1.009 | 1.009 × 1.00 = 1.009 | 1.009 / 1.009 = 1.000 |
| T3 | 60.76% | 60.76/65.00 = 0.935 | 0.935 × 1.30 = 1.215 | 1.215 / 1.009 = **1.204** |
| Total | 65.00% | — | — | — |

**Key limitation:** The loss ratio approach adjusts for distributional bias *only to the extent that current premium already reflects true relativities*. If the current relativities are themselves off, the loss ratio method inherits that error. It partially corrects for exposure correlations (through the premium denominator) but assumes the current rate structure is equitable.

---

### 3. Adjusted Pure Premium Approach

**Steps:**
1. For each exposure unit, calculate the **weighted average current relativity from all rating variables other than the one being analyzed** (i.e., the product of current relativities for all variables except the one being analyzed).
2. **Divide each exposure** by its weighted average current relativity to get an "adjusted" exposure — this normalizes exposures to the base level of every other rating variable.
3. Calculate adjusted pure premium: $PP_i^{adj} = \text{Loss \& ALAE}_i / \text{Adjusted Exposures}_i$
4. Indicated relativity: $R_i = PP_i^{adj} / PP_{base}^{adj}$

**Intuition:** If T1 has many High-AOI risks, those exposures are "worth more" in expected-loss terms. By dividing them by the High-AOI current relativity (1.35), we standardize them to the Medium-AOI basis. The resulting adjusted pure premiums reflect the territory effect alone, free from the AOI distributional bias.

**Key advantage:** Adjusts for distributional bias without requiring on-level premium. The result is closer to the "true" territory relativities than the unadjusted PP approach.

---

### Comparison of the Three Methods

| Method | Adjusts for Exposure Correlations? | Requires On-Level Premium? | Inherits Current-Rate Bias? |
|---|---|---|---|
| Pure Premium | **No** — exposure correlations distort results | No | No |
| Loss Ratio | **Partially** — via current premium structure | **Yes** | **Yes** — assumes current rates are equitable |
| Adjusted PP | **Yes** — via exposure weighting | No | Partially — adjusts using current relativities |

> **Exam tip** — The pure premium approach is distorted by any distributional bias in other rating variables. This is the most commonly tested limitation. The adjusted pure premium approach is the most accurate univariate method but requires knowing the current relativities of all other rating variables to adjust exposures.

## Univariate Methods with Credibility

When the experience for a particular level of a rating variable is sparse or volatile, the actuary blends the indicated relativity with a **complement of credibility** — often the current (or a competitor's) relativity. The blend is:

$$\hat{R}_i = Z \times R_i^{indicated} + (1 - Z) \times R_i^{current}$$

However, because indicated and current relativities are expressed relative to the *base level*, simply weighting them can produce results that no longer balance properly across the book. The correct procedure **re-bases both sets of relativities to the total** before weighting, then re-bases the result back to the base level.

---

### Procedure: PP Method with Credibility

1. **Re-base current relativities to total**: $R_i^{c \to T} = R_i^{current} \div \bar{R}^{current}$ where $\bar{R}^{current}$ is the exposure-weighted average current relativity across all levels.
2. **Re-base indicated relativities to total**: $R_i^{I \to T} = R_i^{indicated} \div \bar{R}^{indicated}$ (exposure-weighted avg of indicated relativities).
3. **Blend**: $R_i^* = Z \times R_i^{I \to T} + (1 - Z) \times R_i^{c \to T}$
4. **Rebase to base level**: $R_i^{final} = R_i^* \div R_{base}^*$

For the **adjusted PP method with credibility**: use the same 4 steps but weight by *adjusted exposures* in steps 1–2.

For the **loss ratio method with credibility**: use the same 4 steps but weight by *premium at base level* in steps 1–2.

---

### Shortcut: "No Change" Complement

If the complement is **no change** (i.e., the current relativities are held fixed at 1.0 for all levels), re-basing can be skipped. Instead, simply credibility-weight the indicated **change factor** vs. 1.0:

$$CF_i^{blended} = Z \times CF_i^{indicated} + (1 - Z) \times 1.0$$

This works because the change factor already expresses the movement relative to the current rate level — a factor of 1.0 means no change.

---

### Worked Example

Two-territory analysis. Territory 2 is base. Exposures: T1 = 100, T2 = 200, T3 = 100. Credibility Z = 0.60 for T1 and T3.

**Data:**

| Territory | Exposures | Loss & ALAE | Current Relativity | Indicated Relativity (PP) |
|---|---|---|---|---|
| T1 | 100 | $4,800 | 0.75 | 0.64 |
| T2 (base) | 200 | $14,000 | 1.00 | 1.00 |
| T3 | 100 | $8,200 | 1.20 | 1.26 |
| **Total** | **400** | **$27,000** | — | — |

Pure premiums: T1 = $48.00, T2 = $70.00, T3 = $82.00.
Indicated relativities to T2: T1 = 48/70 = 0.686, T2 = 1.000, T3 = 82/70 = 1.171. (Note: these differ slightly from the table due to rounding in example.)

**Step 1: Re-base current relativities to total (exposure-weighted avg):**

$\bar{R}^{c} = (100 \times 0.75 + 200 \times 1.00 + 100 \times 1.20) / 400 = (75 + 200 + 120) / 400 = 395/400 = 0.9875$

| Territory | $R^{current}$ | $R^{c \to T}$ = $R^c / 0.9875$ |
|---|---|---|
| T1 | 0.75 | 0.759 |
| T2 | 1.00 | 1.013 |
| T3 | 1.20 | 1.215 |

**Step 2: Re-base indicated relativities to total (exposure-weighted avg):**

$\bar{R}^{I} = (100 \times 0.686 + 200 \times 1.000 + 100 \times 1.171) / 400 = (68.6 + 200 + 117.1) / 400 = 385.7/400 = 0.964$

| Territory | $R^{indicated}$ | $R^{I \to T}$ = $R^I / 0.964$ |
|---|---|---|
| T1 | 0.686 | 0.712 |
| T2 | 1.000 | 1.037 |
| T3 | 1.171 | 1.215 |

**Step 3: Blend (Z = 0.60 for T1 and T3; T2 = base, so it adjusts implicitly):**

| Territory | $R^{I \to T}$ | $R^{c \to T}$ | Blended $R^*$ |
|---|---|---|---|
| T1 | 0.712 | 0.759 | 0.60(0.712) + 0.40(0.759) = **0.731** |
| T2 | 1.037 | 1.013 | (base — no credibility applied) |
| T3 | 1.215 | 1.215 | 0.60(1.215) + 0.40(1.215) = **1.215** |

**Step 4: Rebase to T2.** $R_{T2}^* = 1.037$ (or recompute from blended; use whatever T2 blended to):

| Territory | Final Relativity = $R^* / R_{T2}^*$ |
|---|---|
| T1 | 0.731 / 1.037 = **0.705** |
| T2 | 1.037 / 1.037 = **1.000** |
| T3 | 1.215 / 1.037 = **1.172** |

The final relativities are between the indicated and current relativities, pulled toward the current by the 40% credibility complement.

> **Exam tip** — The re-basing step exists because relativities expressed relative to the base level do not "add up" to a consistent total — the weighted average should equal 1.0 when measured relative to total, not relative to base. Skipping re-basing introduces a systematic error. The shortcut (no re-basing) is valid only when the complement is "no change" and you're working with change factors rather than relativities.

## Multivariate Classification Methods

Univariate methods analyze one rating variable at a time. Their central limitation is that they cannot simultaneously control for all other rating variables — they only approximate the true effect of any one variable. Modern actuarial practice relies on multivariate methods that analyze all variables simultaneously.

### Why Univariate Methods Fall Short

1. **Exposure correlations** — If young drivers and old cars are correlated, the univariate analysis of car age will be contaminated by the youthful-driver effect. Adjusting for this manually (via the adjusted PP approach) works when there are a few variables, but becomes impractical with dozens or hundreds.
2. **Signal vs. noise** — Univariate results include both systematic effects (signal) and random fluctuations (noise). Multivariate methods use statistical techniques to isolate signal.
3. **No diagnostics** — Univariate methods produce no information about the statistical reliability of results.
4. **Interaction terms** — Multivariate methods can model interaction effects: situations where the effect of one variable differs depending on the level of another.

Three factors drove the adoption of multivariate methods in the late 1990s and early 2000s:
1. Dramatic increases in computing power
2. Improved data warehouse initiatives (better granular data)
3. Competitive pressure — when one insurer adopts GLMs, others face adverse selection if they don't

---

### Minimum Bias Procedures

**Minimum bias procedures** are **iteratively standardized univariate approaches** — a conceptual bridge between one-way analysis and full multivariate modeling. The procedure selects a rating structure (multiplicative, additive) and a bias function; the most common is the **balance principle** (multiplicative structure), which requires that the exposure-weighted indicated loss costs equal the exposure-weighted observed loss costs for every level of every rating variable.

**Steps (multiplicative structure, balance principle):**
1. Equate the exposure-weighted loss costs to the indicated loss costs for each level of each rating variable.
2. Calculate seed relativities for all but the first variable (e.g., use the univariate pure premium relativities).
3. Using the seed relativities for other variables, solve for the relativities of the first variable.
4. Discard the seed for the second variable; use the Step 3 result to solve for the second variable.
5. Repeat until convergence is achieved (relativities stop changing materially between iterations).
6. Re-base relativities to the base level.

*Simple 2-variable example (from Werner & Modlin Ch. 10):* Gender (Male/Female=base) and Territory (Urban/Rural=base). Seed: urban relativity $t_1 = $ total urban PP / total rural PP = 1.86. Solve for gender relativities; then use those to re-solve for territory. After one iteration: male = 1.97, urban = 1.68 (vs. seed of 1.86 — reflecting the heavier male exposure in urban areas).

**Limitations:** Computationally inefficient compared to GLMs; no statistical diagnostics; biased if the balance principle is not satisfied exactly.

---

### Generalized Linear Models (GLMs)

**GLMs** are the industry-standard multivariate method for classification ratemaking. They are a generalization of linear models (LMs) that:
- Allow non-normal response variable distributions (e.g., Poisson for frequency, gamma for severity)
- Allow a **link function** to define a non-linear relationship between predictors and the response (e.g., log link → multiplicative rating structure)
- Remove the constant-variance assumption of LMs

#### Why GLMs Analyze Loss Costs, Not Loss Ratios

GLM analysis is performed on loss cost data (or frequency and severity separately), not loss ratios, for four reasons:

1. Modeling loss ratios requires on-level premium at the *granular* (individual policy) level — practically very difficult.
2. Experienced actuaries have a priori expectations about frequency and severity patterns (e.g., youthful drivers have higher frequencies), making signal easier to identify in loss cost models.
3. Loss ratio models become obsolete when rates or rating structures change — the denominator shifts.
4. There is no standard statistical distribution for modeling loss ratios.

#### Benefits of GLMs vs. Univariate Methods

| Benefit | Description |
|---|---|
| **Automatic exposure-correlation adjustment** | Considers all variables simultaneously; no manual adjustment needed |
| **Signal extraction** | Statistical framework separates systematic effects from noise |
| **Model diagnostics** | Produces standard errors, deviance measures, and cross-year consistency checks |
| **Interaction terms** | Can explicitly model response correlation between variables |
| **Transparent** | Output is a series of multipliers — familiar to actuaries and easy to implement in rating algorithms |

#### GLM Diagnostics

1. **Standard errors** — Measure the variability (uncertainty) associated with each parameter estimate; used to assess statistical significance.
2. **Measures of deviance** — Measure how much the fitted values deviate from actual observations; analogous to R² in a linear model but for non-normal distributions.
3. **Practical (cross-year) diagnostic** — Run the GLM separately for individual years across a multi-year dataset; assess consistency of results over time to detect shifts in the underlying patterns.

#### Practical Considerations

- Ensuring **data adequacy** — garbage in, garbage out (GIGO); GLMs require a homogeneous body of claims
- Identifying **unusual outcomes** that may indicate data errors or structural changes
- Evaluating results from both **statistical and business perspectives** — a statistically significant variable may have no practical or actuarial rationale
- **Communicating model results** effectively to management and regulators

#### Supplemental External Data

GLMs can incorporate data from outside the insurer's own database:
- **Geo-demographics**: population density, income levels, crime indices
- **Weather data**: average rainfall, wind speed, temperature
- **Property characteristics**: construction type, building materials
- **Information about insured individuals or businesses**: credit scores, financial data

---

### Data Mining Techniques

Beyond GLMs, actuaries may use data mining techniques for classification. These include:

1. **Factor analysis** — reduces dimensionality by identifying latent variables
2. **Cluster analysis** — groups observations by similarity
3. **Classification and Regression Trees (CART)** — recursive partitioning of the dataset
4. **Multivariate Adaptive Regression Splines (MARS)** — non-parametric extension of linear models
5. **Neural networks** — flexible but often criticized for being **black boxes** — the lack of transparency makes it difficult to communicate results to regulators or management

> **Exam tip** — GLMs are transparent (output is a series of multipliers) and generate statistical diagnostics — these are the two most frequently tested advantages over minimum bias procedures and other multivariate methods. Data mining techniques like neural networks are criticized specifically for their lack of transparency.

## Special Classifications

Certain rating variables have unique characteristics that require specialized ratemaking procedures beyond standard univariate or multivariate analysis.

---

### Territorial Ratemaking

Geography is one of the most established and important rating variables. **Territorial ratemaking** poses two fundamental challenges:

1. **High correlation with other rating variables** — High-value homes and commercial risks tend to cluster geographically, making univariate analysis of territory heavily biased.
2. **Sparse data per territory** — Geographic units are often defined at the zip code or census block level, and each individual unit has limited data.

#### Phase 1: Establishing Territorial Boundaries

**Step 1: Determine the basic geographic unit.** Common options and trade-offs:

| Unit | Advantage | Disadvantage |
|---|---|---|
| Zip codes | Widely available | Change over time |
| Counties | Static, readily available | Often too large and heterogeneous |
| Census blocks | Relatively static | Requires mapping policies to blocks |

**Step 2: Calculate the geographic estimator.** A multivariate model (GLM) is used to isolate the geographic signal from non-geographic factors. The geographic residuals from the model reveal unexplained geographic patterns.

**Step 3: Smooth the geographic estimators.** Geographic risk tends to be similar for nearby units. Two types of spatial smoothing:

| Smoothing Type | Mechanism | Advantage | Disadvantage | Best For |
|---|---|---|---|---|
| **Distance-based** | Weight nearby units by distance | Easy to understand and implement | Assumes equal urban/rural impact; ignores natural/man-made boundaries | Weather-related perils |
| **Adjacency-based** | Weight by rings of adjacent units (inner rings > outer rings) | Accounts for urban/rural differences; respects physical boundaries | More complex | Socio-demographic perils (theft, etc.) |

Balance over-smoothing (masking real spatial variation) vs. under-smoothing (leaving excess noise in estimates).

**Step 4: Cluster units into territories.** Clustering minimizes within-group heterogeneity and maximizes between-group heterogeneity:

| Clustering Type | Description |
|---|---|
| **Quantile methods** | Create clusters based on equal numbers of observations or equal exposure weights |
| **Similarity methods** | Create clusters based on closeness of estimators |
| — Average linkage | Joins clusters with lower variances |
| — Centroid | More responsive to outliers |
| — Ward's method | Produces clusters with similar numbers of observations |

If contiguous territorial boundaries are required (no "islands"), a **contiguity constraint** must be added to the clustering routine.

#### Phase 2: Determining Rate Relativities

Once boundaries are established, rate relativities are determined using the univariate or multivariate techniques from prior sections. Because territory is highly correlated with other variables, multivariate techniques (e.g., GLMs) are recommended.

> **Exam tip** — Distance-based smoothing is preferred for weather perils (distance = good proxy for weather similarity). Adjacency-based smoothing is preferred for socio-demographic perils (urban/rural distinctions and physical boundaries matter more).

### Increased Limits Ratemaking (ILFs)

Insurance products covering third-party liability offer different **limits of insurance**. The lowest offered level is the **basic limit (B)**; higher levels are **increased limits (H)**. An **increased limits factor (ILF)** is the multiplier applied to the basic limit rate to price a policy at limit H:

$$\text{Premium at H} = \text{ILF}(H) \times \text{Premium at B}$$

#### Why Special Techniques Are Needed

Three reasons ILFs are growing in importance:
1. Growth in personal wealth → individuals need higher coverage
2. General and medical inflation has a **leveraged** (disproportionate) effect on increased limits losses
3. Social inflation (larger jury awards, more lawsuits) disproportionately impacts increased limits

Standard ratemaking techniques struggle with increased limits because **large claims are rare** — the data is sparse and results are volatile.

#### Standard ILF Approach (Uncensored Losses)

Under the simplifying assumptions that (1) all expenses are variable, (2) the profit provision does not vary by limit, and (3) frequency is the same regardless of limit chosen:

$$\text{ILF}(H) = \frac{\text{LAS}(H)}{\text{LAS}(B)}$$

where the **Limited Average Severity (LAS)** is:

$$\text{LAS}(H) = \frac{\text{Total losses capped at H} + (\text{Claims} \geq H) \times H}{\text{Total claim count}}$$

**Worked example (5,000 claims, basic limit $100K):**

| Size Range | Claims | Total Losses |
|---|---|---|
| $0 – $100K | 2,324 | $117,629,223 |
| $100K – $250K | 1,923 | $307,599,929 |
| $250K – $500K | 680 | $222,793,514 |
| $500K – $1,000K | 73 | $43,047,470 |
| **Total** | **5,000** | **$691,070,136** |

$\text{LAS}(\$100\text{K}) = \dfrac{\$117,629,223 + (1,923 + 680 + 73) \times \$100{,}000}{5,000} = \dfrac{\$117,629,223 + \$267,600,000}{5,000} = \dfrac{\$385,229,223}{5,000} = \$77,046$

$\text{LAS}(\$250\text{K}) = \dfrac{\$117,629,223 + (1,923 \times \$175{,}000^*) + (680+73) \times \$250{,}000}{5,000}$

*(Using mid-point approximation $175K for the $100K–$250K layer; precise calculation gives $122,696)*

$$\text{ILF}(\$250\text{K}) = \frac{\$122,696}{\$77,046} = 1.59$$

#### Censored Losses

In practice, the insurer's database only records losses up to the policy limit (censored). Policies with a limit less than the target limit **cannot** be used for the layer above their limit. The layered approach:

$$\text{LAS}(y) = \text{LAS}(x) + \frac{\text{Losses in layer } [x,\ y]}{\text{Total claims from policies with limit} > x}$$

This correctly accounts for the fact that only policies with limit $> x$ contribute observable experience to the layer above $x$.

#### Other Considerations

- **Trend historical losses** before calculating LAS — note the leveraged effect: $\text{Basic Limit Trend} \leq \text{Total Limit Trend} \leq \text{Excess Limit Trend}$
- **Develop losses to ultimate** — large claims often develop differently than small claims
- **Fitted curves** (lognormal, Pareto, truncated Pareto) — smooth random fluctuations in the empirical layer data

> **Exam tip** — ILFs are always ≥ 1.0 (since adding coverage cannot reduce expected losses). If a GLM produces ILFs < 1.0 for higher limits, it's capturing the behavioral effect that higher-limit buyers have lower claim frequencies — not a true "discount" for coverage. Both approaches (empirical LAS and GLM) may be used to triangulate.

### Deductible Pricing: LER Approach

A **deductible** makes the insured responsible for the first dollars of loss, reducing premium in exchange. Two types: flat dollar (most common) and percentage deductibles.

Deductibles are popular because they:
1. Reduce insurance premium
2. Eliminate small nuisance claims (reducing LAE)
3. Provide incentive for loss control (insured bears first-dollar cost)
4. Reduce insurer exposure to catastrophic aggregate losses

#### Loss Elimination Ratio (LER)

The **Loss Elimination Ratio** measures what fraction of ground-up losses is eliminated when a deductible D is applied:

$$\text{LER}(D) = \frac{(L + E_L)_B - (L + E_L)_D}{(L + E_L)_B}$$

The indicated deductible relativity (vs. no-deductible base) is:

$$\text{Indicated Deductible Relativity} = 1 - \text{LER}(D)$$

**Same censoring rule as ILFs:** Only policies with deductibles ≤ D can be used to calculate LER(D) for deductible D.

**Worked example (flat $500 deductible):** Suppose ground-up losses total $100,000 and losses capped at $500 (i.e., losses below the deductible, which would be eliminated) total $12,000. Then LER($500) = 12,000/100,000 = 12.0%, and the indicated deductible relativity = 1 − 0.12 = **0.88** (12% credit vs. no-deductible).

#### Limitations of the LER Approach

1. **Ignores claiming behavior differences** — Insureds with high deductibles may not report small claims they would have reported under a full-coverage policy (moral hazard difference). The LER approach uses ground-up loss data that may not reflect this.
2. **Fails to recognize self-selection effects** — Insureds who choose high deductibles tend to be lower-risk (they're comfortable retaining first-dollar losses). A GLM could capture this behavioral effect; the LER approach cannot.

**When premium credit > deductible amount:** If the LER-based credit produces a dollar premium reduction larger than the deductible amount, the insurer can: (1) cap the dollar credit from the deductible, (2) compute separate credits for different policy types, or (3) use percentage deductibles.

### Workers Compensation Size of Risk

Workers compensation rating algorithms have historically not varied rates by the **size of the risk** (measured by premium or payroll). However, insurers generally agree that smaller risks have worse expected loss ratios than larger risks.

#### Why Small Risks Have Worse Experience (3 Theories)

1. Small companies often have **less advanced safety programs**.
2. Small companies may **lack return-to-work programs** designed to help injured employees return to work, leading to longer disabilities.
3. Small insureds may have **less financial motivation** to prevent or manage injuries compared to large insureds (who are more likely to be experience-rated and bear the financial consequences of their own losses).

#### Expense Inequity from the All Variable Expense Method

When all underwriting expenses are treated as a percentage of premium, the same variable expense rate applies to a $1,000 policy and a $1,000,000 policy. In practice, many expenses (e.g., policy issuance, basic account management) are roughly fixed regardless of premium size. This means:
- The All Variable Expense Method **undercharges** large risks for fixed expenses (fixed costs spread across large premium → small %)
- The All Variable Expense Method **overcharges** small risks for fixed expenses (same fixed cost spread across small premium → large %)

#### Three Approaches to Fix Expense Inequity

1. **Cap the variable expense provision** — Apply the variable expense percentage only to the first $X of standard premium; above that threshold, no variable expense load.
2. **Charge an expense constant** — A flat dollar fee added to all risks, regardless of size. This fee represents the fixed costs that don't scale with premium.
3. **Apply a premium discount** — Reduce the premium for policies exceeding a specified size, reflecting the lower average cost per dollar of premium for large risks.

#### Loss Constants

Even after addressing expense inequity, small risks have higher expected loss ratios. A **loss constant** is a flat dollar amount added to the premium of small risks to equalize the expected loss ratio across all insured sizes. The loss constant is calculated by finding the dollar amount needed to bring the small-risk expected loss ratio down to the target (large-risk) level.

### Insurance to Value (ITV)

**Insurance to value (ITV)** describes the relationship between the face amount of insurance selected (F) and the full replacement cost or value of the insured item (V). If F = V, the property is insured to full value. If F < V, it is underinsured.

#### Two Problems with Underinsurance

1. **Coverage inadequacy** — In the event of a total loss, the insurance payment (capped at F) will not be sufficient to restore the insured to the pre-loss condition.
2. **Rate inequity** — If the insurer calculates rates assuming all properties are insured to full value, then an underinsured property will pay less premium than warranted by its expected loss cost. This subsidizes underinsured risks at the expense of fully-insured risks.

> **Key insight** — Rate inequity only exists when **partial losses are possible**. If all claims are total losses, the rates are still equitable (because capped losses and premiums are proportional), though adequacy may still be an issue.

#### Coinsurance Clause

A **coinsurance clause** requires the insured to carry a minimum level of insurance (e.g., 80% of replacement cost) or face a reduction in the indemnity payment proportional to the degree of underinsurance. Notation:

| Symbol | Definition |
|---|---|
| $F$ | Face value (amount of insurance selected) |
| $V$ | Full value or replacement cost of insured item |
| $c$ | Required coinsurance percentage (e.g., 80%) |
| $cV$ | Coinsurance requirement (minimum required coverage) |
| $a$ | Apportionment ratio |
| $L$ | Amount of loss (after deductible) |
| $I$ | Indemnity payment received by insured |
| $e$ | Coinsurance penalty |

Key formulas:
$$a = \min\!\left(\frac{F}{cV},\ 1\right)$$
$$I = \min(aL,\ F)$$
$$e = \min(L, F) - I$$

A non-zero penalty occurs when: (1) $L > 0$, (2) $F < cV$ (underinsured), and (3) $L < cV$ (loss is below the coinsurance requirement).

#### Worked Example

Home value V = $500K, insured for F = $300K, coinsurance requirement c = 80% → cV = $400K.

Since F = $300K < cV = $400K, there is a coinsurance deficiency. Apportionment ratio: $a = 300/400 = 0.75$.

| Loss Amount (L) | I = min(0.75 × L, F) | e = min(L, F) - I | Notes |
|---|---|---|---|
| $200,000 | min(150K, 300K) = **$150,000** | 200K − 150K = **$50,000** | Partial loss below face |
| $300,000 | min(225K, 300K) = **$225,000** | 300K − 225K = **$75,000** | Partial loss at face |
| $350,000 | min(262.5K, 300K) = **$262,500** | 300K − 262.5K = **$37,500** | Loss exceeds face; penalty decreases |
| $450,000 | min(337.5K, 300K) = **$300,000** | 300K − 300K = **$0** | Loss ≥ cV; no penalty (capped at F) |

The coinsurance penalty is largest when the loss equals the face amount (F). For losses ≥ cV, the penalty is zero — the insurer simply pays the face amount.

#### Other Approaches to ITV

**Varying rates by ITV level:** Rather than penalizing losses after the fact, charge a *higher rate per $1,000 of insurance* to underinsured homes. Since partial losses are more frequent than total losses (left-skewed distribution for homeowners), underinsured homes generate a disproportionately high loss per dollar of insurance.

**ITV initiatives:**
1. **Guaranteed replacement cost (GRC)** — Covers replacement even if it exceeds the policy limit, often with a cap (e.g., 125% of policy limit).
2. **Property estimation tools** — Modern component-indicator tools assess specific home features to estimate replacement cost more accurately.
3. **Property inspections, indexation clauses, and insured education** — Encourage annual updating of insured values.

> **Exam tip** — The coinsurance penalty is zero for losses ≥ cV (the insurer simply pays $F and there is no gap). The penalty is also zero when the property is insured to full value ($F \geq cV$). The largest penalty occurs at the loss amount equal to $F$ — so a smaller face amount means both a larger (relative) penalty and a smaller absolute maximum penalty.

## Credibility

**Credibility** is the weight given to observed experience vs. a complement (related experience or a prior estimate):

$$\hat{\mu} = Z \times \text{Observed Experience} + (1 - Z) \times \text{Complement of Credibility}$$

---

### Criteria for the Credibility Measure Z

Any credibility measure Z must satisfy three basic criteria:

1. $0 \leq Z \leq 1$ — credibility is bounded; you cannot give more than full weight to observed data or negative weight.
2. As the size of the risk (number of observations) increases, Z should **increase** — more data → more reliable.
3. As the size of the risk increases, Z should increase at a **decreasing rate** — the marginal value of additional data diminishes.

---

### Classical (Limited Fluctuation) Credibility

The most widely used credibility method in insurance ratemaking.

**Goal:** Limit the effect of random fluctuations on the risk estimate.

**Full credibility standard:** The observed experience is considered fully credible (Z = 1.0) when there is a probability $p$ that the observed experience is within $k$% of the true expected value. Under simplifying assumptions (homogeneous exposures, Poisson claim occurrence, constant severity):

$$E(Y) = \left(\frac{z_{(1+p)/2}}{k}\right)^2$$

where $z_{(1+p)/2}$ is the standard normal value at the $\frac{1+p}{2}$ percentile.

| k | p | $z_{(1+p)/2}$ | Claims Needed |
|---|---|---|---|
| 5% | 90% | 1.645 | **1,082** |
| 5% | 95% | 1.960 | **1,537** |
| 10% | 90% | 1.645 | 271 |
| 10% | 95% | 1.960 | 384 |

**With variation in severity** (relaxing the constant-severity assumption):

$$E(Y) = \left(\frac{z_{(1+p)/2}}{k}\right)^2\!\left(1 + \frac{\sigma_s^2}{\mu_s^2}\right)$$

where $\sigma_s^2/\mu_s^2$ is the squared coefficient of variation of the severity distribution. More variability in severity → higher standard for full credibility.

**Square root rule** (when Y < E(Y)):

$$Z = \sqrt{\frac{Y}{E(Y)}}$$

where Y = observed claim count. If $Y \geq E(Y)$, then $Z = 1.0$.

**Worked example:** Observed pure premium = $200 based on 100 claims. Complement = $300. Standard: p = 90%, k = 5%, so E(Y) = 1,082.

$$Z = \sqrt{\frac{100}{1{,}082}} = 0.304 \approx 0.30$$

$$\hat{\mu} = 0.30 \times \$200 + 0.70 \times \$300 = \$60 + \$210 = \$270$$

**Advantages:** Most commonly used; data is readily available; straightforward calculations.

**Disadvantages:** Derivation requires simplifying assumptions (Poisson frequency, constant severity); uses a selected complement but does not account for the quality of the complement relative to the observed data.

---

### Bühlmann (Least Squares) Credibility

Also called **least squares credibility** because it minimizes the squared error between the estimate and the true expected value.

$$Z = \frac{N}{N + K}, \quad K = \frac{\text{EVPV}}{\text{VHM}}$$

| Symbol | Definition |
|---|---|
| N | Number of observations |
| EVPV | Expected Value of Process Variance — avg variability *within* a risk over time |
| VHM | Variance of Hypothetical Means — variability *between* risks |
| K | Ratio of average within-risk variance to between-risk variance |

**Intuition:** If the risks in the portfolio are very similar to each other (small VHM), there is little information to gain from a single risk's experience — Z → 0. If the risks differ greatly (large VHM), individual experience is more informative — Z → higher. K captures this trade-off.

**Compared to classical credibility:**
- Bühlmann Z approaches 1.0 **asymptotically** as N grows
- Classical Z reaches 1.0 at a finite standard $N_f$ and stays there
- For practical purposes: Bühlmann is more theoretically grounded; classical is simpler and more commonly used

**Worked example:** N = 21 observations, EVPV = 2.00, VHM = 0.50. Prior mean = $225.

$$K = \frac{2.00}{0.50} = 4.00, \quad Z = \frac{21}{21 + 4} = \frac{21}{25} = 0.84$$

$$\hat{\mu} = 0.84 \times \$200 + 0.16 \times \$225 = \$168 + \$36 = \$204$$

**Challenge in practice:** Determining EVPV and VHM requires either a parametric model (subject to model error) or empirical estimation from data (subject to sampling error).

---

### Bayesian Analysis

Bayesian credibility adjusts the prior estimate to reflect new information using **Bayes Theorem** in a probabilistic manner — there is no explicit formula for Z. The new information is introduced via conditional probabilities rather than through a simple weighting factor.

- More theoretically elegant than classical or Bühlmann
- **Not commonly used in practice** due to computational complexity
- Under special distributional assumptions, the Bayesian estimate equals the Bühlmann (least squares) credibility estimate

---

### Desirable Qualities of a Complement of Credibility

Once Z is determined, the actuary must select the **complement of credibility** — the "related experience" or prior mean that receives weight $(1-Z)$. The complement matters enormously: if Z is low (sparse data), most of the rate is driven by the complement.

Per Boor (2004), a good complement of credibility has six qualities:

| Quality | Description |
|---|---|
| **1. Accurate** | Low error variance around future expected losses — consistently close to the true value |
| **2. Unbiased** | Does not systematically over- or under-estimate; differences average to zero over time |
| **3. Statistically independent** | Not correlated with the base statistic — errors in the base won't be compounded |
| **4. Available** | Data exists and can be obtained |
| **5. Easy to compute** | Calculations are straightforward and explainable to third parties (e.g., regulators) |
| **6. Logical** | Has an explainable relationship to the base statistic |

> **Important distinction** — *Accurate* and *unbiased* are not the same thing. An accurate complement is consistently close to the true value (low variance) but may be biased (systematically too high or low). An unbiased complement varies randomly around the true value but may have high variance (not always close). The ideal complement is both, but the trade-off is real.

---

### Complements for First Dollar Ratemaking

**First dollar ratemaking** covers claims from the first dollar of loss (or after a small deductible), as opposed to excess/reinsurance layers. Six common complement methods:

| Method | Key Feature | Accurate? | Unbiased? | Independent? |
|---|---|---|---|---|
| **1. Loss costs of larger group including subject** | Uses the broader group's observed loss cost directly | Yes | Likely biased (group ≠ subject) | If subject doesn't dominate group |
| **2. Loss costs of larger related group** | Uses a separate but similar group | May be inaccurate | Generally biased (direction unknown) | Usually yes |
| **3. Rate change from larger group applied to present rates** | $C = \text{Current LC}_{subject} \times \dfrac{\text{Indicated LC}_{larger}}{\text{Current LC}_{larger}}$ | Accurate over long run | Generally unbiased | Depends on size of subject vs. group |
| **4. Harwayne's method** | Adjusts related-state experience to match subject state's class distribution | Yes | Unbiased | Mostly independent |
| **5. Trended present rates** | Two adjustments here: Prior Indication & Trend:  $C = \text{Present Rate} \times \text{Trend Factor} \times \dfrac{\text{Prior Indicated LC}}{\text{Prior Implemented LC}}$ | Depends on accuracy of historical costs | Unbiased | Depends |
| **6. Competitor's rates** | Uses competitor's filed rates | May be inaccurate | May be biased | Yes |

**Harwayne's Method** (5 steps) — used when subject experience and related experience have different class distributions (e.g., ratemaking for Class 1 in State A using other states' data):
1. Calculate the weighted average pure premium for State A (using State A's exposure distribution).
2. Calculate the weighted average pure premium for each other state, *reweighted using State A's exposure distribution by class*.
3. Calculate an adjustment factor for each other state: State A average PP / other state reweighted PP.
4. Apply the adjustment factor to Class 1's pure premium in each other state.
5. Calculate the complement as the weighted average of the adjusted Class 1 pure premiums from other states.

---

### Complements for Excess Ratemaking

**Excess ratemaking** prices layers of loss above a high attachment point (e.g., umbrella, excess reinsurance). The base statistic may be very sparse or even zero. Four complement methods:

| Method | Description | Accuracy | Bias |
|---|---|---|---|
| **1. Increased limits analysis** | Uses ILF differentials: $C = L_x \times \dfrac{\text{ILF}_{A+L} - \text{ILF}_A}{\text{ILF}_A}$ | Inaccurate if loss severity distribution differs | Biased if distribution differs |
| **2. Lower limits analysis** | Similar to (1) but uses lower-limit data to estimate the excess layer | More available; more biased than (1) | More biased than (1) |
| **3. Limits analysis** | Combines data from all available limits (expected losses derived from premium, ILFs used to allocated expected losses to the layer); time-consuming but straightforward | Same level as (1) and (2) | Same as (1) |
| **4. Fitted curves** | Fit a parametric distribution (e.g., Pareto) to loss data and integrate for the excess layer | Most accurate *if* the fitted curve replicates actual data | Least biased of the four; most complex |

> **Exam tip** — For first dollar ratemaking, the rate change from the larger group applied to present rates (#3) is generally the least biased because it adjusts the current level and doesn't inherit structural differences between the groups. For excess ratemaking, fitted curves are theoretically superior but computationally complex; they may be less independent than the other methods because fitting the curve uses the same underlying data as the base statistic.

## Credibility in Ratemaking: Complement Selection and Justification

Credibility formulas (classical and Bühlmann) tell you **how much weight** to assign to observed experience. But the equally important question is: **what is the complement?** The complement drives the credibility-weighted estimate whenever $Z < 1.0$, so selecting and justifying the right complement is a core actuarial judgment task — and a common exam topic.

---

### Boor's Six Desirable Qualities for a Complement of Credibility

James Boor identified six characteristics of a good complement. A strong complement should be:

| # | Quality | Explanation |
|---|---|---|
| 1 | **Accurate** | On average, should be close to the true expected value for this risk/class |
| 2 | **Unbiased** | Should not systematically over- or understate the true value |
| 3 | **Independent** | Should not be correlated with the observed experience being blended (otherwise, credibility reduces risk less than expected) |
| 4 | **Available** | Must be obtainable in practice at reasonable cost and timeliness |
| 5 | **Easy to compute** | Should not require complex or opaque calculations that reduce auditability |
| 6 | **Logical relationship** | The complement and the observed experience should be related — similar exposures, similar loss characteristics |

> **Exam tip:** The most commonly tested qualities are **accuracy** (does the complement reflect true costs for this risk?), **independence** (is it correlated with the observed data?), and **logical relationship** (is the complement a valid reference group?). You may be asked to evaluate a proposed complement against these criteria.

---

### Common Complements in Classification Ratemaking

| Complement | Description | When Appropriate |
|---|---|---|
| **Current rates (no change)** | Treat the existing relativity as the complement; blend indication vs. 1.0 | Stable lines where existing rates are believed to be reasonable |
| **Competitor rates** | Use rates filed by a dominant competitor as the complement | When the competitor has a larger, more credible book in the same segment |
| **Industry data** | ISO, NCCI, or other bureau rates/relativities | Commodity lines where bureau data is available and applicable |
| **Statewide or countrywide data** | Aggregate indications from the full book; complement for a specific class or territory | When the class is a sub-segment of the total book |
| **Complement from complementary criteria** | Derive a logical expected relativity from related variables | E.g., use house age as a proxy for fire risk when direct fire data is sparse |
| **Trended prior period** | Apply a trend factor to last period's indication | For lines with stable trends where data is too thin for a new indication |

---

### Application in Ratemaking: Blending a Class Indication

**Setup:** You have an indicated relativity for Class A from one-way analysis, but the class is small. You want to blend it with the complement.

**Given:**
- Observed claims: Y = 150
- Full credibility standard: E(Y) = 1,082 (at p=90%, k=5%)
- Indicated relativity: 0.75
- Complement (statewide): 0.90

**Step 1: Credibility**
$$Z = \sqrt{\frac{150}{1{,}082}} = \sqrt{0.1386} = 0.372$$

**Step 2: Credibility-weighted relativity (to total)**
$$R^* = 0.372 \times 0.75 + (1 - 0.372) \times 0.90 = 0.279 + 0.565 = 0.844$$

**Step 3: Rebase to base class** *(if needed — divide by the base class's credibility-weighted relativity)*

The blended result of **0.844** represents what Class A's relativity should be, given both the sparse observed data and the statewide benchmark.

---

### Justifying the Complement in an Exam Essay

When asked to justify a selected complement, address three things:

1. **Logical relationship:** "The statewide indication represents the same line of business, coverages, and policy terms — it is a reasonable prior estimate for any class within the book."
2. **Independence:** "The statewide data is dominated by classes other than Class A, so the complement is not materially correlated with Class A's own observed experience."
3. **Accuracy/lack of bias:** "The statewide indication is based on a large, credible volume of data and is not expected to systematically over- or understate Class A's true costs."

> **Exam tip:** Never state that the complement is "better" than the observed data — the point is that **neither alone is fully reliable**; credibility blending combines both optimally. The complement is only as good as its logical relationship to the target class. A complement from a completely different line of business (e.g., using commercial auto data as a complement for homeowners) would fail Boor's "logical relationship" criterion.

---

### Credibility in the Rate Level Indication

Credibility applies not just to classification relativities but to the **overall rate level indication** itself, especially for small companies or niche lines:

$$\text{Indicated Change} = Z \times \text{Indicated Change (Own Data)} + (1 - Z) \times \text{Indicated Change (Industry)}$$

When own-company data is thin (e.g., a new entrant or a highly specialized book), regulators and management alike expect the actuary to explain how much weight is given to industry benchmarks vs. company experience, and to justify that choice using the Boor criteria.
