Learning Task A17: Understand the considerations for implementing rates to achieve an organization's goals (e.g. non-pricing solutions, minimum premium, rating algorithms).


### Not Every Problem Needs a Rate

The manager cried, "We are losing good risks!"  
And everyone reached for their pricing spreadsheets.  
But the actuary paused and she asked with some care,  
"Is price really broken, or is something else there?"  

If service is slow and the quotes take too long,  
If billing is clumsy, if claims handling's wrong,  
Then cutting the rate may not fix what is real,  
A non-pricing solution may better heal.  

Maybe tighten the classes, improve appetite,  
Maybe change how the agents submit every risk right;  
Maybe fix inspections, renewals, or speed,  
For goals are achieved by more than pure need.  

So implementation means choosing the tool,  
Not worshiping blindly the rate-making rule;  
Price is one lever, but not the whole slate,  
And not every problem needs only a rate.  

### Algorithms, Floors, and Goals

The filed indication may say minus three,  
But the company still needs a minimum fee;  
Tiny accounts still consume time and expense,  
So a minimum premium can make perfect sense.  

A rating algorithm layers factor on factor,  
Base rate, class, territory, every multiplier after;  
The sequence must work and the logic be clear,  
Or odd results for policyholders appear.  

Caps and floors may be used when the rollout is tough,  
To keep swings from being too sharp and too rough;  
But every constraint should be tested with thought,  
So the business goal reached is not dearly bought.  

Organization goals can be growth, mix, or gain,  
Retention, new business, or smoother terrain;  
Implement rates with the whole system in view,  
And let judgment and structure work jointly for you.
