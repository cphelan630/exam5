Learning Task A10: Demonstrate proper handling of extraordinary losses for ratemaking purposes (e.g. large losses, catastrophes).


### Cap the Shock

In the land of the rates where the pure premiums grow,  
there lived a great loss that had put on a show.  
Five million it cost, in a year of its own,  
while the rest of the years barely averaged a groan.  

"Include it!" cried one. "It is part of the past!"  
But the rates that came out were too high to forecast.  
The policyholders screamed as they ran for the door,  
for the rate was three times what it had been before.  

"Exclude it!" cried two. "It will never come back!"  
But the actuary paused and then stepped off that track.  
"A shock CAN recur," said the actuary with care,  
"we must price for the cost, not pretend it's not there."  

So they capped every loss at five hundred K flat,  
and loaded for excess — imagine that!  
The ILF told them what excess would cost,  
so the rate held the truth and no dollar was lost.  

Cap the shock at the threshold and load for the rest,  
that is the method that passes the test.  
Not too high, not too low — the expected is right,  
and the rate tells the future what losses look like.


### Dealing with Large Losses and Anomalies for Ratemaking  

When shock losses thunder and rattle the page,  
You do not let one year set the rate for an age.  
Three tools stand ready to steady the floor:  
Cut them out wholly, or cap them, or cap them once more.  

One says, “Throw out the giants; they distort what we see.”  
One says, “Keep only the basic-limit piece with me.”  
One says, “Choose some other threshold where volatility starts,  
Keep what lies under, remove what departs.”  

But none of these methods may end with “take away.”  
You must add back expected large loss in a stable, long way.  
For if you remove every giant and never restore,  
Your rates understate what the future has in store.  

**Excluding shock losses entirely**

The first road is brutal, clean, and severe:  
“If this claim is a shock loss, it does not belong here.”  
Take the whole thing out, let the historical years breathe,  
Remove the distortion the giants can weave.  

The mechanics are simple, almost too neat:  
Identify the shock claims, delete them from the sheet.  
Then study the remaining, more stable base,  
And later add a long-run provision in their place.  

The advantage is clarity, speed, and force,  
A sharp reduction in random short-course.  
For small books, one giant can poison the well,  
And this method keeps that from driving the spell.  

But the weakness is obvious, maybe too much:  
It throws away signal along with the crutch.  
A big loss is rare, but not wholly absurd;  
By dropping it all, some truth may be blurred.  
And if your thresholding judgment is loose,  
You may excise too much and bias the use.  

Its assumption is strong and should trouble your head:  
That the removed claims are “special” enough to be shed,  
And that a later provision, built over more years,  
Will restore the right average despite short-run fears.  

**Capping at the basic limit**

The second road says, “Keep part of the beast.  
Retain the basic-limit layer at least.”   
So a million-dollar claim, if the cap is one hundred,  
Becomes one hundred retained, the rest now sundered.  

The mechanics are layered and exam-worthy too: 
Cap each claim at the basic limit when rebuilding the view.  
The kept part becomes basic-limit loss for the rate,  
The removed part becomes excess, handled later in fate.  
Then if using a loss-ratio indication, be warned:  
Premium must also be rerated so basic-limit is formed.  

The advantage is structure, especially when rates  
Are really base rates before limits create  
The classification differences layered above;  
This matches the pricing framework you love.  

But the disadvantage comes quickly in sight:  
The basic limit may be far too low for the right  
Point where volatility truly begins to distort;  
Many ordinary claims may exceed it in court.  
Then you are not just removing the strange and the wild,  
You are chopping out too much of the normal compiled.  

Its assumption is this: that the base rate should price 
Only the basic-limit layer, nothing more precise;   
And that higher layers are better deferred  
To increased limits work, where those costs are inferred.  

**Capping at some other selected large-loss threshold, apply excess loading factor**

The third road is subtler: “Do not worship the base.  
Choose a threshold that better fits this actual case.”  
Maybe the 99th percentile, maybe some share   
Of insured value, maybe where the tail gets rare.  

The mechanics: pick a threshold meant to divide  
Ordinary severity from volatility’s tide.  
For each large claim, keep the loss up to the cap,  
Call that non-excess; remove the rest from the map.  
Then over a longer horizon, steady and broad,  
Compute excess over non-excess, not trusting one odd   
Year to define what the future should be;  
Apply that loading back to the stable loss sea.   

The advantage is balance: more losses stay in,  
Yet the wildest volatility does not still win.  
This often fits better than basic-limit art,  
Because the cap is chosen where instability starts.  

But the weakness is judgment, and judgment can drift.  
Pick the cap too low, and too much gets a lift.  
Pick it too high, and the volatility stays,  
Still polluting indications in sharp, noisy ways.  
And if policy size varies a lot by design,  
One fixed threshold may not really align. 

Its assumption is gentler than method two’s creed:  
That there is some practical point where you need  
To split ordinary from extraordinary pain,  
And that long-run tail ratios are stable enough to sustain.  

### Adjusting the cap for changing severity levels

A cap set in old dollars will not hold its form,  
When severity rises through inflation’s storm.  
If claims are trended upward but the threshold stands still,  
Too much slides above it by mere trend, not by will.  

So two roads appear for the cap through time:  
One trends the losses first, then caps at the line.  
Take ground-up claims, lift them to future cost air,  
Then apply the fixed threshold once both are up there.  

The other road keeps losses where history began,  
But trends the threshold itself by an indexed plan.  
Raise the cap with severity, year after year,  
Then apply that moving cap to old losses right here.  

Both aim at one truth in a different disguise:  
Put claim and threshold on comparable size.  
For if only one moves and the other stays old,  
Your excess provision becomes biased and cold.  

And note one more wrinkle the syllabus may bring:  
Higher layers may leverage inflation’s sting.  
The tail can inflate faster than the body below,  
So crude handling here may distort the whole show.    

So whether you trend claims, or trend caps instead,  
The point is the same, and keep it in your head:  
A threshold in yesterday’s dollars cannot  
Judge today’s trended losses the way that it ought.  

### The ILF and the Cap

There once was a loss that was fabulously large,  
it sailed through the data like a battleship barge.  
It swamped every year and it doubled the rate,  
and the policyholders said, "This is simply not great."  

The actuary arrived with a cap in his hand,  
"Five hundred K," said he, "that's where we stand.  
Above that, I'll load — I know what excess will cost,  
the ILF table ensures nothing is lost."  

The ILF said at a million: one-six-nine-eight.  
At five hundred K: one-five-oh-one, and that's great.  
The difference times basic means excess per claim,  
times frequency — now we can play the right game.  

And the cat, said the actuary, is different indeed,  
it corrupts the whole book with remarkable speed.  
Remove it from history, load from the model,  
the vendor will tell you just what's in the coddle.  

So cap the large loss and remove every cat,  
load them back with the tools, and remember all that:  
the rate must reflect what the FUTURE will bring,  
not whatever odd loss the past year happened to sing.
