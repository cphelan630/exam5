## Which Base to Choose?

Two bases are sitting and waiting to be  
The unit that measures your risk, you see.  
Base A and Base B are both vying for fame -  
How do you pick? Well, you play this game:  

First ask: does it GROW when the expected loss grows?  
Does more of this number mean more of those woes?  
If double the base gives you double the claim,  
Proportionality is there - a strong, sturdy frame.  

Then ask: is it PRACTICAL, easy to get?  
Before the policy starts, is the number all set?  
Can the insured manipulate, fudge it, or cheat?  
If yes, find another - that base cannot compete.  
 
And last: has the INDUSTRY done this before?  
Have regulators nodded and opened the door?  
A base with no history is hard to defend  
When the regulator asks you to justify, friend.  

So miles are proportional - they grow with the drive,  
But car-years are practical, helping rates thrive.  
For auto, car-years wins the three-part test score;  
Miles fails practicality - we have said this before.  

For workers comp, it is payroll - not employee count -  
Because wages move with the benefit amount.  
More pay means more claims cost if someone gets hurt;   
Employee count misses the wage-driven hurt.  

So check all THREE criteria, one, two, and three,  
Write them all down - that is what earns you the fee.  
"Proportional, practical, precedent" - say it out loud,  
And walk into exam day confident, proud.
