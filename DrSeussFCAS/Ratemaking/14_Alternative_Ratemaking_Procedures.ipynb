Learning Task A14: Perform calculations related to alternative ratemaking procedures (e.g. classification, territory, deductibles, increased limits, coinsurance, commercial lines rating mechanisms).


### Every Risk Has Its Rate

The overall rate change is needed, no doubt,  
but the class and the territory need to be sorted out.  
Class A has a loss ratio twice that of B,  
and if they pay the same rate, B will flee.  

For B will go shopping and find a better deal,  
from a company that prices with actuarial zeal.  
And A will remain, overpaying no more,  
while the adverse selection walks in through the door.  

So every risk has its rate — that's the rule,  
the relativity system is the actuary's tool.  
Pure premium by class, divided by base,  
gives the relativity factor to put in its place.  

The territory is spatial, the class is a trait,  
but both are just factors that multiply the base rate.  
Compute them from data, apply credibility too,  
and every risk pays what is actuarially true.


### One Rate Does Not Fit All

One rate does not fit all — the insurer learned fast,  
when the good risks departed and the bad ones amassed.  
Classification and territory sorted them out,  
but commercial lines needed a different route.  

Schedule rating inspects what you currently see:  
the sprinklers, the training, the safety decree.  
No losses required — it's prospective and true,  
a debit for hazards, a credit for new.  

Experience rating looks back at the past:  
your own three-year losses, first, second, and last.  
A modification — a credibility blend,  
of your actual results with the manual trend.  

Retro rating is strangest and boldest of all:  
it waits for the year-end to answer the call.  
The premium adjusts based on losses that came,  
a minimum, maximum, and a converted-loss game.  

Schedule for the future, experience for the past,  
retro for the sharing that makes premiums last.  
Three tools for the commercial lines actuary's kit —  
for one rate does not fit all, and that's the right fit.


### Deductibles, Limits, and Shares

A deductible nibbles the small losses first,  
So loss elimination tells how much is reversed;  
The credit is not just a discount by feel,  
It comes from the losses the deductible steals.  

Increased limits move upward from basic to high,  
So ILFs tell how severities climb;  
The bigger the limit, the more loss you keep,  
And the premium follows that laddering steep.  

Coinsurance says: carry enough of the value,  
Or a partial loss payment may not fully pay you;  
The insured shares the pain when the limit is shy,  
So rating must notice that feature nearby.  

Deductibles, limits, and sharing of cost,  
Are all rating mechanisms lest adequacy's lost;  
One base rate begins it, but factors then call,  
Because one simple rate cannot fit them all.
