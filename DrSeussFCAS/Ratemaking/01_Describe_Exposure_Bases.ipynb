## Find Your Base

Oh, find your base! Your unit! Your thing!  
The number that makes your pure premium sing.  
It must grow with the losses — proportional, true —  
So double the risk means double the due.  

A car on the road for a year and a day  
Is a car-year of driving — that's how we say  
"This much exposure, this much of the cost" —  
Without this proportion, the signal is lost.  

The second great rule is the practical rule:  
Can you MEASURE it cheaply? Is it a useful tool?  
Can the insured manipulate it, game it, or fake?  
If yes, then that base is a terrible mistake.  

Miles-driven is fine as a theoretical tale,  
But how do you VERIFY it? Odometers can fail.  
The insured could lie, could under-report,  
So car-years we use, of the practical sort.  

The third is tradition — historical weight,  
The industry uses it, regulators relate.  
A base with long history is easy to defend;  
It's trusted by everyone, start to end.  

So: PROPORTIONAL, PRACTICAL, PRECEDENT — three!  
The criteria locked in your memory, you see.  
On exam day you'll list them with confident flair,  
And prove that you chose your exposure with care.
