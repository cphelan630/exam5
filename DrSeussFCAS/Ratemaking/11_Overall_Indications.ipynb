Learning Task A11: Construct an overall rate level indication using the pure premium and loss ratio methods.

### Big-picture poem: two roads, one answer

Two roads can price the rate we need,  
If losses, LAE, and expenses lead.  
One starts with **loss per exposure** in view,  
The **pure premium** path comes straight on through.  

The other asks what **premium now** can bear,  
And builds a **permissible loss ratio** there.  
Different starting points, same actuarial aim:  
An overall indication for the future rate change.  

### Pure premium method

Start with losses **per exposure** first,  
Trend and develop, quench the burst.  
Add LAE and fixed expense too,  
Then profit and variable loads come due.  

A future **pure premium** stands in sight,  
Compared to current average premium right.  
Indicated change is what rates must do,  
When expected cost per exposure shines through.  

This method feels natural when exposure is strong,  
When premium history may have gone wrong.  
If premium is noisy, but losses are sound,  
Pure premium keeps your footing on ground.  

### Loss ratio method
Take premium first, but set it straight,  
On-level and trend it to future rate.  
Take losses too, made future-clean,  
Then form the **projected loss ratio** scene.  

But not every premium dollar can pay  
For losses alone in the future day.  
Expenses and profit must claim their share,  
So a **permissible loss ratio** waits there.  
  
Compare projected to permissible line,  
If actual is higher, rates must climb.  
If actual is lower, rates may fall—  
That’s the loss ratio method after all.  

### Equivalence

Do not be tricked by different dress,  
These methods should not truly guess  
At different answers if inputs align—  
They’re two forms of the selfsame spine.  

Same future losses, same expense view,  
Same trend dates and on-leveling too.  
If one says up and one says down,  
Some mismatch likely lurks around.  
  
Maybe premium was handled one way here,  
But not the same in the sister gear.  
Check consistency first, then try once more—  
Equivalent methods should agree at the core.  

### Exam-trap poem

Develop the losses, then trend them ahead,  
But do not let one-time changes hide in the spread.  
On-level premium; do not forget,  
Old rate levels can poison the set.  

Use matching dates from where to where,  
And keep fixed versus variable clear.  
If the indication feels oddly bizarre,  
Check whether your denominators are  
On the same basis, clean and true—  
Most exam mistakes start there, not new.  


### Quick Poem

Pure premium asks: “What cost per risk?”  
Loss ratio asks: “How much premium’s at risk?”  
One builds up from exposure-side pain,  
One checks whether premium can carry the strain.  

Pure premium: future cost / current avg prem view.  
Loss ratio: projected LR / permissible one due.  
Different lenses, same final mission:   
Find the overall indicated rate revision.  


### When to use which 

If exposures are stable, meaningful, strong,  
Pure premium may sing the cleaner song.  
If premium data are central and clean,  
Loss ratio may better fit the scene.  

But on Exam 5, know both with ease,  
Not just the formulas meant to please.  
Know what is adjusted, why, and when—  
That is how you collect the points in the end.  


### Two Methods, One Truth

There are two ways to find what the rate ought to be,  
one starts with exposures, one starts with EP.  
The pure premium method says: what does one unit cost?  
The loss ratio says: how much was in the past tossed?  

The pure premium actuary opens her spreadsheet with glee,  
trending the losses and developing with fee.  
She divides by the exposures to get the pure PP,  
then divides by the PLR — that's the indicated, you see!  

The loss ratio actuary opens a neighboring file,  
he on-levels the premium with care and with style.  
He divides all the losses by on-leveled EP,  
then divides by the PLR, same answer — whee!  

"How can this be?" cried a student one night,  
"Two methods, same answer — is this magic or sleight?"  
"No magic," said both actuaries with pride,  
"it's algebra, friend — they're the same thing inside."  

So learn both the methods, for the exam will ask one,  
and sometimes it asks for both — better have fun.  
The truth is the truth whether pure PP or LR,  
one truth, two approaches, one prospective rate star.


### Which Method Today?

A student once asked, "Which method do I need?  
There are two on the page and I'm trying to read.  
The pure premium method, the loss ratio too —  
how do I know which is the right one to do?"  

"It's simple," said the actuary, "ask yourself this:  
is the premium on-leveled? You won't want to miss.  
If the rates have been stable and the book's in good shape,  
use the loss ratio method — it's a well-tested cape."  

"But if you've been reclassifying and changing the mix,  
or starting from scratch with no premium to fix,  
then the pure premium method is the one you should choose,  
for the premiums are muddled and there's nothing to use."  

The student nodded and wrote it all down:  
New class → pure premium, stable → loss ratio crown.  
Both give the same answer when the data is true,  
but the right choice of method depends on what's askew.  

On-level the premium! the actuary cried,  
don't use the raw EP — it will lead you aside.  
The whole loss ratio method stands or it falls,  
on on-leveled premium in those denominator halls.
