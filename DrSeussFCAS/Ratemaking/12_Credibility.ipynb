Learning Task A12: Apply credibility to ratemaking analyses using different methods and justify the choice of complement.


### Z Is for Credibility

Z is for credibility, great and small,  
it blends what you saw with what you'd expect from them all.  
When the data is thin and the variance is wide,  
bring in the complement to sit at your side.  

Z times your data, plus one-minus-Z times the rest,  
that is the credibility method — the blend that is best.  
Not zero for the part that does not come from you,  
but M times one-minus-Z — the complement comes through.  

When n is a thousand and k is a thousand and two,  
the classical method says Z equals the square root view.  
Take root of n over k, cap it at one,  
and partial credibility's formula is done.  

The Bühlmann approach has a different Z flavor,  
n over n plus k is the number to savor.  
And k is the EVPV divided by VHM —  
the ratio of process noise to signal, you see.  

So learn your Z formulas, classical and B,  
and never forget what one-minus-Z means to thee:  
it's not a waste and it isn't ignored,  
it's M times that weight — the complement aboard.


### Blend and Believe

Class C had a hundred claims, good year through and through,  
the loss ratio came in at sixty-two!  
"Cut rates by five percent!" cried the manager with glee.  
The actuary said "Slowly — let us compute Z."  

Root of one hundred over a thousand and eighty-two,  
Z equals point three — not much to construe.  
So thirty percent of the own data's tale,  
seventy percent from the statewide pool's trail.  

The statewide said eight percent increase is due,  
so blend point-three times minus-five with the rest of the crew.  
One-minus-Z times the statewide, plus Z times the own,  
the credibility estimate is four percent grown.  

The manager frowned: "But our data says down!"  
"Your data's too thin," said the actuary with a gown.  
"A hundred claims cannot the future foretell,  
we blend with the stable — and all will be well."
