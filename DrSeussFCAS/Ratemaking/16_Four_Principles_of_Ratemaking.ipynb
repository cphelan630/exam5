Learning Task A16: Apply the four principles of ratemaking to a scenario.


### The Four Pillars

**Principle 1 — Prospective**

A rate is not a backward glance,  
Not punishment for old mischance.  
It prices futures yet unseen,  
Expected costs of what is to be  

**Principle 2 — All costs**  

Premium must carry every load:  
The claims, the LAE on the road,  
The expenses, too, and profit’s share—  
A sound rate leaves no cost out there.   

**Principle 3 — Individual risk**

Not every risk should pay the same,  
Though pooled they are within the frame.  
If one risk’s data is thin and slight,  
Use similar risks to price it right.   

**Principle 4 — Actuarially sound**

Reasonable rates are built this way:  
Not too high, not too low to pay,  
Not unfairly split from risk’s true story—  
Sound cost estimates earn that glory.   

**Summary poem — all four together**

Look forward first, not back in fear.  
Include all costs that will appear.  
Match risk to risk as best you can.  
Be sound, not flawed, in pricing plan.   

**Ultra-short memory version**

Prospective.  
All costs.  
Individual risk. 
Actuarially sound.   

Key takeaway: **The four principles are future costs, full costs, individual-risk equity, and actuarial soundness.**


### Principle Four: Actuarially Sound

A rate should reflect what the future will cost,  
not padded too high, not so low you are lost.  
Not excessive, inadequate, nor unfairly askew,  
if it’s actuarially sound, it can carry you through.  

If one risk costs more, then more rate may be due,  
but that idea belongs first to Principle Two—  
and Principle Three says each individual guy  
should pay for its own expected slice of the pie.    

If you charge all alike when their costs are not same,  
the safer risks leave and the worse stay in frame.  
Adverse selection comes in, and the average won’t last;  
your book gets more expensive, and trouble comes fast.  
  
So classify with care, and support what you do;  
let expected cost differences guide what is true.    
Territory, age, use, prior claims may belong—  
but only if data says they predict cost strong.   

And fairness here does not mean every rate’s the same;  
it means cost-based distinctions, not arbitrary blame.  
If the cost difference isn’t there, the indication must stop;  
unfair discrimination means the support has to drop.  

So Principle Four is the final test:  
Are the rates sound overall, and do they hold up best?  
Reasonable, balanced, and grounded all through—  
that’s the standard ratemaking is aiming to do.
