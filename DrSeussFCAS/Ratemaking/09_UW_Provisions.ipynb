Learning Task A9: Calculate the underwriting provisions underlying the overall rate level indication (e.g. fixed and variable expenses, profit and contingency, transaction costs).


### Load It All In

The premium must cover every cost that we bear —  
The losses, the adjusting, the overhead to share,  
The commission paid out and the taxes we owe,  
The profit target and the reinsurance flow.

The variable expenses are a fraction of rate —  
Commission, taxes, policyholder rebate.  
They come out of premium as a percent, you see —  
So move them to the denominator: 1 minus V.

The fixed expenses — the cost per policy each —  
Are not a percent; they don't rise with premium's reach.  
Put them in the numerator, one fixed dollar amount,  
They go into the rate regardless of the count.

Add Q for the profit — the target that's set —  
1 minus V minus Q: the PLR net.  
Divide the numerator by this fraction below —  
And the indicated rate begins to show.

Load it all in, every penny, every cent —  
From reinsurance to overhead to the profit intent.  
The rate that emerges is adequate, full —  
When every expense has been loaded to pull.


### Every Penny Has a Place

Every penny of premium must cover a cost —  
No penny unassigned, no penny be lost.  
The commission goes here, the taxes go there,  
The overhead and profit must each get a share.

The variable expenses are taken as a percent —  
From each dollar of premium, that fraction is spent.  
They live in the denominator, subtracted below:  
One minus V minus Q — that's the PLR flow.

The fixed expenses don't change with the premium size —  
A policy's a policy, same cost, no surprise.  
They ride in the numerator, flat dollar and true —  
Added to losses before dividing by PLR through.

The reinsurance cost — on gross basis — is V,  
A percentage of premium you pay to be free  
Of the worst of the losses the treaty will share —  
On net basis, leave it out — it's already there.

The PLR tells you what losses may be —  
Sixty-seven cents per dollar? That's all that we're free  
To spend on the claims without losing our way —  
Exceed it, and rates must rise the next day.

Look back at the expense ratio — that's the past,  
But load what the FUTURE will ask you to cast.  
Remove the one-timers, adjust for what's new —  
Every penny has a place when the math is true.


### Expense Methods 

Three roads to carry expenses through,  
When pricing the book and building the view.  
One says, “Treat them all as premium-tied,”  
And let every expense with rate change ride.  
 
One says, “Some are fixed, some vary with pay,  
Project premium first, then split the way.”  
A middle road, common and clean,  
Using premium dollars to shape the scene.  

One says, “No, fixed costs live per risk or file,  
Project them by exposure, policy, style.”  
Best when the count of risks tells more  
Than premium ever really bore.  

So one makes all variable, simple and fast.  
One separates but lets premium forecast last.  
One separates and anchors fixed cost to count.  
The right choice depends on what costs really amount.  

**All-variable expense method**  
 
Make all expenses dance with premium’s song,  
As if when premium moves, they all move along.  
Commissions, taxes, general expense,  
All treated variable for convenience.  

The mechanics are easy, the math is light:  
Put everything as a ratio, keep it tight.  
Losses and LAE take their share,  
Expense ratio joins them neatly there.  
Then profit is loaded on top with grace,  
And the indication falls into place.  
 
Its advantage is speed, its advantage is ease,  
A clean little method that aims to please.  
When data are thin or time is small,  
This method is handy, perhaps best of all.  

But the weakness is clear if you stop and stare:  
Not all expenses truly belong there.  
Some costs do not rise just because premium grew,  
So calling them variable is not really true.  
If rates go up, the method may say  
Expenses rise too in the very same way.  
That can distort what the indication should be,  
Because fixed costs are not rate-level debris.  
 
Its assumption is blunt, and that is the game:  
That expense behavior follows premium the same.  
Useful for rough work, but weak if you know  
The fixed/variable split really matters below.  

**Premium-based projection method**  

Now enters the method more balanced in tone:   
Not all expenses are premium-grown.  
Some are variable, yes, and some are fixed,  
So the expense structure is properly mixed.  

The mechanics go something like this:  
Project premium first to the future abyss.  
Variable costs stay as a percent of premium earned,  
While fixed expense dollars must separately be learned.  
But fixed costs are projected with premium in mind,  
Using historical premium as the scaling bind.  
So fixed expense ratio in future disguise  
Often comes from fixed dollars over premium size.  

Its advantage is better realism than one,  
Because it admits fixed costs are not all spun  
From premium changes alone in the year;  
It makes that distinction at least somewhat clear.  
It is practical too, often easier to do  
When premium data are stronger than counts in view.  

But the disadvantage lies under the skin:  
Fixed costs do not always move with premium within.  
If average premium shifts but policy count does not,  
This method may miss what fixed expenses ought.  
A growing premium base can dilute fixed expense share  
Even when the actual workload is still there.  
So if fixed costs follow policies more than premium flow,  
This method can quietly misstate the show.  

Its assumption, then, is the critical thread:  
That premium is a decent proxy ahead  
For how fixed expenses will behave through time,  
Or at least good enough for the actuarial rhyme.  
 
**Exposure/policy-based projection method**  

The third road says, “Count what drives the work.  
Do not let premium alone do the clerk.”  
Fixed expenses often arise per policy, file,  
Per bill, per renewal, per risk all the while.   

So the mechanics start with a split made clean:  
Variable expenses stay premium-based in the scene.  
But fixed expenses are projected per exposure or policy unit,   
Then multiplied by future counts to run it.  
If policy count grows, fixed expense grows too;  
If count stays flat, it may largely stay true.  
Then divide through by projected premium or rate base  
To place the fixed load in the indication case.  

Its advantage is conceptual strength:  
It often matches operations at length.   
When fixed costs are driven by how many risks you touch,  
This method captures that logic much better, as such.  
For lines where exposure units matter a lot,  
Or where average premium changes but workload does not,  
This method often gives the sounder read  
Of the real fixed cost that rates must feed.  

But the disadvantages are not hard to find:  
It needs better data and a sharper mind.  
You need future exposure or policy projections too,   
And those can be harder to estimate through.   
Also some “fixed” costs are not purely per count,  
So even this method may not fully account.  
Reality is messy; expense behavior can blend,  
Neither wholly fixed nor variable end to end.  

Its assumption is stronger, but still not exact:  
That exposure or policy count best tracks  
The future level of those fixed expense streams,   
Better than premium does in the ratemaking schemes.  
When that assumption is right, this method shines.  
When not, it too can blur the lines.  


### Comparison of Expense Methods 

Three ways to load the expense side,  
Each asks what truly moves with price.  
One keeps it simple, broad, and fast;  
One splits the stream; one looks past  
Premium toward the count of risks—  
Three methods, three actuarial twists.  

**All-variable** is blunt and clean:  
“Let premium drive the whole machine.”  
If premium rises, expenses too;  
A tidy shortcut, easy to do.  
Its strength is speed, its math is light;  
Its weakness: fixed costs leave the sight.  
For not all dollars ebb and flow  
Just because premiums rise or slow.  

**Premium-based projection** stands midway:  
“Some costs are fixed, some vary with pay.”  
Variable loads stay premium-tied,  
But fixed expense is set aside.  
Still, premium helps project the whole,  
So premium remains the guiding role.  
More realistic than method one,  
Yet still can miss what fixed costs run  
When workload stays but premiums shift—  
Then premium gives a partial lift.  

**Exposure- or policy-based** goes deep:  
It asks what drivers fixed costs keep.  
Per policy, file, renewal, touch—  
Some costs depend on count, not much  
On premium size or average rate;  
This method tries to match that fate.  
Its strength is logic, grounded, sound;  
Its weakness: data must be found.  
More inputs, more forecasts, more room to miss,  
But conceptually it cuts to this:  
If fixed costs follow units, not price,  
Then count-based projection is more precise.  

So all three methods seek one end:  
Project the costs that rates must tend.  
One says **price drives all**. One says **price drives some**.  
One says **count of risks tells where fixed costs come**.  
The closer your method fits how expenses behave,  
The more believable the rates you pave.
