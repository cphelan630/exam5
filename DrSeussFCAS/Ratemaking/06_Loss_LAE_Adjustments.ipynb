Learning Task A6: Calculate loss and loss adjustment expenses to be used for ratemaking (e.g. common ratios, adjustments to losses, claims-made vs. occurrence coverages).


### Prepare the Losses

The losses are coming — they're raw and they're rough,  
They're reported and partial and not nearly enough.  
To use them in ratemaking you must prepare them first:  
Develop, adjust, trend — quench the ratemaker's thirst.

Step one: Take reported and multiply through  
By the CDF — to the ultimate view.  
The triangle's columns reveal what will be —  
At infinite age the ultimate is free.

Step two: Adjust the benefits — if limits have changed,  
Or WC indemnity has been rearranged —  
Apply the adjustment to every prior year,  
Restate them to current, make the comparison clear.

Step three: Add the ALAE — the allocated expense —  
Develop it separately or combined, and hence  
Add in the ULAE at the end as a load,  
As percent of losses — that's the standard code.

Step four is the trending — but ONLY AFTER  
You've developed to ultimate. Don't trend before —  
The LDFs capture the TIME-based detail,  
The trend captures INFLATION — don't mix them or fail.


### Adjust for Benefit Changes

The benefit changed in the twenty-second year,  
The cap on indemnity was raised, loud and clear.  
But our historical data still sits at the old —  
So we can't use it raw; it must first be retold.

Apply the adjustment to EVERY prior year —  
Not just the year before the change appeared.  
All years must be stated at the current level's bar,  
Or your trends will be tilted by how benefits are.

The BCA factor comes from the wage distribution,  
From how many claimants the cap caused restriction.  
Divide expected new by expected old —  
The factor that tells you the benefit story told.

Once restated, your data speaks one language clear,  
Your trends are pure signal — no benefit noise here.  
Develop and trend with confidence now,  
For your historical losses all live at today's vow.


### Adjust for WC Benefit/Coverage Level Change  

In workers comp the statute speaks,  
A benefit shifts, the old law creaks.  
The losses shown are from days before,  
But rates must price the future door.  

Do not be fooled by history’s face,  
Past dollars lived in a different place.  
If med or indemnity rose by law,  
The triangle hides a silent flaw.  

The exam will likely set the scene:  
“On Jan first, benefits changed between  
The experience years and forecast span;  
Adjust the losses.” That is the plan.  

Perhaps TTD now lasts more weeks,  
Or max weekly pay climbs to new peaks.  
Perhaps a coverage broadened wide,  
Or offsets shrank the carrier’s side.  

Then comes the task, both plain and sly:  
Find which claims the change would modify.  
Not every loss gets hit the same;  
You must isolate the altered frame.  

For medical changes, think with care:  
Which services now fall in there?  
For indemnity, ask who is paid,  
How much, how long, what caps were made.  

The direct effect becomes your sword:  
Old benefit level to new restored.  
A factor built from old and new,  
Applied to losses that pass through.  

If all claims are changed in equal way,  
A simple factor may save the day.  
But more often segments lead the dance:  
By injury type or severity bands.  

Then multiply, restate, adjust,  
Make old experience future-trust.  
Because ratemaking is not to bless  
What losses were, but what they’ll express.  

So when Exam 5 lays down the test,  
The winning move is not a guess:  
Identify the law change, scope the claims,   
Calculate the factor, and rename  
Those historical losses to what they’d be  
Under today’s reality.  

### Downward Development

Sometimes losses shrink as the claim matures,
not every triangle only climbs and lures.
Case takedown, deductibles recovered, salvage too —
these pull the incurred back to something more true.

**Case Takedown**

When case reserves are set a little high,
and later the adjuster says goodbye —
the case closes lower than it sat before,
and incurred losses dip below the prior floor.

It shows as negative age-to-age in the frame,
a downward factor — development in reverse name.
Select factors that account for this decline,
or incurred ultimates will sit above the line.

**Deductible Recovery**

Large deductible policies shift the pay:
the insurer advances the claim on the day,
then bills the policyholder back for their share —
that recovery offsets what was sitting there.

If data is gross before recovery comes in,
adjust to net, or the ratemaking will spin.
Match the basis of losses to the premium in force,
and track the billing lag — it can alter your course.

**Salvage or Subrogation**

After paying the claim the insurer may find
the damaged property or a liable party behind.
Salvage is selling the wreck that remains,
subrogation pursues a third party's pains.

These recoveries reduce the net cost of the claim,
so ultimate losses should reflect that net frame.
Develop or ratio the recoveries out,
then subtract from gross — that is what net is about.

# Claims Made

**Why claims-made exists**

Long tails made the pricing blind,  
Bad trends took years to clearly find.  
So insurers changed the trigger date:  
Not when it happened, when told to rate.   

**Occurrence vs. claims-made**

Occurrence listens to when loss began,  
Claims-made asks, “When reached the plan?”  
Same risk in spirit, different clock—  
One by diagonals, one by row and block.  

**Report year aggregation**

Let (L(y,r)) mark report year and lag,  
Each row a claims-made pricing bag.  
Each diagonal is occurrence land—  
Report-year rows are the claims-made stand.   

**Retroactive date**

A claim must happen after retro day,  
And be reported while the policy’s in play.  
That retro line is crucial law—  
Without it, overlap or gaps you saw.  

**Immature to mature**

First-year claims-made is thin and small,  
It catches only part of all.  
Renewals stack more lagged claims in sight,  
Till mature coverage reaches full report-year height.  

**The five principles**

If costs rise, claims-made should cost less still,  
Its shorter forecast helps the rate fit bill.  
Trend shocks hurt occurrence more than made,  
Report-lag shocks too, less reserve to shade.  
No pure IBNR sits in its chest,  
Less float to invest—though pricing risk is less.  

**Switching coverage forms**

Occurrence to claims-made: set retro with care,  
Too early means overlap sitting there.  
Claims-made to occurrence leaves a hole behind,  
So tail coverage guards what late reports might find.  

**Retirement**

When the doctor stops, the risk does not,  
Old acts may surface long after thought.  
The final claims-made year is not enough alone—  
A tail may be needed for claims still unknown.   

**Maturity, switching, and diagrams**

Each claims-made year grows wider scope,  
More accident time is brought in hope.  
Step factors climb year after year,  
Till mature claims-made is finally here.  

Switching to CM can work just right  
If retro meets prior expiry night.  
Switching from CM leaves gaps to see,  
So tail coverage is the remedy.  

In report-year pictures, keep this law:  
Claims-made takes rows that reporters saw;  
Occurrence slices diagonal flow—  
That’s the key exam fact to know.  


**The Five Claims-Made Principles**

When costs are climbing year by year,  
occurrence rates must price more fear;  
claims-made looks shorter down the road,  
so it should bear the lesser load.  

If trend breaks hard and runs off course,  
both may miss the hidden force;  
but claims-made, built on nearer ground,  
will land closer when truth is found.  

If fewer new claims get reported in the year,  
mature claims-made will barely feel the veer;  
its rows are already locked with prior acts inside,  
so reporting lag shifts leave its pricing as guide.  

Occurrence, on the other hand, cannot hide:  
delayed claims report with higher costs applied.  
So rates for occurrence end up with defects  
when reporting lag changes leave uncounted effects.  

No pure IBNR waits in shade,  
that ghost is not on claims-made's page;  
so reserve risk shrinks, and far less dread  
lives in the unpaid claims ahead.  

But every gain has got its due:  
less float means less investment, too.  
Claims-made pays sooner, cash won't stay,  
so income earned gives up some play.  

So sing it once, and know the theme:  
less price than occ when costs all steam,  
closer on trend shocks, calmer on lags,  
less IBNR in reserve bags,  
but less investment income made —  
that is the logic of claims-made.  

**Short chant version**

Cheaper than occ when costs increase.  
Closer if trend forecasts lose peace.  
Reporting lag shifts hurt mature CM less.  
No pure IBNR to guess.  
Less investment income, nonetheless.

### Caveat to Reporting Lag Principle: Mature vs Immature 

If claims-made still is young and thin,  
New layers of old acts enter in;  
A lag shift then can change the grade,  
More than for fully mature claims-made.  

Once mature claims-made has filled its span,  
Its response grows steadier than it began;  
Reporting changes still may bite,  
But less than occurrence, and less than in youth’s first write.
