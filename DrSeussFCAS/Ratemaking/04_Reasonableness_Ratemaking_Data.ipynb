### The Data Check

Before you compute, before you project,  
Before you develop or trend or correct —  
You must look at your data, you must check it first,  
For the cleanest of methods can give you the worst.

If the data is broken, if the codes are amiss,  
If a system migration produced a wrong list,  
Your LDF's perfect, your trend is just right,  
But your rates will be wrong when you turn on the light.

So check year over year — does the frequency jump?  
Does severity fall like a boulder off aump?  
Is the claim count suspiciously low in year three?  
Is the premium per exposure wildly off-key?

Compare to the benchmarks, compare to the trade,  
Ask why the unusual pattern was made.  
Document what you found, and document why —  
The ASOP says: disclose, don't let anomalies lie.

The actuary who skips the data review  
Is building their castle on fog and on dew.  
Check the data. Check it twice. Check it again.  
Then proceed with your models — and do it with zen.


### Trust But Verify

The data arrives in a neat little file,  
With columns and numbers arranged with some style.  
But trust it? Oh no — you must verify too,  
For the data may carry a flaw or two.  

ASOP Twelve says: your classes must stand   
On four solid legs — they must all be planned:  
Actuarially sound, and practical, too,  
Legal in all states, and fair through and through.  

Correlated with losses? A good start, it's true,  
But that's just the first of four gates you pass through.  
Can you measure it? Prove it? Is it allowed by law?  
Does it treat all the people without any flaw?  

ASOP Thirteen insists: when you trend, be fair,  
Pick your period with reason — not strategic flair.  
Don't choose the two years that give you the low,   
When seven years' data says trends ought to grow.  

Internal, external — consider them both,    
The CPI and your own data: each sworn on their oath.   
Document why you chose what you chose,   
And disclose every assumption your judgment bestows.  

ASOP Forty-Three, the last of the three,  
Says any unpaid estimate — mine, yours, and free —   
Must be reasonable, documented, disclosed —  
Whether ratemaking or reserving, the standard's not closed.  

So trust but verify. Check the data. Document.  
Disclose the assumptions. Show where your judgment went.    
The actuary who follows these standards with care  
Will produce work that's honest, defensible, fair.
