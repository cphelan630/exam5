### Trend Forward

The past is the past and the data is old —    
The losses of yesterday can't be retold.  
The future will cost more (or less, but let's say more) —  
So trend your historical losses to the future's door.  

Medical inflation drives severity up,  
Attorney's fees fill the settlement cup.  
Legal environment shifts, juries award more —  
The losses of tomorrow outpace the ones before.  

Trend from the midpoint of history's view  
To the midpoint of rates that are coming up new.  
Not from start to end, but midpoint to mid —  
That's where the average accident sits and did.  

Multiply ultimate losses by the trend factor tall,  
But only AFTER developing — don't trend before the call.  
Development and trend are two separate things —  
Conflating the two is the error that stings.  

Trend forward with confidence, trend with your best rate,  
And your historical data will speak to the future state.  
Without the adjustment the rates will be wrong —  
Trend is the bridge between old and tomorrow's song.  


### Calculating Trend Factor  

Four roads wind through the trendland hills,  
Each with strengths and each with bills.  
The exam may ask which path to take,  
So learn the song each method makes.  

The first is **percent change**, blunt and clean,  
A simple glance at what has been seen.  
Take one year’s level, compare the next,    
Read the growth straight from the text. 
Good when the pattern is steady and slow.  
But one odd year can swing the view,  
And noise may whisper lies to you.  

Then comes the **linear fit**, calm and plain,  
Drawing a line through loss or claim.  
It says each year adds just so much more,  
A fixed amount, no less, no more.  
Good when dollars rise in equal bands,  
Like climbing steps with level spans.  
But if the world compounds instead,  
A straight-line choice may mislead your head.  

The **exponential fit** is trend’s true flame,  
Growing by percentage, year to year the same.  
Not plus ten dollars, but plus ten percent,  
A compounding path where costs are sent.  
This often suits inflation’s art,  
Where each year builds on the prior part.  
For many exam questions, this is king,  
When severities swell with an economic spring.  

But losses are harder, sly, and deep,  
Not every pattern is yours to keep.  
For loss trend has two extra doors,  
Beyond the simple fitted scores.  

One door is **GLMs or econometric might**,  
Where many forces come to light.  
Frequency here, severity there,   
Wages, laws, inflation, care.  
A richer model, more refined,  
To separate causes intertwined.  
Not just “losses rose,” but *why* they grew,  
And which variable pushed them through.  
Powerful stuff, but harder ground,  
More data, more judgment, more traps are found.  

The other door looks diagonal, sly:  
**Incremental calendar year data by accident year**.  
A mouthful, yes, but sharp in aim,  
To match old losses to the risks that came.  
It says: do not let plain CY deceive,  
When growing books make false patterns weave.   
Split the calendar year by AY source,  
Then pair each piece with its proper course.  
This guards against mismatch in time,  
And gives loss trend a cleaner rhyme.  

So what is the lesson, what should you know,  
When the exam says, “Which trend would you show?”  
**Percent change** is quick and rough.  
**Linear fit** adds equal stuff.  
**Exponential fit** compounds the climb.  
**GLMs/econometrics** explain through time.  
And **CY by AY** helps losses stay    
Matched to the business that made them that way.  

Choose not the fanciest just for show,  
Choose what the data truly know.  
For trend is not a ritual chant,  
But a judgment call the numbers grant.  

### Approaches (2) to Step-1 of 2-Step  

Step one can start in either way:  
Rebase old years to today’s observed lay,  
Using the latest level as anchor and guide,  
Then step two projects to the future from that side.  

Or roll each old year through time’s own gate,  
Trending each one to a common date.  
One anchors to latest level, one to latest time—  
Two paths to the same prospective rhyme.  

### The Two-Step Trend

The trend has two steps — don't forget them, my friend —  
From history's midpoint to the present, then bend  
From the present to future, where the rates will be set —  
Two legs of a journey you can never forget.

Step one: take the recent observed rate and apply —  
How quickly did losses inflate, measure high?  
From the midpoint of data to the date of today —  
Use the recent-trend rate to get partway.

Step two: take the expected trend going forward —  
From today to the future rate period's word.  
If inflation is calming, the rate may be low;  
If medical costs are still surging — still grow.

Multiply step one by step two — that's your trend —  
A two-factor adjustment from the start to the end.  
Then multiply ultimate losses — the trended result  
Is losses in future-period dollars, exult!

But remember the midpoints — not start, not the end —  
The AVERAGE accident date is where both legs bend.  
Experience midpoint to future rate midpoint true —  
Between midpoints is where all the trending runs through.

And after development, trend — in that order alone —  
The overlap fallacy leaves us lost, all alone.  
Develop then trend, always trend after the last —  
Ultimate, then trended: from future and past.


### Trend Dates  

Trend the losses, trend them right,  
But first decide the target date in sight.  
For dates are where the factors live,  
And bad date choices wrong answers give.  

A year of data is not one day’s claim,  
So midpoint thinking is the usual game.  
For accident year, the middle is king,  
Halfway through is the anchoring thing.  

For policy year, pause and think twice,  
The average accident date is not so nice.  
Policies write through all of the year,  
And losses occur as coverage draws near.  
So often you land not at year start plain,  
But further out on the timing chain.  

Trend from the average date of loss,  
Not from the valuation point by force.  
Development gets you ultimate first,  
Then trend moves cost levels, quenches the thirst.  

And where do you trend to? Not just “today,”  
But the average date losses will happen that way.  
If annual policies start on Jan one,  
The future average loss date is not at begun.  
It sits in the middle of the policy flow,  
Often a half-year where losses will grow.  

For premium, the date may differ instead,  
Average written or earned dates must be read.  
Premium trend follows premium’s own stream,  
Not always the same as the loss-date scheme.  

Beware of double counting in haste,  
A common exam and workplace waste.  
On-level fixes rate changes gone by,  
Trend handles time; do not let roles tie.  

And if a one-time change altered the ground,  
Adjust that separately, do not let trend drown.  
Trend is for movement through passing years,  
Not law changes, benefit jumps, or quirks in arrears.  

So ask each time before you compute:  
What is my from-date’s actual root?  
What is my to-date in future design?  
And does that match the item’s timeline?  

For trend is easy once dates are true;  
Most of the battle is choosing the two.
