Learning Task A3: Aggregate and/or organize ratemaking data in the following ways: calendar year, policy year, accident year, report year, close year, in-force, net of reinsurance.

### Get Your Data Right

Oh, the data! The data! You must get it right,  
Or your rates will be broken by morning and night.  
You must match up your losses with premiums too —  
The SAME kinds of risks, the same time period through.

There are three ways to sort them, yes three ways in all,  
The Calendar, Accident, and Policy call.  
The Calendar closes on December the last —  
It sees what you booked, but the year has been passed.

The Accident Year asks: when did it occur?  
The loss and the premium are joined in a blur.  
It develops and grows through the columns of time,  
From one year to ten in a triangular rhyme.

The Policy Year is the most careful of all —  
Each loss goes back home to the policy's call.  
It takes even longer, two full years or more,  
To know what the policy was truly in store.

So don't mix your calendar with accident days,  
Or your loss ratio walks in a fog and a haze.  
Get the numerator matched to the denominator base —  
Get your data right and you'll win the ratemaking race!


### In-Force, Close Year, Net and Gross

In-force is the business still living today,  
The risks on the books in a current display;  
It helps when you ask what is active right now,  
Not only what history happened to allow.  

Close year looks backward to when claims shut tight,  
A closure-date grouping, not accident's light;  
Useful for some claim studies, but mind what you pair,  
For close-year with accident costs is a mismatch to bear.  

Gross is before any treaty has cut,  
The full claim amount before reinsurance shuts;  
Net is what stays after ceded is gone,  
So match net losses with net premium all along.  
  
Do not mix net with gross and declare it okay,  
Or your ratios will wander and drift far away;  
Choose one clean basis and stay with it through,  
And your data organization will stay sound and true.
