Learning Task A13: Select and justify a final rate change to implement beyond the calculated overall rate level indication (e.g. operational, marketing, regulatory constraints, lifetime value).


### Indicated Is Not Final

The indication came in at plus thirty-eight,  
and the actuary delivered the news to her fate.  
The management blinked, then they coughed, then they sighed,  
"Thirty-eight percent? We'll lose half our inside!"  

"The regulator will laugh and the agents will run,  
thirty-eight percent cannot be the one.  
We need to phase this in over two or three years,  
smooth out the edges and quiet the fears."  

The actuary nodded — she understood well,  
the indication's a guide but not always a bell.  
She documented the gap and the reasons with care:  
competitive constraints and retention impair.  

But she noted: "This rate will remain inadequate still,  
until we have closed the full gap with goodwill.  
Year two we'll file more, and year three we'll complete,  
and the cumulative change will be actuarially neat."  

Indicated is not final — that much is true,  
but the deviation must always be documented too.  
State your reason, the direction, the gap to be filled,  
and the indication's integrity will be fulfilled.


### Indication and Judgment

The indication says plus-thirty, but the CEO says no.  
"We'll lose all our customers — I'm telling you, slow.  
We can handle fifteen this year, ten the next,  
and a few in year three — that's the management text."  

The actuary listened and opened her file,  
"The indication is thirty — I'll document the while.  
Year one is fifteen — that's oh-point-eight-seven of need,  
and the remaining is thirteen — just compound and heed.  

Not fifteen minus fifteen, for rates are compound math,  
one-thirty divided by one-fifteen on the path.  
The remainder is thirteen, not fifteen, you see,  
multiplicative factors are the key.  

Year two we file ten, and the cumulative stands,  
at twenty-six-five — and the plan nearly lands.  
Year three with a three-percent closes the gap,  
and the thirty-percent indication snaps back.  

So phase it with care, and compound with might,  
and document every deviation in light.  
The indication's the compass; the judgment's the wheel —  
both matter together to seal the good deal.
