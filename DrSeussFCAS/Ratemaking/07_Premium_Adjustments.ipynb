Learning Task A7: Calculate various adjustments to the premium used for ratemaking (e.g. on-leveling, premium audit, common ratios, claims-made vs. occurrence premium issues).


### On-Level Everything

The premium was fine when the rates were in place,  
But the rates changed in October — we must embrace  
The fact that old premium was charged at old rates,  
And current-rate dollars must flow through the gates.

If we skip the on-leveling, the ratio will lie —  
Old denominator makes the loss ratio fly.  
The actuary cries: "Rates must go up thirty percent!"  
When in truth, the rate change already happened and went.

So on-level the premium — bring it to today,  
With extensions or parallelograms, pick your way.  
Compute the OLF, multiply through,  
And your denominator finally speaks truth anew.

On-level everything — every year in the file —  
Before you compute the first decimal and smile.  
The premium adjusted is premium right,  
And your rate indication will land in the light.


### The Parallelogram

Draw me a unit square, flat on the page,  
Time going right and the term on the stage.  
The bottom is January, the right side December,  
And premium earned is the area — remember.

A rate change in April? Draw a line going up —  
A vertical cut that divides up the cup.  
The left of the cut is the old rate in force,  
The right of the cut is the new, better course.

The old slice is a triangle, pointy at top —  
Its area is half base times height — then you stop.  
The new slice is the rest, the trapezoid wide —  
Together they sum to one whole, side by side.

Weight each region by its area fraction,  
Multiply by its index — that's the action.  
Average the indices, weighted by size —  
Then divide by the current: the OLF prize.

OLF greater than one when rates went up the scale —  
Old premium was small; current rates prevail.  
Multiply historical EP by the factor you found —  
And your on-level premium is actuarially sound.


### Audit the Premium

The policy started with payroll guessed light,  
But workers worked harder by autumn's last night;  
When audit came calling, the premium grew,  
Because earned exposure was bigger than knew.  

If you rate from the estimate written at start,  
You miss the full charge that belongs in the chart;  
Premium audit true-up is part of the game,  
For final earned premium may not match the same.  

Endorsements and audits can change what was due,  
So use the adjusted premium all the way through;  
Historical dollars need current-rate care,  
But they also need final audited measure in there.  

On-level the premium and audit it too,  
Both steps matter greatly for what you must do;  
The denominator must match what was earned,  
Or the indicated change is a lesson unlearned.
