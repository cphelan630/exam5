Learning Task A15: Analyze results of predictive models (e.g. GLMs).


### The GLM Machine

In the land of one-way, the actuary would go,  
computing the pure premium, row by row.  
For age, then for territory, then vehicle type too,  
and the relativities came out — but they weren't quite true.  

For age and for territory were tangled inside,  
young drivers in cities, not spread far and wide.  
The age relativity soaked up the city's cost,  
and the one-way result was confused and then lost.  

Then came the GLM, glorious and right,  
it put all the variables in simultaneous flight.  
Hold age constant, and territory stays true,  
each beta comes out in the right residue.  

Exponentiate the beta and there is your rate,  
the relativity pure, fully unconfounded and great.  
The log link is key — it's multiplicative math,  
the structure of insurance is right on this path.  

So let the GLM do what one-way cannot:  
untangle the correlations that muddled the plot.  
The model is rigorous, the data is clean,  
and the GLM machine makes the relativities keen.


### The Model Speaks

with betas and link functions polished and sleek.  
The log link connected the linear to the mean,  
and the Poisson distribution kept everything clean.  

"Exponentiate me!" said the beta with pride,  
"I am not the relativity — I'm logged inside.  
E to the beta is the number you need,  
for that is the factor that the rating will heed."  

The offset stood quietly, offset from the rest,  
"I am the log of exposure — I've taken the test.  
Without me the model confounds the count,  
but with me it models the rate, not the mount."  

The AIC said: add parameters with care,  
I penalize each one — so only add what's fair.  
The lift curve then asked: do your deciles align?  
If predicted and actual agree, then the model is fine.  

And the actuary smiled at the output with glee,  
"The GLM tells me what one-way can't see:  
the pure effect of age, holding all else the same —  
unconfounded, unbounded, and actuarially sane."
