Learning Task A5: Separate data into groups that balance homogeneity and credibility and summarize the considerations for determining such groups.

### The Tug-of-War

On one side stands Homogeneity, tall,  
She wants every class split — the granular call.  
"Put sports cars with sports cars! Sedans with sedans!  
Divide them by driver and territory plans!"

On the other side, Credibility stands,  
With data and samples in both of his hands.  
"You've split them too thin — there are only three claims!  
Your Z is a zero — your estimate's frames!"

They pull and they pull on the long actuarial rope,  
And the actuary stands in the middle with hope.  
She blends the two forces with judgment and care:  
"Z times the class data plus complement — fair!"

Not too thin, not too broad — find the balance, the sweet  
Spot where your classes are both homogeneous AND meet  
The minimum claims threshold, the k that you need,  
And then blend with a complement — that's how you succeed.

Homogeneity, Credibility — neither wins alone.  
The answer is both of them sharing the throne.  
Use Z and the complement to balance the two —  
The tug-of-war ends when you blend old with new.


### Group Them Right

You want to split policyholders into classes, you say —  
And charge different rates by the risk they display.  
But before you create a new variable or two,  
There are four actuarial questions you MUST make it through.

First: Is it justified? Does the data confirm  
That this factor relates to the losses — the firm  
Statistical link, the correlation clear?  
If yes, then Gate One you may now pass, my dear.

But correlated alone is not nearly enough —  
Gate Two asks: is it PRACTICAL? Is it not tough  
To measure, to verify, to put on a form?  
If the data can't be checked, it can't be the norm.

Gate Three is the law — and the law has its say.  
Some states say "Yes, use it." Some states say "No way."  
Actuarial soundness won't override the rule —  
A legally banned variable is not in your tool.

Gate Four is the hardest: Is it FAIR through and through?  
Does it proxy for race or religion or crew?  
A proxy for protected class, however disguised,  
Is unfairly discriminatory — and should be revised.

Four gates. All four must be opened with care.  
Miss any one gate and your variable's bare.  
Group them right, group them fair, group them with reason and art —  
ASOP Twelve is the guide for the classifier's heart.
