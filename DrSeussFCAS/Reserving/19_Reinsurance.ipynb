Learning Task B19: Define and apply reinsurance concepts to calculate net, ceded, and gross losses.

Gross is the loss before the shield,  
the full raw wound the claims have sealed.  
Ceded is the part passed on,  
to reinsurers when pain grows strong.  

Net is what the company keeps,  
the portion left when ceded leaves.  
So three must live in every view:  
gross, ceded, net — all tied and true.  

Start with gross, the total claim,  
before recoveries stake their claim.  
Then read the treaty, layer by layer,  
to see what burden others share.  
   
A quota share will split the cost,  
by fixed percent of every loss.  
If forty percent is ceded away,  
then reinsurer pays that same share each day.  

So gross times forty gives ceded due,  
and net is what remains with you.  
Simple, clean, a proportional song:  
same slice of premium, same slice of wrong.  

Excess of loss now works by band,  
not by a fixed proportional hand.  
The company keeps up to retention,  
then ceded picks up next dimension.  

“100 xs 50” means this instead:  
the first 50 stays with the cedent’s bed.  
The next 100 may ceded be,  
and above that top comes back to thee.  

If gross is 120, then net is 50,  
ceded is 70 — plain and nifty.  
If gross is 30, no reinsurer pays,  
the loss stays net in full always.  

If gross is 200, ceded takes 100,  
net keeps 100 too when all is done.  
Because 50 below and 50 above  
sit with the insurer, not the treaty thereof.  

Aggregate covers may also appear,   
where many losses combine through the year.  
Per risk, per occurrence, catastrophe too—  
the treaty wording decides what to do.  

So define the terms and apply with care:  
what is the base, and what layer is where?  
Because one wrong step in gross-to-net flow  
can wreck the reserve numbers you show.  

Gross is the whole.  
Ceded is shared.  
Net is retained  
once treaty has cleared.
