Learning Task B18:  Communicate results and drivers of change to various stakeholders (internal management, investors, regulators).

Three audiences wait beyond the page,  
each reading numbers in a different way.  
Management asks, “What changed this year?  
What drives the move, and what should we fear?”  

Investors look for profit’s shape,  
for strength that isn’t thin as tape.  
Regulators watch with sterner eyes,  
for signs masked over by reserve disguise.  

So do not merely state the pick,   
the carried number, clean and quick.  
You must explain what made it move—  
the why beneath the point you prove.  

Was frequency higher? Severity steep?  
Did case reserves strengthen? Did closure slow creep?  
Did large losses fade, or reopenings rise?  
Did assumptions shift with fresher eyes?  

Friedland warns that bad reserve speech,  
or numbers whose meaning you fail to teach,   
can bend decisions far off course—  
in pricing, capital, growth, and force.  
Management may chase a broken plan,  
expand too fast, or wrongly ban  
a line whose weakness was not real,  
or miss the wound they should have healed.   

Investors, too, may read the wrong tale,  
if reserves are weak or prudence frail.  
Too low, and strength looks overstated bright;  
too high, and value dims from sight.   
The balance sheet can speak untrue  
if reserve communication is careless through and through.     

And regulators need more than trust—  
they need reserve stories clear and just.  
Too little booked, too softly said,   
and trouble may spread before they tread.  
For if the warning comes too late,  
the company may not escape its fate.    

ASOP says: first know the use—  
who is the user? What’s the juice?  
A point may fit, a range may do,  
depending on what the work is for and who.  
Present the estimate in fitting form,  
not every crowd needs the selfsame storm.  
  
Disclose the purpose. Disclose the scope.  
Disclose what limits bound your hope.  
Give accounting, valuation, review date too,  
so users know what facts ran through.  
Name risks and uncertainties standing near;  
say what assumptions matter here.  
If something relied on lies outside your hand,  
say that plainly so all understand.  

And when this year updates the last,  
don’t hide behind “it changed” too fast.  
State which assumptions, methods, models turned,  
and why those changes now were earned.  
The driver of change is part of the art:  
not just the answer, but how it departs.   

So write for humans, not for sheets.  
Tie movement back to claim-world beats.   
One audience wants action, one value, one control—  
but each still needs the honest whole.   

Reserve work isn’t done at “selected” or “booked”;  
it ends when the right people can see what should be looked at.  
For the number alone is never the end—  
the real skill is explaining the forces that bend.
