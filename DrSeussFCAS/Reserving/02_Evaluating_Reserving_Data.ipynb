Learning Task B2: Evaluate reserving data and analyses for errors and reasonableness. 

### Question the Data

Oh, the data! The data! It all looks so clean!  
But things are not always exactly what they seem.  
That triangle cell that says eight hundred and one?  
Could be a mistake — could include the wrong run.  

Did someone reclassify claims in year three?  
Did the system migration confuse what we see?  
Did the adjuster change how they set reserves last fall,  
Making the incurred triangle jump up the wall?  

Question the large loss that showed up in year two,  
Question the diagonal spike — is that really true?  
Check all your paid-to-incurred ratios with care,  
And look at the benchmarks to see what's in there.  

For a method applied to bad data is worse  
Than no method at all — it's a quantitative curse.  
So before you compute a single CDF,  
Ask hard questions first — that's no bluff.  

The actuary who questions the data with skill  
Will select better LDFs and get a better fill.  
The one who just runs the numbers and calls it done  
Will have much explaining when the adverse runs come.


### Inspect Before You Project

Before you apply your LDF to the row,  
Ask yourself first: does this triangle really flow?  
Does each ATA sit peacefully in line,  
Or does one jump out like a big neon sign?  

That factor of two-point-one sitting alone,  
In a sea of one-seven-fives — it can't be home.  
It says something happened in calendar year four,  
A settlement push? Or a data error? Explore!  

Did the claims team settle a case that was late?  
Did a coding mistake sneak through the gate?  
Did a large single loss distort the whole row?  
Ask all these questions before you let it go.  

For the actuary who selects without a look  
Will find their reserves are way off the book.  
But the one who inspects with a critical eye  
Will select better LDFs — and here's why:  

The data is history, imperfect and real,  
Full of glitches and ghosts and occasional zeal.  
Your job is to filter what's signal from noise,  
And project the future with professional poise.


### Poem: "ASOP 43 Says: Look Before You Leap"

Before the first projection is placed on a page,
The actuary must set the reliability stage;
ASOP 43 says: *evaluate with care*
The data you're given before going there.

Is the data complete? Are there coding breaks?
Did the company change how it enters and takes
Its claim information — are the systems in line
With the triangles you're building across every spine?

Check internal consistency — do counts match the cost?
Does ALAE move with paid, or has something been lost?
Are the year-over-year ratios stable and clean,
Or are there spikes and reversals that want to be seen?

Evaluate the *source* of the data you trust:
Is it from a system migration, adjusted, or mussed?
Were there accounting changes, reclassifications done?
If so, the triangle may need to be run

On a restated basis — or at least with a note
That clearly discloses the limitations to quote;
**Document your reliance** on what you were given,
And **disclose assumptions** by which you were driven.

ASOP 43 blesses the actuary who says:
"The data has limits — here's what that weighs."
A reasonable range, well-explained and transparent,
Is more defensible than false precision apparent.
