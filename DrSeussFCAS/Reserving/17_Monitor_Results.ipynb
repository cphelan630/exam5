Learning Task B17: Monitor results for adequacy and reasonability including interim valuations (e.g., actual versus expected, roll forward analysis).

### AvE  

At quarter-end the question stands:  
Did claims emerge as planned?  
Not just, “What reserve did last year show?”  
But, “Did reality now follow?”  
 
Take selected ultimate from before,  
take latest reported at the door,  
then use the pattern you had picked  
to find expected emergence quick.  

Expected reported for the slice  
is unreported times the rise  
in cumulative percent reported through  
the two dates you are moving to.  

And paid works too the selfsame way:  
take unpaid left at latest day,   
then multiply by the extra part  
of percent paid in that new chart.  

Compare expected to the real:  
did actual paid or reported exceed?  
If actual tops expected by a lot,  
ask what change the triangles caught.  

A big new loss may spike report,  
yet not mean ultimate should distort.  
A stronger case reserve can do  
the same for reported, not paid too.  

If paid is close but reported high,  
case adequacy may explain why.  
If both are high and keep that trend,  
maybe the old ultimate must bend.  

If actual lags what you foresaw,  
don’t slash reserve by reflex law.  
Maybe reporting sped up before,  
or timing noise just hit the floor.  

Interpolate when dates sit between,  
for month-end spots not plainly seen.  
Use percent reported, not raw LDF,  
that trap costs points and time on test day.  

So B17 is not just math—  
it’s judging what sits underneath the path.  
The discrepancy is the clue,  
not proof by itself of what to do.  

### Roll Forward  

You did a full reserve review,  
but now a new close date is due.  
No time to rebuild every piece?  
Then roll it forward, gain a read.  
 
Start with the prior selected base,  
the unpaid estimate in place.  
Then let the calendar unwind:  
pay claims, report claims, change what’s mine.  

The old reserve should run off some,  
as actual emergence comes.  
What was expected should appear;  
that is the monitoring test right here.  

Smaller shops often use this way:  
between full studies, quarter to quarter day.  
A roll-forward is the bridge you build  
from detailed review to later field.  

But do not merely move the old  
and hope the new date fits the mold.  
Exposure changes matter too;  
current period shifts must pass through.  

For financials, planning, budgets, rates,  
this monitoring informs the stakes.  
It helps you see if picks still stand,  
or if the next deep dive is at hand.  

Check reasonableness beyond one test:  
loss ratios, severities do the rest,  
pure premium, frequencies, case implied,  
open-claim averages side by side.  

Compare methods, not just one.  
If estimates spread, ask what’s going on.  
Operations change, environment shifts,  
and reconciliation brings new insights.  

Roll-forward is not a full reserve,  
but it tells whether picks still hold their nerve.  
Actual versus expected is the drum:  
did the old estimate beat right as claims have come?  

So watch the runoff, month by month,  
by AY if you want the sharpest hunt.  
Reserve analysis does not end at pick—  
B17 says: monitor it.
