Learning Task B11: Adjust data and/or estimation techniques for changes in the internal and external environment (e.g., claims processes that result in shift in the adequacy of case outstanding or shift in settlement rates, change in mix of business, change in rate level, inflationary or legal environment).

### Case OS Shift 

When reported triangles start to swell,  
ask first what story they truly tell.  
Did claims get worse in honest size,  
or did case reserves simply rise?  

Friedland warns the method’s core:  
reported development works no more  
if case adequacy changed with time,  
beyond mere inflation’s steady climb.  

So test the claims department’s way,  
what changed in process day to day?  
Compare the average case per open claim  
to paid claim trends in the same game.  

If case grows faster than paid would show,  
reserve adequacy may be the hidden flow.  
Then Berquist-Sherman enters the scene:  
restating reported to a common mean.  

Pick a diagonal, often the last,    
trend average case outstanding backward fast.  
Multiply by open counts, add paid in place,  
and rebuild reported on a common base.  

Because if case strength rose in recent years,  
unadjusted reported will exaggerate fears.  
The raw method then will often overstate  
the ultimate claim and unpaid estimate.  

But mind the trap: the trend you choose  
can make the whole reserve win or lose.  
And shifts by size of loss can bend  
what looks like adequacy in the end.  

### Settlement Rate  

Not every change is case reserve strength;  
some lie in how claims close at length.  
If claims now settle fast and soon,  
paid patterns shift beneath the moon.  

A speedup in settlement changes paid,  
so old paid factors mislead the grade.  
The development method, left alone,  
may overstate ultimates it has shown.  
Reported may feel little sting,  
while paid triangles warp from closing speed.  

So watch disposal rates, closed counts too,  
and ask what the claims unit chose to do.  
Did staffing grow, did workflows change,  
did litigation handling rearrange?  

Then Berquist-Sherman can help again,  
not just for case strength now and then.  
Adjust paid for settlement pace,  
adjust reported with open-claim space.  
Use adjusted closed counts in the art,  
so open claims are right in part.  

The environment inside the firm  
can twist the triangles as they turn.  
So never trust a method blind   
when operations have redesigned.  

### Mix, Rate Level, Inflation, Law 

Some changes come from outside air,  
some from the book you chose to bear.  
A mix shift moves the claims you see—  
new business is not old business, see?  

If heavier risks now fill the page,  
development may change with the age.  
If classes, limits, or retentions sway,  
old patterns may not map today.  

If rate levels changed, or premium mix,  
expected-loss methods may need their fix.  
Bring losses and premiums to common ground;  
do not let old price levels confound.  

Inflation lifts severity’s song,  
legal changes can move it strong.  
Tort reform may narrow what courts allow;  
social inflation may press it now.  
The actuary must adjust the view   
for what the external world will do.   

Internal or external, the task is the same:  
find what changed and rename the game.    
Change the data, or change the method,  
but do not act like nothing shifted.  

For Exam 5 says the skill is this:  
not just triangles, not just a list—  
adjust for case, for pace, for mix,  
for rate, inflation, law, and shifts.
