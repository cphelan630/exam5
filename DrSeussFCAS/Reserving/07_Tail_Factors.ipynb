Learning Task B7: Apply a tail factor.

### The Tail Extends

The triangle ends at a hundred-and-twenty,  
But the claims? They keep coming — there's still quite aplenty.  
The WC disability that started back then?  
It's still paying out — it won't end until when?  

The med mal claim from the operation of five?  
It's still in the courts while the plaintiff's alive.  
The asbestos exposure from decades before?  
Still being reported — still knocking the door.  

So you cannot just stop at the last column's right,  
And say: "That's the ultimate — we're done for the night."  
There's development lurking beyond what you see,  
And the tail factor captures how much that will be.  

Too low a tail and your reserves are thin,  
Too high a tail and you over-put in.  
The art is in fitting the curve that extends,  
And extrapolating where the triangle ends.  

For the tail is not zero — don't make that mistake,  
The long-tail line's tail is more than a flake.  
Fit the inverse power, extrapolate true,  
And the tail you select will be honest and true.  

Don't forget the tail — it extends from the known,  
Into territory still waiting to be shown.  
The actuary who ignores it has done a disservice,  
To the truth of the ultimate, however nerve-wrenching.
