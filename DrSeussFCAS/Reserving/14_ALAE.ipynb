Learning Task B14: Calculate and evaluate the estimation techniques for allocated loss adjustment expenses

### What Is ALAE?

ALAE is the cost that sticks to the claim,  
the attorney, the expert — each one has a name.  
Unlike ULAE, which floats through the air,  
ALAE belongs to one claim with care.  

Defense counsel bills, court costs assigned,  
each dollar of ALAE to one claim is lined.  
So the question the actuary must answer and face:  
how much ALAE is still unpaid in this case?  

### Paid ALAE Development (Chain Ladder)

Build a triangle of paid ALAE by year,  
then find the age-to-age factors that steer.  
Select the LDFs, apply them in turn,  
project to ultimate — that is the concern.  

It works just like losses — same triangle game,  
the mechanics of development stay just the same.  
The challenge lies in data that's thin or unclear,  
since ALAE triangles can shift year to year.  

### Loss + ALAE Combined Development

Sometimes you fold the ALAE inside,  
combine it with losses and let them both ride.  
A single triangle of loss plus ALAE paid,  
develops together — one combined cascade.  
  
The LDFs now reflect both streams at once,  
and the ultimate gives loss plus ALAE, blunts  
the need to separate — but if ALAE trends  
differently than losses, combined it bends.  

### ALAE Ratio to Losses

The ratio method picks a simpler track:  
divide paid ALAE by paid losses, look back.  
Select a ratio from a range of years,  
then apply it to ultimate losses — it clears.  

ALAE ultimate = ratio times loss ultimate found.  
ALAE IBNR = ultimate minus what's been around.  
It is quick, it is easy, it relies on the past,  
but check that the ratio has held steady and fast.  

### Bornhuetter-Ferguson on ALAE

When paid ALAE data is sparse and young,  
the BF method can give it more lung.  
Start with an a priori ALAE ratio in mind,  
apply it to expected losses — that is the find.  
 
Weight the a priori by what is unreported still,  
add it to paid ALAE to give the fill.  
The older the year, the less prior weight it holds;   
the younger the year, the more a priori molds.  

### Comparison of ALAE Methods

Paid ALAE development works like any CL,  
but sparse data means factors may not hold well.  
Combined loss + ALAE keeps it together in one frame,    
but masks divergence if the trends aren't the same.  
  
The ratio method is the simplest tool here:  
pick a ratio, apply to ultimates, that is clear.  
BF brings in prior when the data runs thin,  
and credibility blends where you begin.  
 
Know which method fits the data you own,  
and whether ALAE trends separately alone.
