Learning Task B1: Organize reserving data in the following ways:  
* Calendar Year  
* Accident Year  
* Policy Year  
* UW Year  
* Report Year  

### Build the Triangle

In the land of the actuary, losses are born,  
They trickle and tumble from evening to morn.  
They come in at 12 months and inch up to 24,  
And nobody knows what will come through the door.  

So you lay out your data in rows and in lines,  
With the accident year here and the age over there,  
The first row says "year one" and points out it shines  
At a hundred and twelve thousand dollars to bear.  

At twelve months it's small, at twenty-four more,  
At thirty-six bigger, it grows like a plant.  
The triangle shows you what happened before,  
So you know what to do and you know what you can't.  

Each row is a year and each column an age,  
The diagonal is now — it's the most current page.  
Below it is zero, the future unknown,  
Above it is data, the past fully shown.  

So build your triangle, row after row,  
The pattern will tell you which way losses go.  
Without the triangle you'd only have guesses,  
But with it you'll pass all your Exam 5 stresses.


### Read the Triangle

Look at the triangle and what do you see?  
A grid full of numbers, all arranged carefully.  
Each row is a year when the accidents occurred,  
Each column an age — don't you worry, don't be deterred.  

The diagonal there — that bold line on the right —  
Is the latest known value for each row in sight.  
The upper triangle, that's history done,  
The lower triangle is where projection's begun.  

So read the top row first — it's the oldest and best,  
Sixty months in, it's already passed every test.  
The bottom row though? Just twelve months along,  
It barely knows what will become of its song.  

The ATA factor — divide row by row —  
It shows you how quickly the losses all grow.  
At twelve months to twenty-four, factor of one-seven-five,  
Means early-reported is barely alive.  

So build your averages, select what is true,  
Then multiply out and the ultimate's in view.  
Subtract what you have on the diagonal today,  
And the IBNR estimate will show you the way.


### Poem: "Four More Ways to Slice the Claims"

The accident year asks where the accident fell,
The calendar year tracks what moved on the books' bell,
But three more aggregations deserve their own verse,
For underwriting year and report year traverse.

The **policy year** gathers all claims from a policy's date —
Whatever occurred while that contract was great,
Whether accident year one or accident year two,
It follows the policy from start all the way through.

The **underwriting year** is cousin to policy year's call:
Policies written in year X bear all claims that fall
From those contracts — wherever and whenever they land;
Reinsurance loves UY — it's how treaties are planned.

It develops across two full calendar years or more,
The slowest to mature, but beloved by the treaty's lore;
When evaluating which vintage made money or not,
UY tells the reinsurer which year the loss was brought.

**Report year** says: I only care when the claim arrived —
The year it was first reported and registered and survived;
Claims-made lines use this logic by design:
If it wasn't reported this year, it isn't mine.

So CY, AY, PY, UY, RY — that is five,
Each answers a different question to thrive;
Match your aggregation to the question at hand,
And your reserving analysis will land where it's planned.
