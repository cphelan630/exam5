Learning Task B12: 12. Consider the impact of and adjust for the presence of large losses in a reserving analysis.

Large losses thunder through the page,  
and warp the look of every age.  
One claim can make a year look worse,  
and drag the factors off their course.  

So first ask: is this loss a trend,  
or just a shock that will not blend?  
If it is rare and outsized too,  
the triangle may not speak the truth.  

You might remove it, cap it low,  
or treat it separate from the flow.  
Project the bulk with ordinary care,  
then add the large loss back in there.  

Or segment layers, ground-up and excess,  
so ordinary claims cause less distress.  
For if one giant claim drives all,  
the selected factors may stand too tall.  

But do not smooth away real change,  
if large losses truly now are in range.  
The goal is not to make data neat,  
but keep the estimate fair and complete.  

So handle shock losses with judgment first,  
decide if the year is truly cursed.  
Then let the triangle show the norm,  
and treat the outlier in its proper form.
