Learning Task B15: Calculate and evaluate the estimation techniques for unallocated loss adjustment
expenses.

### Classical Dollar-Based  

Classical ULAE follows claim size,  
with paid ULAE and paid claims as guides.  
Pick a ratio from years gone by,  
then let unpaid claims tell what’s still high.  

Pure IBNR gets the full weight due,  
case on open claims gets half-way through.  
IBNER joins in with the unpaid stream,   
and ratio times basis gives the reserve scheme.  

### Kittel's Refinement  

Kittel says opening claims costs too,  
even when no payment yet came through.  
So ULAE is tied not just to paid,  
but to claims newly opened and work still made.  

Paid on old claims, opened-and-paid,  
opened-and-still-open all get weighed.  
It fixes the flaw in the classical view:  
some claim work happens before dollars do.  

### Conger-Nolibos Generalized Method   

Conger-Nolibos breaks the labor apart:  
opening, maintaining, closing each get a part.  
Reported claims start it, payments keep it alive,  
closed claims finish the work to derive.  

It says handled dollars drive the spend,   
but different activities matter end to end.  
So instead of one crude ratio for all,  
it builds the reserve from three cost calls.  

### Mango-Allen Dollar Based 

Mango-Allen keeps the structure in sight,  
but shifts the expected basis more right.  
Rather than only what paid claims show,  
it uses expected claims to help it grow.  

So if raw paid amounts distort the frame,  
a broader expected basis can steady the game.  
Still opening and maintenance share the role,  
with judgment needed to complete the whole.  

### Brian  

Brian counts the work instead of the dollars,  
for some claim tasks cost similar effort from handlers.  
Open it, manage it, pay it, close,  
reopen too — that’s how the workload flows.  

Project transaction counts still left to come,  
apply unit costs, then total the sum.  
It works when effort tracks claim events more cleanly  
than claim size alone can ever mean.  

### Wendy Johnson Technique  

Wendy Johnson trims the list down small,  
fewer transaction types can still tell all.  
Reporting and maintenance lead the way,  
to estimate future ULAE pay.  

Weighted counts become the measuring stick,  
historical cost per weighted count does the trick.  
Less detail than Brian, but same broad theme:  
count the work, not just the dollars seen.  

### Mango-Allen Claim Staffing  

Claim staffing asks a practical thing:  
how many people will future claims handling bring?  
Forecast staff needed for claims ahead,  
then multiply by cost per head.  

It is less about each claim’s exact route,  
and more about total department output.  
If future workload drives future staff,  
this method can give a sensible path.  

### Rahardjo  

Rahardjo says long-open claims consume  
more ULAE as they fill the room.  
A claim that lingers keeps drawing care,  
so duration open matters there.  

Not just count it once and call it done,  
maintenance cost grows while the claim still runs.  
So the reserve should reflect that staying power,  
not merely opening hour by hour.  

### Spalla   

Spalla watches time the workers spend,  
claim by claim from start to end.  
Use systems data to trace the load,  
where staff hours actually flowed.  

It is operational, close to real,  
showing how resources truly feel.  
Harder to build, but rich in sight,  
when time-tracking data is done right.  

### Triangle-Based  

Triangle ULAE builds a shape,  
though “actual by AY” may escape.  
So allocate calendar ULAE first,  
often by claims payment pattern rehearsed.  

Then develop the triangle toward the end,  
as if ULAE follows a development trend.  
Useful in theory, less common in play,  
because allocation assumptions can get in the way.  

### Comparison of ULAE Methods 

Classical follows unpaid dollars still due,  
Kittel adds opening effort to the view.  
Conger-Nolibos splits work into three,  
Mango-Allen tweaks the expected key.  

Count-based methods count tasks, not size,  
staffing counts people the future requires.  
Triangles force ULAE into AY form,  
but need allocation assumptions to weather the storm.
