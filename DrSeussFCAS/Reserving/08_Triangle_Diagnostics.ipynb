Learning Task B8: Use development triangles as diagnostic tools to identify changes and trends.

### Triangle Diagnostics

Triangles diagnose trend and shift,  
frequency rise or severity lift.  
Rate inadequacy, mix, reserve strength,  
settlement speed, and calendar length.  

Do not just build them, learn to read,  
for diagnostics come before the need  
to pick a method, factor, or tail —  
first ask why the patterns look frail.  

A triangle is not just built to reserve,  
it tells you how the claims system will swerve.  
Look by accident year, age, and date,  
to see if the pattern is early or late.  

If rows bend oddly or columns break stride,  
some change in the process may be inside.  
So before you project and call it done,  
ask what the triangle is telling you, one by one.  

Losses ask: are dollars changing?  
Counts ask: are claims rearranging?  
Severity asks: what’s the cost each claim?    
Ratios ask: is pricing still in game?  
 
Paid asks: how fast is cash going out?  
Reported asks: what do reserves say about  
the claims already known today?  
Put them together — that’s the actuary’s way.  



### LR  

Loss ratio rows can quietly speak,  
of pricing strength turning strong or weak.  
If mature years trend upward through time,  
rate adequacy may be missing the climb.  
  
But do not rush and call it rate,  
premium shifts can also change the state.  
On-level, trend, and mix must clear,  
before a loss ratio tells you what to fear.  

### Paid vs Reported 

Reported speaks earlier, paid speaks late,  
each sees the claim at a different state.  
If reported looks high but paid looks calm,  
case reserve changes may sound the alarm.  

If paid is slow but reported seems fine,  
settlement shifts may be crossing the line.  
Use both together, not one alone,  
for each reveals a different tone.  

### Case Adequacy 

Average case on open claims can tell,  
whether reserve strength is rising as well.  
If average case grows faster than paid severity trend,  
stronger case reserving may be the end.  
 
Then raw reported development may lie,  
and ultimate estimates drift too high.  
So triangles diagnose not just cost,  
but whether reserve judgment has been lost.  

### CY Effect  

When diagonals line up in one strange band,  
a calendar influence may be at hand.  
Court change, inflation, law, or staff,  
can tilt each year by the same path.  

Not row by row, but slanting through,  
calendar effects leave a diagonal clue.  
So if many accident years shift at once,  
a calendar force is the smarter hunch.  

### Business Mix  

A triangle may worsen, not from one claim trend,  
but from a changing book from end to end.  
More risky classes, higher limits too,  
can make old averages no longer true.  

If counts, severity, or loss ratios stray,  
ask whether the mix has changed the play.  
The triangle shows the symptom, not always the cause,  
so mix shifts matter before you pause.  

### Large Loss 

One giant claim can stain a row,  
and make development factors blow.  
If one year spikes beyond the rest,    
ask if a shock loss drove the test.  
      
Counts may not move, but dollars scream,  
so severity jumps and distorts the theme.   
A triangle used without this check  
may project one freak year through the deck.  


### Claims Made 

For claims-made, report timing leads,  
so report-year views may fit the needs.  
If reporting lag shifts through the file,  
the pattern can change by more than style.  

What looks like worse loss cost may be  
claims arriving sooner than expected to be.  
So diagnostics here must keep in mind,  
that reporting speed can reshape the line.
