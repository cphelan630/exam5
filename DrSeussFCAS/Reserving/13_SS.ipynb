Learning Task B13: Calculate and evaluate the estimation techniques for recoveries (e.g., salvage and
subrogation, reinsurance).

### S&S - Development Method  

Salvage and subro come back in time,  
after gross claims have started their climb.  
Build a triangle, reported or received,  
then develop to ultimate as patterns are believed.  

Fast salvage settles, property lines move quick,  
while subrogation may take years to stick.  
So ultimate recovery less what you’ve seen  
gives what is still unpaid or unreported between.  

### S&S - Ratio Method  

The ratio method ties recoveries near  
to gross ultimate claims for each accident year.  
First estimate gross claims, then pick  
an ultimate S&S ratio that seems to fit.  

It gives less leveraged factors than raw dollars alone,  
and lets judgment shape the ratio you own.  
But miss the gross losses, and soon you will see  
the S&S estimate miss too, naturally.  

### S&S - Data Caveat  

Some insurers track it, some net it away,  
some hold only payments by accident day.  
Some post it as negatives, hidden from view,  
so the first task is knowing what the data will do.  

Gross or net claims? Case estimates there?  
Separate by type, or all in one share?  
Before you reserve what recoveries mean,  
make sure you know how they entered the machine.
