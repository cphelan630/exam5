Learning Task B3: Describe the role of homogeneity and Credibility of data in the process of estimating unpaid claims. 

### Separate and Estimate

Oh, you mustn't mix the fast ones with the slow,  
The property with the GL in one great big row.  
For the auto pays out in a matter of weeks,  
But the liability claim? It takes years as it speaks.  

If you pile them together in one triangle bright,  
You'll get a blended LDF that isn't quite right.  
It's right for the mix that you had in the past,  
But what about now? Will that mix always last?  

No! The mix shifts! The GL grows and grows!  
And your blended-up factor just doesn't know.  
It still says one-sixty when it ought to say more,  
And your reserve is too low — you're behind on the score.  

So separate your data by type, by kind,  
Short tail on the left and long tail behind.  
Build one triangle for each homogeneous group,  
Then blend your estimates — that's the right loop.  

But wait! said the actuary. If I split things too fine,  
I'll have five claims per cell and that isn't divine.  
I need credibility — data with mass,  
Or my LDFs dance like a fast-shuffling class.  

So balance the homogeneity with size,  
Use benchmarks and blends to be actuarially wise.  
Separate enough to get meaningful flows,  
But keep enough volume for the pattern to show.


### Long Tail or Short?

Is it long tail? Is it short tail? Let's find out today,  
Does the claim get paid quickly or does it delay?  
Auto PD pays in weeks — that's your short-tail friend,  
But the GL excess claim? It goes on without end.  

A short-tail line has a tail near to one,  
When 24 months pass, it's essentially done.  
A long-tail line at 24 months? Barely begun!  
The CDF might be five — hold on to your gun.  

High frequency means: many claims in the pot,  
The law of large numbers says patterns you've got.  
The chain ladder works well when the counts are high,  
Each year's factor is stable — the reasons don't lie.  

But low frequency? One claim or maybe two?  
That single large payment makes LDFs askew.  
The chain ladder breaks down — the patterns aren't there,  
Use BF or expected loss — handle with care.  

Occurrence means: the accident is the trigger,  
Report it in five years? The liability's bigger.  
Claims-made means: report it this year or it's gone,  
Less IBNR — but the IBNER goes on.  

Tail coverage says: we'll extend your reporting date,  
For the claims-made policy about to terminate.  
It turns the claims-made into something like occurrence —  
The insurer takes back the full reserve endurance.  

So know your tails, long and short, high and low,  
And know your triggers — occurrence or reporting flow.  
The method you choose and the tail that you select  
All depend on which category applies to protect.


### Poem: "Occurrence and Claims-Made: Two Different Tails"

The **occurrence** policy covers what happened last year,
Even if the claim doesn't surface for five years — it's clear;
The trigger is the *event*, not the date of the claim,
So the IBNR tail stretches long in this game.

For occurrence lines the triangle spans wide:
Losses keep developing well past the right side,
The tail factor matters — it can be large and real,
For med mal and WC and asbestos appeal.

The **claims-made** policy has a different clock:
The claim must be *reported* within the policy's lock;
No matter when the injury happened in time,
If reported this year, it falls in this line.

The tail for claims-made is shorter and neater —
There's little IBNR after the first year's meter;
But watch for the **extended reporting period (ERP)**,
The "tail coverage" that extends the claims-made decree.

And watch the **retroactive date** — it cuts backward hard:
Claims from before that date are barred from the yard;
A new claims-made policy has a *retro date of today*,
So the first year is thin — prior acts kept away.

For reserving: occurrence needs development far,
Claims-made needs less — but ERP leaves a scar;
Know which trigger applies before building your view —
The tail factor and IBNR depend on the two.
