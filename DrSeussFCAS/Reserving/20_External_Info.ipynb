Learning Task B20: Utilize external information in a reserve analysis (e.g., inflation, tort reform, attorney
involvement).

When outside winds begin to blow,  
your triangles may fail to show  
the whole of what is coming through,  
because the world is changing too.  

Inflation lifts the severities high,  
old dollars fade as years go by.  
A legal shift, a court’s new tone,  
can make old patterns not your own.  

Tort reform cuts where claims may rise,  
less pain and suffering in the eyes  
of what a plaintiff once might win,  
so future ultimates may thin.    
  
Attorney use can slow the pace,  
add friction to the claims race,  
stretch settlements, raise costs to pay,  
and bend development on its way.  

So do not stare at triangles blind,  
as though the past is all you’ll find.  
Ask what external facts have turned  
since those old accident years were earned.  

Use bureau data, price indices,   
court decisions, social tendencies.  
Let management tell what they have seen,  
what shifted late, what laws now mean.  

Then choose your method with more care:  
development may lag unaware,  
while expected loss or freq-sev  
can build in changes more adeptly.  

Adjust the case, adjust IBNR,  
not paid that’s gone too near, too far.   
For what is fixed is fixed indeed;  
reserve the change where futures bleed.  
 
The actuary’s task is this:  
not just to crunch, not just to list,   
but use the world beyond the page  
to judge the liability’s age.  

So if the outside world has moved,  
don’t trust old patterns unimproved.  
A reserve is not just what trails show—  
it’s history plus what you now know.
