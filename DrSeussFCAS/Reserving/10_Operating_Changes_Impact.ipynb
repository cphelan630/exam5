### Poem: "Split the Count from the Cost"

One method says, "Do not let losses stay one,
Split count from severity before you are done."
Reported claim counts can develop one way,
While average cost per claim may travel another array.

Build one triangle for counts and another for size,
Then project each component with separate eyes;
Ultimate count times ultimate cost per claim,
Rebuilds the ultimate loss from a two-part frame.

When frequency and severity move out of step,
This method can show what a combined triangle kept;
It gives the actuary a sharper view,
Of whether more claims or bigger claims came through.


### Poem: "Develop What Is Still Open"

Case outstanding is what remains yet to pay,
The unpaid on open claims still standing today;
So one reserving method says, "Start right there,
Develop the open-case dollars with care."

Take paid losses plus case times a factor selected,
Add pure IBNR if claims are expected;
This method can help when open claims tell
A story of future payments reasonably well.

But if case adequacy shifted with time,
Developing raw case may not be sublime;
So know what the adjusters were booking back then,
Before you let case outstanding drive you again.
