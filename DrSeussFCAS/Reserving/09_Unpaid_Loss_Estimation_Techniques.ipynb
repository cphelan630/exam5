Learning Task B9: Calculate and evaluate unpaid loss estimation techniques (e.g., development/chain ladder,
case outstanding development, expected losses, Bornhuetter-Ferguson, Cape Cod,
frequency-severity, Berquist-Sherman, Benktander).

### ELR  

Expected Claims begins with what you know,  
not what the triangles seem to show.  
Before you trust immature years to speak,  
ask what ultimate should roughly peak.  

Take exposure times an expected rate,  
or premium times ELR to estimate.  
That gives the losses you’d expect to see,  
before experience has full credibility.  

No link ratios needed to lead the way,  
no heavy faith in what young years say.  
For the newest accident years on the page,  
this method is calm where development would rage.  
 
It works best when data is thin or green,  
or the claims process changed behind the scene.  
If history is noisy, sparse, or new,  
Expected Claims may be safer to pursue.  

But there is a price for being so still:  
it barely listens to the actual spill.  
If real experience is better or worse,  
the method won’t react with much force.  

So older years may fit it less,  
when reported data could tell you more, not guess.  
That is where development methods shine,  
because maturity gives a clearer sign.   
  
Expected Claims says: start with a prior view,  
what should this book of business roughly do?  
Then use that estimate as ultimate loss,  
a stable first anchor, though somewhat gross.  

It is simple, sturdy, often wise,  
especially when immature triangles disguise.  
But on a fully seasoned accident year,  
its weakness is clear: it may ignore what’s here.  

### ChainLadder   

Chain Ladder listens to what claims have done,  
and projects that pattern further on.  
Take each age to age link from year to year,  
select development factors that seem sincere.  

Build LDFs from now to ultimate end,  
then let each immature year extend.   
No outside prior, no expected loss theme,  
just history’s pattern pushed downstream.  
 
Reported Chain Ladder starts with paid plus case,  
reported losses already set the pace.  
Because case reserves give an early view,  
reported tends to stabilize sooner too.  

So for immature years it often speaks fast,  
with more information gathered from the past.  
But if case adequacy changed through time,  
reported factors may commit a crime.  
  
Stronger case reserves can make losses look high,   
and the method may overstate ultimate thereby.  
Weaker case strength can do the reverse,  
so changing adequacy can badly curse.  
 
Paid Chain Ladder starts from cash gone out,  
what the company actually paid out.  
No case reserve judgment clouds the frame,  
just actual payments in the triangle’s name.  

That makes paid data cleaner in one way,   
less exposed to reserve opinion day by day.   
But paid develops slower, especially long tail,  
so immature years can be thin and frail.  

If settlement rates changed, paid patterns bend,  
and old link ratios stop being your friend.  
Close claims faster, the triangle looks mature,  
close them slower, and it looks less sure.  
 
So both forms ask one central thing:  
will the past pattern still hold in spring?  
If yes, chain ladder can work just fine,  
if not, you need more judgment by design.    
 
Reported is quicker, but case-strength prone.  
Paid is cleaner, but timing alone  
can make it lag and mislead the view,  
if settlement behavior changed on you.  

Use chain ladder when the triangle is stable,  
when operational changes are not on the table.  
Then let development carry the rest,  
because mature history often knows best.  

### BF

Bornhuetter-Ferguson walks a middle line,  
between pure expectation and chain ladder’s design.  
It starts with an ultimate you choose in advance,  
then lets reported or paid give partial credence.  

The basic thought is steady and clear:  
don’t trust immature development too near.  
If only part of the losses have emerged to view,  
only that part should come from experience true.  
 
The rest stays anchored to what you expected before,  
so young accident years do not swing too far off shore.  
That is why BF is calmer than chain ladder’s claim,  
it does not force thin data to carry the game.  

Reported BF uses reported emergence by age,  
paid plus case reserve already on stage.  
If thirty percent is still unreported today,  
then thirty percent of expected ultimate stays away.  

So ultimate becomes, in thought and in sum:  
reported to date plus expected yet to come.  
Because reported emerges earlier, faster, more soon,  
reported BF reacts earlier than paid BF does in tune.  

But it still depends on case reserve quality’s role,  
since reported losses include case in the whole.  
If case adequacy shifted, beware what you see,  
for “reported to date” may not mean what it should be.  

Paid BF uses paid emergence instead,  
actual cash out, not case reserves instead.  
If sixty percent remains unpaid by now,  
then sixty percent of expected ultimate carries the plow.  

So paid BF says: take paid losses you know,  
then add the expected unpaid still to flow.  
It is usually slower, because paid lags behind,  
but it avoids case reserve judgment in kind.  
 
That makes it useful when case strength is weak,  
or when reserve adequacy makes reported less sleek.  
Still, if settlement timing changed through the years,  
paid emergence assumptions may bring their own fears.  

Both BF methods say one powerful thing:  
the immature years should not fully sing.  
Let emerged loss speak for the part that is shown,  
and let expected loss support what is unknown.  

Chain ladder says the triangle can tell,  
Expected Claims says trust your prior well.  
BF blends the two with a measured view:  
experience for the emerged part, expectation for the due.  

So choose reported BF when reported is sound,  
and claim emergence gets on the table sooner around.  
Choose paid BF when payments feel more clean,  
and case reserve noise muddies the scene.  

### Benktander  

Benktander blends with a lighter hand,  
between BF and chain ladder it takes its stand.  
Start with an expected ultimate in place,  
then let development add some experience grace.  
 
More responsive than BF, less jumpy than chain,  
it gives immature years a measured gain.  
Not all the way prior, not all the way past,  
a middle-road method built to last.  

### Cape Cod  

Cape Cod begins where BF would start,  
but earns its expected loss ratio from the data’s own heart.  
Not picked from outside by judgment alone,  
the ELR is implied from the experience shown.  

Take developed losses for older years,  
match them to earned exposure through trending gears.  
Then solve for the rate that best fits the book,  
using the data itself as the hook.  

For each accident year, use what has emerged,  
and keep the unreported part still expectation-merged.   
So like BF, it tempers immature years,  
but unlike BF, the prior comes from the triangle’s peers.  

It works best when the book is fairly stable,  
with exposures and trends you can place on the table.  
But if mix or adequacy changed a lot,  
the implied ELR may miss the spot.  

### Case OS Methods  

Case Outstanding comes in two main ways,  
both use case reserves to forecast future days.   
Each asks the same thing at the valuation date:  
how much more will known claims develop to ultimate fate?  

**Method One: Development on Case Incurred**  
Take reported losses, let link ratios be heard.  
Paid plus case sits in the triangle’s view,  
then develop to ultimate as chain ladder would do.  

It is simple and direct, and often can fit,  
when case reserves are stable and reasonably legit.  
But if adequacy changed from year to year,  
the reported pattern may not be clear.  

**Method Two: Average Case Outstanding**  
This one gets more granular all around.  
Look at case per open claim by age,  
that average carries the reserve-strength gauge.  

Estimate open counts still left to close,  
and multiply by average case as it grows.  
Then add paid losses already on the page,  
to build ultimate loss at each maturity age.  

This method shines when claim counts behave,  
and average case gives insight you can save.  
But it leans on count data being clean and sound,  
and on reserve patterns being meaningfully found.  

So one develops **reported losses as a whole**,  
the other uses **open counts and average case** in the role.  
Both are case-outstanding methods at heart,  
just different ways to let case reserves play their part.  

### Freqency Severity Techniqiues 

Frequency-Severity splits the loss in two:  
how often claims happen, and what they run through.  
Count the claims first, then price what they cost,  
so ultimate loss is not easily lost.  

**Method One: Reported Claims × Ultimate Severity**  
Start with reported counts in the year,  
then develop claim numbers till ultimate is clear.  
Next pick a severity expected at close,  
and multiply through where the final count goes.  

This method works best when counts are more known,  
but final dollars per claim are not fully shown.  
If reporting patterns are stable and sound,  
it gives you a clean way to build losses around.  

**Method Two: Ultimate Counts × Paid Severity-to-Date + Future Paid per Open Claim**  
Here counts still matter, but payments do too,  
split what is already paid from what’s left to come through.  
Take ultimate claim counts as one needed piece,  
then ask how much more each open claim will increase.  

So part of the loss is already on file,  
paid amounts observed up to now in the pile.  
The rest is projected from open claims still live,  
with future average payments helping them arrive.  

The first method leans harder on final severity,  
the second gives paid data more practical clarity.  
Both break the problem to counts and claim size,  
so the reserve is built from more basic supplies.  

### FS Technique 3 - DR 

Disposal Rate Frequency-Severity counts claims by close,  
not just how many report, but how many compose  
a settled path through the triangle’s frame,  
with closure patterns built into the game.  

Start with reported claims entering the file,  
then use disposal rates by age and by style.  
These tell what portion should already be done,  
and what open claims still have distance to run.  

From there, estimate ultimate claim count clear,  
then split closed and open at each development year.  
Closed claims give severity already known,  
open claims need future cost still to be shown.  

So unlike the simpler count-times-severity view,  
this method tracks claim **status** as it passes through.  
Not just how many claims will finally exist,  
but how many are settled and how many persist.  

It helps when closure timing matters a lot,  
because open and closed claims are not the same spot.  
An open claim’s future may still be wide,  
while a closed claim’s cost has mostly died.  

But the method needs counts that are stable and clean,  
and disposal patterns that actually mean  
the claims department behaved in a comparable way,  
not shifting its closure habits year by day.  

So disposal rate F-S says: follow the flow,  
through reported to closed, and let severity grow.  
Use claim counts plus status to sharpen the view,  
of what’s been settled and what’s still due.  


### Reported BS

Reported Berquist-Sherman, clean and bright,  
fixes the triangle before you trust its sight.  
If case reserves changed strength from year to year,  
raw reported losses may not be clear.  

Start with average case on claims still open,  
that’s where the adequacy clue is often spoken.  
Compare its rise to severity’s pace,  
to see if stronger reserving drove the case.  

Keep the latest diagonal in your hand,  
it reflects today’s adequacy stand.  
Then trend it backward, year by year,  
using severity trend to make old years peer.  
 
Not across maturities, that’s not the aim,  
but down each column at the same age frame.    
Twelve with twelve, and twenty-four with twenty-four,  
make them comparable to what they were before.  

Then rebuild reported losses from the ground:  
paid losses plus open counts times case re-found.  
Now the triangle speaks with less disguise,   
less polluted by reserve-strength shifts in size.  

If adequacy strengthened through the years,  
unadjusted reported development often oversteers.  
It reads high case reserves as worsening cost,  
and books too much ultimate that was never lost.  

So Reported B-S says: pause, repair,  
make the years consistent, then develop there.    
First fix case adequacy’s changing role,   
then let chain ladder estimate the whole.  

### Paid BS   

Paid Berquist-Sherman watches closing speed,  
for paid triangles bend when claim habits lead.  
If adjusters settle faster year by year,  
paid losses look more mature than they appear.  

If settlements slow, the reverse is true,  
the triangle looks younger than the losses due.  
So before you trust paid development’s song,  
ask whether settlement timing changed along.  

Start with closed counts, or disposal rate,  
to measure how fast claims leave the gate.  
Select a common pattern by maturity age,  
a steady settlement rhythm for the page.  

Then restate paid losses to that pace,  
as if each year had closed at the same rate base.   
You are not changing what ultimate should be,  
just removing timing distortion so the pattern is free.  

After that, the adjusted paid triangle can speak,  
with less noise from shifts in settlement technique.  
Then apply paid development as you normally would,  
now the signal is cleaner and the method more good.  

Reported B-S fixes changing case strength’s role,   
Paid B-S fixes changing closure control.  
One corrects adequacy, one corrects speed,  
both clean the data before you read.  

So if paid emergence changed because claims closed fast,  
don’t let raw factors chain that quirk to the past.  
Normalize settlement, rebuild what was paid,  
then develop the triangle you’ve honestly made.  

### Comparing Methods  

Reserving methods each try to say  
what unpaid claims will cost one day.  
Some trust patterns, some trust priors,  
some split counts, some clean old wires.  

**Chain Ladder** looks to what claims have done,  
and carries that pattern further on.  
For **reported**, case reserves help it see,  
so it matures faster, more speedily.  
For **paid**, it waits for cash to flow,  
cleaner perhaps, but slower to know.  
If case strength changed, reported can bend.  
If settlement changed, paid can mislead in the end.  

**Expected Claims** is calmer still,  
it starts with judgment, not triangle skill.  
Exposure times ELR, or some expected base,  
gives ultimate losses for the whole year’s case.  
It works when data is young or thin,  
but older years may want more experience let in.   

**Bornhuetter-Ferguson** stands in between,  
part prior belief, part what has been seen.  
With **reported BF**, emerged reported loss  
gets some weight before the full unknown is crossed.  
With **paid BF**, cash paid to date  
anchors the known, while the rest must wait.  
BF is less jumpy than ladder’s art,  
because only emerged claims play their part.  
 
**Cape Cod** is BF’s close kin,  
but gets its ELR from the data within.  
Not chosen outside by judgment alone,  
but implied by experience the book has shown.  
So BF brings a prior you mainly select,  
while Cape Cod lets the experience shape that expect.  

**Benktander** takes one step more,  
letting experience enter a bit stronger than before.  
Not as reactive as chain ladder’s climb,  
not as anchored as BF all the time.  
A middle road when years are green,  
but you still want some experience seen.  

**Frequency-Severity** breaks loss in two:   
how many claims, and what claims do.  
One form takes **reported counts to ultimate**, then  
multiplies by final severity again.   
Another splits **paid to date and future cost**,  
so open claims show what’s not yet lost.  
The **disposal-rate** version goes one stage deeper,  
tracking open and closed as a sharper keeper.  
These methods help when counts tell more truth  
than dollars alone in the triangle’s youth.  
 
**Case Outstanding methods** lean on case,  
the adjuster’s view of each open case.  
One develops **reported losses as a whole**,   
letting paid-plus-case play the role.  
The other uses **open counts times average case**,  
a more granular, claim-status-based place.  
They help when case reserves are meaningful and sound,    
but suffer if adequacy shifts around.  

That is why **Berquist-Sherman** enters the scene,  
when the triangle is dirty, not truly clean.  
**Reported B-S** fixes changing case strength first,  
so stronger reserves do not make ultimates burst.  
**Paid B-S** fixes changing settlement pace,   
so payment timing does not distort the case.  
B-S is not just another reserve projection line;  
it cleans the data before standard methods opine.  

So here is the contrast, stated plain:  
**Chain Ladder** trusts patterns in the chain.  
**Expected Claims** trusts an outside view.  
**BF** blends prior with what came through.  
**Cape Cod** lets the data imply the prior.  
**Benktander** lets experience climb a little higher.  
**Freq-Sev** splits claims into count and cost.  
**Case methods** use reserve views not yet lost.  
**Berquist-Sherman** repairs the past,   
so whichever method can be applied at last.  

No single method wears the crown,
for every line, each year, all around.  
Stable data may let chain ladder lead,  
immature years may make BF what you need.  
Count-sensitive books may want F-S’s eye,  
operational shifts may make B-S apply.  
Good reserving is not just plugging and play,  
but asking what the data can honestly say.
