Learning Task B16:  Evaluate the results of a reserve analysis for adequacy and reasonableness using loss ratios, severities, pure premiums, frequencies, indicated unpaid losses, etc..

A reserve is not a number alone,  
not just a pick the actuary owns.  
You test its shape, you test its sound,  
by how it fits the business ground.  

Look first at loss ratios through the years:  
do they align, or raise new fears?  
If one jumps high with no clear cause,  
ask what has shifted in claims or laws.  

Then severities—claim cost per file—  
have they been climbing all the while?  
A higher reserve may make good sense  
if average claim costs grew intense.  

And frequency too must earn its place:  
more claims can change the reserve’s face.  
If counts are flat but dollars soar,  
the story’s severity, not something more.  

Pure premium joins the two as one,  
loss per exposure, neatly done.  
A useful check on overall need,  
to see if the reserve fits the book indeed.  

Then indicated unpaid losses say,  
“What reserve does each method weigh?”  
If methods differ far and wide,  
don’t simply pick a middle and hide.  

Ask why chain ladder leans too high,  
why BF lets the number lie  
below the rest—perhaps because  
the recent data broke old laws.  

A reasonable reserve should live  
where diagnostics also give  
a story that is clear and true,  
not one that only one method knew.  

So test by ratio, count, and claim,  
by whether trends all tell one name.  
Adequate, reasonable—that’s the art:  
the selected reserve must fit every part.  

If metrics clash, then slow your hand.  
Find what changed in the underlying land.  
For a reserve is only sound  
when all the evidence gathers round.
